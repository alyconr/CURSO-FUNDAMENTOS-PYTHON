1: Comparación y Análisis de Listas de Estudiantes Imagina que se tiene un conjunto con los números de identificación de los estudiantes de dos cursos. Se debe:

Determinar si ambos conjuntos tienen estudiantes en común.

Verificar si el conjunto de un curso está completamente contenido en el conjunto del otro.

Mostrar el conjunto de estudiantes que aparece en ambos cursos (común) y también aquellos que están solo en uno de ellos (diferencia y diferencia simétrica). Ejemplo de salida esperado:

```python
curso1 = {102, 403, 504, 104}
curso2 = {403, 998, 283, 504}

print(f"Ambos cursos tienes estudiantes en comun: {curso1.issubset(curso2)}")
print(f"Un curso esta contenido en otro: {curso1.issubset(curso2)}")
print(f"Estudiantes en comun: {curso1.intersection(curso2)}")
print(f"Estudiantes en curso1 pero no en curso2: {curso1.difference(curso2)}")
print(f"Estudiantes en curso2 pero no en curso1: {curso2.difference(curso1)}")
```

2: Gestión de Inventarios en Almacenes Se dispone de dos conjuntos que representan los códigos de productos en dos almacenes diferentes. El desafío consiste en:

Mostrar los productos que se encuentran en ambos almacenes.

Determinar cuáles productos están únicamente en uno de ellos.

Unificar los productos disponibles en ambos almacenes en un solo conjunto.

Finalmente, eliminar (descartar) un producto que se ha dado de baja de uno de los almacenes.

```python
Almacen1 = {"zapatos", "blusas", "chaquetas", "pantalones"}
Almacen2 = {"blusas", "bufandas", "tenis"}

print(f"Productos en ambos almacenes: {Almacen1.intersection(Almacen2)}")
print(f"Productos en Almacen1 pero no en Almacen2: {Almacen1.difference(Almacen2)}")
print(f"Productos en Almacen2 pero no en Almacen1: {Almacen2.difference(Almacen1)}")
print(f"Unificar productos en ambos almacenes: {Almacen1.union(Almacen2)}")
Almacen1.discard("pantalones")
print(f"Almacen1 actualizado tras eliminar 'pantalones': {Almacen1}")
```

3: Análisis de Preferencias de Hobbies Se tiene un conjunto con los hobbies de un grupo de amigos y otro conjunto con los hobbies de otro grupo. Se solicita:

Identificar qué hobbies comparten ambos grupos.

Conocer cuáles son exclusivos de cada grupo (usando la diferencia y la diferencia simétrica).

Comprobar si los grupos no tienen ningún hobby en común (verificar disjoint).

Además, eliminar un hobby específico de uno de los conjuntos para actualizar las preferencias. Ejemplo de salida esperado:

```python
Hobbies = {'leer', 'correr', 'pintar'}
Hobbies2 = {'correr', 'viajar', 'cocinar'}

print(f"Hobbies comunes: {Hobbies.intersection(Hobbies2)}")
print(f"Hobbies únicos en el Grupo A: {Hobbies.difference(Hobbies2)}")
print(f"Hobbies únicos en el Grupo B: {Hobbies2.difference(Hobbies)}")
print(f"los grupos no tiene elementos en comun?: {Hobbies.isdisjoint(Hobbies2)}")
Hobbies.discard('pintar')
print(f"Grupo A actualizado tras eliminar 'pintar': {Hobbies}")
```
