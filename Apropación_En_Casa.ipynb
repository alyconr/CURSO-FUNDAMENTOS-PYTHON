Procesamiento de Números del 1 al 20 Se requiere desarrollar un programa que trabaje con los números comprendidos entre el 1 y el 20, de forma que se identifiquen y muestren aquellos que sean divisibles por dos, acompañados de su posición en la secuencia. Además, el sistema debe recorrer la colección para determinar cuántos números no cumplen esa propiedad y calcular la suma de los que sí la cumplen. El objetivo es diseñar la lógica necesaria para examinar y procesar los números según las condiciones indicadas.

```python
# cree la lista hasta el "21" porque nos va a enviar hasta un numero antes y necesitamos hasta el 20.
numeros = list(range(1, 21))


divisibles_por_dos = []
no_divisibles_contador = 0
suma_divisibles = 0

enum_numeros = list(enumerate(numeros, start=1))
for posicion, numero in enum_numeros:
    if numero % 2 == 0:
        divisibles_por_dos.append((posicion, numero))
        suma_divisibles += numero
    else:
        no_divisibles_contador += 1

print("Números divisibles por 2 con su posición:")
for posicion, numero in divisibles_por_dos:
    print(f"Posición {posicion}: {numero}")

print(f"Cantidad de números no divisibles por 2: {no_divisibles_contador}")
print(f"Suma de los números divisibles por 2: {suma_divisibles}")
```

Administración de Actividades Diarias Imagina una aplicación para gestionar las actividades cotidianas de una persona, donde se disponga de una serie de tareas, algunas de las cuales indican que ya han sido concluidas. El desafío consiste en presentar cada actividad junto a su posición en la colección, filtrar aquellas que contengan una marca de finalización y actualizar el conjunto de actividades. Finalmente, se debe exhibir el estado del conjunto antes y después de aplicar el filtro, permitiendo al usuario observar los cambios realizados.

```python
 #poner las actividades
actividades = [
    "Comprar víveres",
    "Hacer ejercicio - SI",
    "Llamar al banco",
    "Enviar correo - SI",
    "Estudiar programación",
    "Preparar la cena - SI"
]


print("Lista original de actividades:")
for idx, actividad in enumerate(actividades, start=1):
    print(f"{idx}. {actividad}")

actividades_pendientes = [act for act in actividades if "SI" not in act]


print("\nLista de actividades después de eliminar las completadas:")
for idx, actividad in enumerate(actividades_pendientes, start=1):
    print(f"{idx}. {actividad}")
```

Filtrado y Transformación de Texto Propón un sistema en el que se solicite al usuario una secuencia de caracteres. A partir de la entrada, se debe descomponer el texto en sus componentes individuales y extraer únicamente aquellos que correspondan a letras del alfabeto, omitiendo números y símbolos. Una vez realizado este filtrado, la colección resultante deberá ser convertida en una estructura que no permita modificaciones, y el resultado final se debe mostrar al usuario.

```python
def filtrar_y_transformar_texto():

  texto_ingresado = input("Ingresa una secuencia de caracteres: ")
  letras = [letra for letra in texto_ingresado if letra.isalpha()]
  tupla_letras = tuple(letras)
  print(f"Las letras filtradas son: {tupla_letras}")

filtrar_y_transformar_texto()
```

Posteriormente, el sistema debe permitir al usuario consultar la nota de un alumno específico introduciendo la posición correspondiente, finalizando la consulta si se ingresa un valor fuera del rango válido. La solución deberá vincular ambas colecciones de manera sincronizada y ofrecer una interacción dinámica.

```python
 nombres_estudiantes = ["Laura","Camilo","Mauricio","Luicia", "Pablito","Samanta"]
 notas_estudiantes = ["80","90","100","60","20","55"]

 for i in range(min(len(nombres_estudiantes), len(notas_estudiantes))):
  print(i, nombres_estudiantes[i] + " = " + notas_estudiantes[i])

  while True:
    posicion = int(input("Ingrese el número de posicion del alumno: "))

    if posicion < 0 or posicion >= len(nombres_estudiantes):
          print("Se finaliza el programa...")
    break

  print(f"El alumno con posicion {posicion} es {nombres_estudiantes[posicion]} con nota {notas_estudiantes[posicion]}")
```
