Pregunte al usuario cuántos elementos desea ingresar en una lista, luego solicite cada uno de ellos y presente el contenido de la lista y su contenido invertido.

```python
numero_a_lis = int(input("Cantidad de numeros que quieres en la lista"))
numeros=[]
for n in range(numero_a_lis):
    print(f"\nNumero {n + 1}:")
    numero = int(input("Ingrese el numero"))
    numeros.append(numero)
    print(numeros)
    numeros.reverse()
    print(numeros)

#for lista in numeros:
    #print(lista)
```

2. Crear una lista de números y solicitar un número. Borrar de la lista ese número en todas las posiciones donde se encuentre. Presentar la lista final

```python
numeros=[1,2,3,4,5,6,9,5,1,2,5,5,5,]
elimina = int(input('numero que quiere borrar'))
while elimina in numeros:
    numeros.remove(elimina)

print(numeros)
```

3. Solicite al usuario dos(2) frases y devuelva una lista con todas las letras que se repiten en la misma posición de ambas listas

```python
frase1 = input("Ingrese una frase")
frase2 = input("Ingrese otra frase")
lista3=[]
for i in range(len(frase1)):
    if frase1[i]==frase2[i]:
        lista3.append(frase1[i])
print(lista3)
```
