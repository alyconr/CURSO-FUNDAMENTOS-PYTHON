Ejercicio 1

```python
inventario = {
    "Manzana" : 15,
    "Pera" : 22,
    "Coco" : 36,
    "Mandarina" : 7,
    "Mandarina" : 5,
}
umbral = 10

print("Inventario actual: ")

for producto in inventario:
    cantidad = inventario[producto]
    if cantidad < umbral:
        print(producto, cantidad, "unidades, stcok bajo")
    else:
        print(producto, cantidad, "unidades")

actualizar = input("Ingrese el nombre del producto a actualizar: ")

if actualizar in inventario:
    print("Actualziando stock de", actualizar)
    print("Antes", inventario[actualizar], "unidades")

    nueva_cantidad = int(input("Ingrese la cantidad: "))

    inventario[actualizar] = nueva_cantidad

    print("Despues", inventario[actualizar], "unidades")
else:
    print("Producto no encontrado en el inventario")


print("Inventario actualizado: ")
for producto in inventario:
    cantidad = inventario[producto]
    if cantidad < umbral:
        print(producto, cantidad, "unidades, stock bajo")
    else:
        print(producto, cantidad, "unidades")
```
