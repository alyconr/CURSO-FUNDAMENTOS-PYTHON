1. Gestión de Inventario de Productos
Descripción:
Se debe desarrollar un sistema que administre un inventario compuesto por productos y sus cantidades. El programa presentará el listado inicial y resaltará aquellos artículos cuyo stock esté por debajo de un umbral definido. Luego, permitirá actualizar la cantidad de un producto específico, mostrando el inventario antes y después de la modificación.

```python
import random

def adivina_el_numero():
    numero_secreto = random.randint(1, 100)
    intentos = 0
    print("¡Bienvenido al juego de Adivina el Número Secreto!")
    print("Estoy pensando en un número del 1 al 100. ¿Puedes adivinar cuál es?")

    while True:
        try:
            intento = int(input("Introduce tu intento: "))
            intentos += 1

            if intento < numero_secreto:
                print("Demasiado bajo. Intenta con un número más alto.")
            elif intento > numero_secreto:
                print("Demasiado alto. Intenta con un número más bajo.")
            else:
                print(f"¡Felicidades! Adivinaste el número en {intentos} intentos.")
                break
        except ValueError:
            print("Por favor, introduce un número válido.")


adivina_el_numero()
```

2. Búsqueda de Palabras en un Texto
Descripción:
Crea una aplicación que solicite al usuario un párrafo y, a continuación, le permita buscar una palabra específica. El sistema debe contar cuántas veces aparece dicha palabra en el texto, mostrar las posiciones donde se encuentra (sin distinguir mayúsculas de minúsculas) y comunicar el resultado al usuario.



```python
def buscar_palabra_en_texto():
    texto = input("Ingresa un párrafo de texto:\n")
    palabra_buscada = input("Ingresa la palabra que deseas buscar:\n")

    texto_lower = texto.lower()
    palabra_lower = palabra_buscada.lower()

    posiciones = []
    inicio = 0

    while True:
        index = texto_lower.find(palabra_lower, inicio)
        if index == -1:
            break
        posiciones.append(index)
        inicio = index + len(palabra_lower)

    cantidad = len(posiciones)

    if cantidad > 0:
        print(f"\nLa palabra '{palabra_buscada}' aparece {cantidad} veces.")
        print(f"Posiciones en el texto: {posiciones}")
    else:
        print(f"\nLa palabra '{palabra_buscada}' no se encontró en el texto.")


buscar_palabra_en_texto()
```

3. Conversión y Limpieza de Texto
Descripción:
Diseña una solución que pida al usuario una frase, la divida en palabras y elimine cualquier carácter no alfabético de cada una. Una vez procesada la información, la estructura resultante se transformará en una forma inmutable, y se presentará el contenido final al usuario de manera ordenada.

```python
import re

def limpiar_y_convertir_texto():
    frase = input("Ingresa una frase:\n")


    palabras = frase.split()

    
    palabras_limpias = [re.sub(r'[^a-zA-ZáéíóúÁÉÍÓÚñÑ]', '', palabra) for palabra in palabras]

    
    palabras_filtradas = sorted([p for p in palabras_limpias if p])

    
    resultado_inmutable = tuple(palabras_filtradas)
    print("\nPalabras procesadas (ordenadas y sin caracteres no alfabéticos):")
    print(resultado_inmutable)

limpiar_y_convertir_texto()
```

5. Calculadora de Estadísticas de Números
Descripción:
Desarrolla una aplicación que trabaje con una colección de números predefinidos. El sistema debe ofrecer al usuario un menú interactivo con opciones para calcular el promedio, identificar el valor máximo y mínimo, o eliminar un número. Tras cada operación, se mostrará tanto el resultado como el estado actual de la colección.

```python
def mostrar_menu():
    print("\n--- MENÚ ---")
    print("1. Calcular promedio")
    print("2. Mostrar valor máximo")
    print("3. Mostrar valor mínimo")
    print("4. Eliminar un número")
    print("5. Mostrar colección actual")
    print("6. Salir")

def calcular_promedio(numeros):
    if numeros:
        promedio = sum(numeros) / len(numeros)
        print(f" Promedio: {promedio:.2f}")
    else:
        print("⚠ La colección está vacía.")

def mostrar_maximo(numeros):
    if numeros:
        print(f" Valor máximo: {max(numeros)}")
    else:
        print("⚠ La colección está vacía.")

def mostrar_minimo(numeros):
    if numeros:
        print(f" Valor mínimo: {min(numeros)}")
    else:
        print("⚠ La colección está vacía.")

def eliminar_numero(numeros):
    try:
        num = float(input("Ingresa el número que deseas eliminar: "))
        if num in numeros:
            numeros.remove(num)
            print(f" Número {num} eliminado.")
        else:
            print(f" El número {num} no está en la colección.")
    except ValueError:
        print(" Entrada inválida. Ingresa un número.")

def mostrar_coleccion(numeros):
    print(f"Colección actual: {numeros if numeros else 'vacía'}")

def calculadora_estadisticas():
    numeros = [10, 20, 30, 40, 50]  # colección predefinida
    print(" Bienvenido a la Calculadora de Estadísticas")
    mostrar_coleccion(numeros)

    while True:
        mostrar_menu()
        opcion = input("Selecciona una opción (1-6): ").strip()

        if opcion == "1":
            calcular_promedio(numeros)
        elif opcion == "2":
            mostrar_maximo(numeros)
        elif opcion == "3":
            mostrar_minimo(numeros)
        elif opcion == "4":
            eliminar_numero(numeros)
        elif opcion == "5":
            mostrar_coleccion(numeros)
        elif opcion == "6":
            print("👋 Saliendo de la aplicación. ¡Hasta pronto!")
            break
        else:
            print(" Opción no válida. Intenta de nuevo.")

calculadora_estadisticas()
```
