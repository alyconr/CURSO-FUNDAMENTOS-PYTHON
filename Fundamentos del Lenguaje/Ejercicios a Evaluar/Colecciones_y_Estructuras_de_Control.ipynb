1. Gestión de Inventario de Productos
Descripción:
Se debe desarrollar un sistema que administre un inventario compuesto por productos y sus cantidades. El programa presentará el listado inicial y resaltará aquellos artículos cuyo stock esté por debajo de un umbral definido. Luego, permitirá actualizar la cantidad de un producto específico, mostrando el inventario antes y después de la modificación.

```python
inventory_list = ["Manzana", "Pollo", "Cerdo", "Pescado"]
stock_list = [30, 40, 27, 3]

for x in range(min(len(inventory_list), len(stock_list))):
    print(f"{x + 1}. {inventory_list[x]} - {stock_list[x]}")
print("Productos por debajo del stock(15): ")

for x in stock_list[:]:
    if x < 15:
        print(f"{inventory_list[x]} - {stock_list[x]}")  # Este print tiene un error

update_product = int(input("Ingrese el producto a actualizar(1,2,3,4): ")) - 1
if 0 <= update_product < len(inventory_list):
    print(f"{inventory_list[update_product]} - {stock_list[update_product]}")
update_quantity = int(input(f"Ingrese la cantidad a actualizar del producto {inventory_list[update_product]}: "))
stock_list[update_product] = update_quantity
print(f"{inventory_list[update_product]} - {stock_list[update_product]}")
print(f"El inventario luego del cambio quedo de la siguiente manera:")
for x in range(min(len(inventory_list), len(stock_list))):
    print(f"{x + 1}. {inventory_list[x]} - {stock_list[x]}")
```

2. Búsqueda de Palabras en un Texto
Descripción:
Crea una aplicación que solicite al usuario un párrafo y, a continuación, le permita buscar una palabra específica. El sistema debe contar cuántas veces aparece dicha palabra en el texto, mostrar las posiciones donde se encuentra (sin distinguir mayúsculas de minúsculas) y comunicar el resultado al usuario.



```python
paragraph = input("Ingresa un párrafo: ")
word = input("Ingresa la palabra a buscar: ")

word1 = word.lower()
paragraph1 = paragraph.lower()
word_found = paragraph1.count(word1)
print(f"Cantidad de palabras encontradas: {word_found}")

for x in range(len(paragraph1)):
    if paragraph1[x:x+len(word1)] == word1:
        print(f"La palabra '{word}' se encuentra en la posición {x}")
```

3. Conversión y Limpieza de Texto
Descripción:
Diseña una solución que pida al usuario una frase, la divida en palabras y elimine cualquier carácter no alfabético de cada una. Una vez procesada la información, la estructura resultante se transformará en una forma inmutable, y se presentará el contenido final al usuario de manera ordenada.

```python
sentence = input("Ingresa una frase: ")

words = sentence.split()

sorted_words = []

for word in words:
    sorted_word = ""
    for character in word:
        if character.isalpha():
            sorted_word += character
    sorted_words.append(sorted_word)

print("Se ordenaron las palabras", sorted_words)

words_tuple = tuple(sorted_words)
print("Se creó la tupla", words_tuple)
```

4. Registro y Consulta de Calificaciones de Exámenes
Descripción:
Implementa un sistema que gestione un registro de estudiantes y sus calificaciones. El programa debe permitir agregar nuevos registros, listar todos los existentes y consultar la calificación de un estudiante introduciendo su nombre. Se espera que la información se actualice y se presente de forma clara en cada paso.

```python
students = ["Brandon", "Juan", "Camilo", "Sergio"]
grades = [100, 91, 86, 24]

for x in range(min(len(students), len(grades))):
    print(f"{x + 1}. {students[x]} - {grades[x]}")

save_student = input("¿Desea guardar un nuevo registro de estudiante(Si/No)?")

if save_student == "Si":
    name = input("Ingrese el nombre del estudiante ")
    students.append(name)
    grade = int(input("Ingrese la calificacion del estudiante"))
    grades.append(grade)

    for x in range(min(len(students), len(grades))):
        print(f"{x + 1}. {students[x]} - {grades[x]}")

elif save_student == "No":
    print("Ok, sigamos")

else:
    print("Opción no válida. Por favor escriba 'Si' o 'No'.")

search_student = input("Ingresa el estudiante a consultar: ")
if search_student in students:
    index = students.index(search_student)
    print(f"El estudiante que consultaste fue: {students[index]} - {grades[index]}")
else:
    print("El estudiante no se encuentra en el registro.")
```

5. Calculadora de Estadísticas de Números
Descripción:
Desarrolla una aplicación que trabaje con una colección de números predefinidos. El sistema debe ofrecer al usuario un menú interactivo con opciones para calcular el promedio, identificar el valor máximo y mínimo, o eliminar un número. Tras cada operación, se mostrará tanto el resultado como el estado actual de la colección.

```python
fixed_values = [5, 25, 10, 55, 5, 80, 10]
print(fixed_values)

while True:

    print("\nMenú de opciones:")
    print("1. Calcular el promedio")
    print("2. Encontrar el número máximo")
    print("3. Encontrar el número mínimo")
    print("4. Salir")
    
    option = input("Ingrese una opción: ")
    
    if option == "1":
        total_sum = 0  
        count = 0  
        for value in fixed_values:
            total_sum += value
            count += 1
        if count > 0:
            average = total_sum / count
            print(f"El promedio es: {average}")
        
    elif option == "2":
        maximum = fixed_values[0]  
        for value in fixed_values:
            if value > maximum:
                maximum = value  
        print(f"El número máximo es: {maximum}")
            
    elif option == "3":
        minimum = fixed_values[0]  
        for value in fixed_values:
            if value < minimum:
                minimum = value  
        print(f"El número mínimo es: {minimum}")
        
    elif option == "4":
        print("Saliendo del programa")
        break
    else:
        print("Opción no válida.")

```
