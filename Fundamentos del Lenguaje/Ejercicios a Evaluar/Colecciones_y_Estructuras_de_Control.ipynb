1. Gestión de Inventario de Productos
Descripción:
Se debe desarrollar un sistema que administre un inventario compuesto por productos y sus cantidades. El programa presentará el listado inicial y resaltará aquellos artículos cuyo stock esté por debajo de un umbral definido. Luego, permitirá actualizar la cantidad de un producto específico, mostrando el inventario antes y después de la modificación.

```python
productos = ["Manzanas", "Naranjas", "Piñas"]
cantidades = [50, 25, 10]

print("Inventario inicial:")
for i in range(len(productos)):
    print(f"{i + 1}. {productos[i].upper()} - {cantidades[i]} UNIDADES")

stock_minimo = 15
print("\nPRODUCTO POR DEBAJO DEL STOCK:")
for i in range(len(cantidades)):
    if cantidades[i] < stock_minimo:
        print(f"- {productos[i].upper()} ({cantidades[i]} UNIDADES)")

indice_actualizar = int(input("\nINGRESE EL PRODUCTO A ACTUALIZAR (número): ")) - 1
cantidad_actualizar = int(input("INGRESE LA CANTIDAD A ACTUALIZAR: "))

cantidades[indice_actualizar] += cantidad_actualizar

print("\nInventario actualizado:")
for i in range(len(productos)):
    print(f"{i + 1}. {productos[i].upper()} - {cantidades[i]} UNIDADES")
```

2. Búsqueda de Palabras en un Texto
Descripción:
Crea una aplicación que solicite al usuario un párrafo y, a continuación, le permita buscar una palabra específica. El sistema debe contar cuántas veces aparece dicha palabra en el texto, mostrar las posiciones donde se encuentra (sin distinguir mayúsculas de minúsculas) y comunicar el resultado al usuario.



```python
parrafo = input("Ingrese un párrafo: ")

palabra_buscada = input("Ingrese la palabra a buscar: ")

parrafo_minuscula = parrafo.lower()
palabra_buscada = palabra_buscada.lower()

palabras = parrafo_minuscula.split()

contador = 0
posiciones = []

for i in range(len(palabras)):
    
    palabra_limpia = palabras[i].strip(".,;:!¿?()[]{}\"'")

    if palabra_limpia == palabra_buscada:
        contador += 1
        posiciones.append(i + 1) 

print(f"\nLa palabra '{palabra_buscada}' aparece {contador} veces.")
if contador > 0:
    print("Posiciones:", posiciones)
else:
    print("La palabra no fue encontrada en el texto.")
```

4. Registro y Consulta de Calificaciones de Exámenes
Descripción:
Implementa un sistema que gestione un registro de estudiantes y sus calificaciones. El programa debe permitir agregar nuevos registros, listar todos los existentes y consultar la calificación de un estudiante introduciendo su nombre. Se espera que la información se actualice y se presente de forma clara en cada paso.

```python
estudiantes = {
    "ANA": 90,
    "LUIS": 50,
    "MARLON": 20
}
print("REGISTRO INICIAL DE ESTUDIANTES")
for nombre in estudiantes:
    print(f"- {nombre.capitalize()}: {estudiantes[nombre]}")

respuesta = input("\n¿DESEA GUARDAR UN NUEVO REGISTRO DE ESTUDIANTES? ").upper()
if respuesta == "SI":
    nuevo_nombre = input("\nINGRESE NOMBRE DE ESTUDIANTE: ").upper()
    nueva_nota = int(input(f"INGRESE LA CALIFICACIÓN DE {nuevo_nombre.capitalize()}: "))

    estudiantes[nuevo_nombre] = nueva_nota

    print("\nREGISTRO ACTUALIZADO:")
    for nombre in estudiantes:
        print(f"- {nombre.capitalize()}: {estudiantes[nombre]}")

consulta = input("\nINGRESE EL NOMBRE DEL ESTUDIANTE A CONSULTAR: ").upper()

if consulta in estudiantes:
    print(f"\nLA CALIFICACIÓN DE {consulta.capitalize()} ES: {estudiantes[consulta]}")
else:
    print(f"\n{consulta.capitalize()} NO SE ENCUENTRA EN EL REGISTRO.")
```

5. Calculadora de Estadísticas de Números
Descripción:
Desarrolla una aplicación que trabaje con una colección de números predefinidos. El sistema debe ofrecer al usuario un menú interactivo con opciones para calcular el promedio, identificar el valor máximo y mínimo, o eliminar un número. Tras cada operación, se mostrará tanto el resultado como el estado actual de la colección.

```python
nums = [12, 34, 7, 90, 23, 11]

while True:
    print("\nNúmeros:", nums)
    print("1.Promedio  2.Máximo  3.Mínimo  4.Eliminar  5.Salir")
    op = int(input("Opción: "))

    if op == 1: print("Promedio:", sum(nums)/len(nums))
    elif op == 2: print("Máximo:", max(nums))
    elif op == 3: print("Mínimo:", min(nums))
    elif op == 4:
        n = int(input("Eliminar número: "))
        if n in nums: nums.remove(n); print("Eliminado.")
        else: print("No está.")
    elif op == 5:
        print("Adiós."); break
    else: print("Opción inválida.")
```
