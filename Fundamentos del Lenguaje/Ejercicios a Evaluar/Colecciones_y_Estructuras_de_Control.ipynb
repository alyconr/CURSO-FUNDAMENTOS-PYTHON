1. Gestión de Inventario de Productos
Descripción:
Se debe desarrollar un sistema que administre un inventario compuesto por productos y sus cantidades. El programa presentará el listado inicial y resaltará aquellos artículos cuyo stock esté por debajo de un umbral definido. Luego, permitirá actualizar la cantidad de un producto específico, mostrando el inventario antes y después de la modificación.

```python
inventario = {
    "Manzanas": 15,
    "Bananas": 5,
    "Naranjas": 2,
    "Peras": 8,
    "Uvas": 12
}

umbral_bajo_stock = 6

def mostrar_inventario(inventario, umbral=None):
    print("\nInventario actual:")
    for producto, cantidad in inventario.items():
        if umbral is not None and cantidad < umbral:
            print(f" {producto}: {cantidad} (Stock bajo)")
        else:
            print(f"{producto}: {cantidad}")

def actualizar_stock(inventario):
    producto = input("\nIngrese el nombre del producto a actualizar: ").strip().capitalize()
    
    if producto not in inventario:
        print(" Producto no encontrado en el inventario.")
        return

    try:
        nueva_cantidad = int(input(f"Ingrese la nueva cantidad para '{producto}': "))
        print("\nInventario ANTES de la actualización:")
        mostrar_inventario(inventario)

        inventario[producto] = nueva_cantidad

        print("\n Actualización exitosa.")
        print("\nInventario DESPUÉS de la actualización:")
        mostrar_inventario(inventario, umbral=umbral_bajo_stock)
    except ValueError:
        print(" Cantidad no válida. Debe ser un número entero.")

mostrar_inventario(inventario, umbral=umbral_bajo_stock)

actualizar_stock(inventario)
```

2. Búsqueda de Palabras en un Texto
Descripción:
Crea una aplicación que solicite al usuario un párrafo y, a continuación, le permita buscar una palabra específica. El sistema debe contar cuántas veces aparece dicha palabra en el texto, mostrar las posiciones donde se encuentra (sin distinguir mayúsculas de minúsculas) y comunicar el resultado al usuario.



```python
def buscar_palabra_en_texto(texto, palabra_objetivo):
    palabras = texto.lower().split()
    palabra_objetivo = palabra_objetivo.lower()

    posiciones = [i for i, palabra in enumerate(palabras) if palabra.strip(".,;!?") == palabra_objetivo]
    cantidad = len(posiciones)

    return cantidad, posiciones

texto_usuario = input("Ingrese un párrafo de texto:\n")

palabra_busqueda = input("\nIngrese la palabra que desea buscar: ")

cantidad, posiciones = buscar_palabra_en_texto(texto_usuario, palabra_busqueda)

print(f"\n Resultados de búsqueda para: '{palabra_busqueda}'")
print(f"Cantidad de apariciones: {cantidad}")

if cantidad > 0:
    print(f"Posiciones (índice de palabra): {posiciones}")
else:
    print("La palabra no se encontró en el texto.")
```

3. Conversión y Limpieza de Texto
Descripción:
Diseña una solución que pida al usuario una frase, la divida en palabras y elimine cualquier carácter no alfabético de cada una. Una vez procesada la información, la estructura resultante se transformará en una forma inmutable, y se presentará el contenido final al usuario de manera ordenada.

```python
import re

def limpiar_y_convertir(frase):
    palabras_limpias = [re.sub(r'[^a-zA-ZáéíóúÁÉÍÓÚñÑ]', '', palabra) for palabra in frase.split()]
    
    palabras_limpias = [palabra for palabra in palabras_limpias if palabra]

    resultado = tuple(palabras_limpias)

    resultado_ordenado = tuple(sorted(resultado, key=lambda x: x.lower()))

    return resultado_ordenado

frase = input("Ingrese una frase: ")
resultado_final = limpiar_y_convertir(frase)

print("\n Palabras procesadas y ordenadas (estructura inmutable):")
print(resultado_final)
```

4. Registro y Consulta de Calificaciones de Exámenes
Descripción:
Implementa un sistema que gestione un registro de estudiantes y sus calificaciones. El programa debe permitir agregar nuevos registros, listar todos los existentes y consultar la calificación de un estudiante introduciendo su nombre. Se espera que la información se actualice y se presente de forma clara en cada paso.

```python
registro_estudiantes = {}

def agregar_registro():
    nombre = input("Ingrese el nombre del estudiante: ").strip().title()
    try:
        nota = float(input(f"Ingrese la calificación de {nombre}: "))
        registro_estudiantes[nombre] = nota
        print(f" Registro agregado: {nombre} -> {nota}")
    except ValueError:
        print(" Calificación no válida. Debe ser un número.")

def listar_registros():
    if not registro_estudiantes:
        print(" No hay registros aún.")
        return
    print("\n Lista de todos los estudiantes y sus calificaciones:")
    for nombre, nota in registro_estudiantes.items():
        print(f" - {nombre}: {nota}")

def consultar_calificacion():
    nombre = input("Ingrese el nombre del estudiante a consultar: ").strip().title()
    nota = registro_estudiantes.get(nombre)
    if nota is not None:
        print(f" {nombre} tiene una calificación de {nota}")
    else:
        print(" Estudiante no encontrado en el registro.")

def menu():
    while True:
        print("\n--- MENÚ ---")
        print("1. Agregar nuevo registro")
        print("2. Listar todos los registros")
        print("3. Consultar calificación de un estudiante")
        print("4. Salir")
        opcion = input("Seleccione una opción: ")

        if opcion == "1":
            agregar_registro()
        elif opcion == "2":
            listar_registros()
        elif opcion == "3":
            consultar_calificacion()
        elif opcion == "4":
            print(" Saliendo del sistema.")
            break
        else:
            print(" Opción inválida. Intente de nuevo.")


menu()
```

5. Calculadora de Estadísticas de Números
Descripción:
Desarrolla una aplicación que trabaje con una colección de números predefinidos. El sistema debe ofrecer al usuario un menú interactivo con opciones para calcular el promedio, identificar el valor máximo y mínimo, o eliminar un número. Tras cada operación, se mostrará tanto el resultado como el estado actual de la colección.

```python
numeros = [12, 45, 23, 89, 5, 34, 77]

def mostrar_coleccion():
    print(f"\n Colección actual: {numeros}")

def calcular_promedio():
    if numeros:
        promedio = sum(numeros) / len(numeros)
        print(f" Promedio: {promedio:.2f}")
    else:
        print(" No hay números en la colección.")

def valor_maximo_minimo():
    if numeros:
        print(f" Máximo: {max(numeros)}")
        print(f" Mínimo: {min(numeros)}")
    else:
        print(" No hay números en la colección.")

def eliminar_numero():
    try:
        num = int(input("Ingrese el número que desea eliminar: "))
        if num in numeros:
            numeros.remove(num)
            print(f"Número {num} eliminado.")
        else:
            print(" El número no está en la colección.")
    except ValueError:
        print(" Entrada inválida. Debe ser un número entero.")

def menu():
    while True:
        print("\n--- MENÚ ---")
        print("1. Calcular promedio")
        print("2. Mostrar valor máximo y mínimo")
        print("3. Eliminar un número")
        print("4. Mostrar colección actual")
        print("5. Salir")
        opcion = input("Seleccione una opción: ")

        if opcion == "1":
            calcular_promedio()
            mostrar_coleccion()
        elif opcion == "2":
            valor_maximo_minimo()
            mostrar_coleccion()
        elif opcion == "3":
            eliminar_numero()
            mostrar_coleccion()
        elif opcion == "4":
            mostrar_coleccion()
        elif opcion == "5":
            print(" Saliendo de la calculadora.")
            break
        else:
            print(" Opción no válida. Intente de nuevo.")

menu()
```
