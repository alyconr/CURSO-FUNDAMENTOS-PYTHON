1. Gestión de Inventario de Productos
Descripción:
Se debe desarrollar un sistema que administre un inventario compuesto por productos y sus cantidades. El programa presentará el listado inicial y resaltará aquellos artículos cuyo stock esté por debajo de un umbral definido. Luego, permitirá actualizar la cantidad de un producto específico, mostrando el inventario antes y después de la modificación.

```python
inventario = [
    ["Pinturas al óleo", 10],
    ["Lienzos", 5],
    ["Esculturas de arcilla", 2],
    ["Acuarelas", 8]
]
umbral = 3

print("Inventario actual:")
for producto in inventario:
    nombre = producto[0]
    cantidad = producto[1]

    if cantidad < umbral:
        print(nombre + ": " + str(cantidad) + " (Bajo stock)")
    else:
        print(nombre + ": " + str(cantidad))

producto_actualizar = input("Ingrese el nombre del producto a actualizar: ")

encontrado = False
for producto in inventario:
    if producto[0].lower() == producto_actualizar.lower():
        nueva_cantidad = int(input("Ingrese la nueva cantidad: "))

        print("\nInventario antes de la actualización:")
        for prod in inventario:
            nombre = prod[0]
            cantidad = prod[1]
            if cantidad < umbral:
                print(nombre + ": " + str(cantidad) + " (Bajo stock)")
            else:
                print(nombre + ": " + str(cantidad))

        producto[1] = nueva_cantidad

        print("\nInventario después de la actualización:")
        for prod in inventario:
            nombre = prod[0]
            cantidad = prod[1]
            if cantidad < umbral:
                print(nombre + ": " + str(cantidad) + " (Bajo stock)")
            else:
                print(nombre + ": " + str(cantidad))

        encontrado = True
        break
if not encontrado:
    print("Producto no encontrado en el inventario.")
```

2. Búsqueda de Palabras en un Texto
Descripción:
Crea una aplicación que solicite al usuario un párrafo y, a continuación, le permita buscar una palabra específica. El sistema debe contar cuántas veces aparece dicha palabra en el texto, mostrar las posiciones donde se encuentra (sin distinguir mayúsculas de minúsculas) y comunicar el resultado al usuario.



```python
texto = input("Ingresa un párrafo: ")
palabra = input("Ingresa la palabra que deseas buscar: ")

texto_lower = texto.lower()
palabra_lower = palabra.lower()

palabras = texto_lower.split()

posiciones = []
cantidad = 0

for i in range(len(palabras)):
    if palabras[i] == palabra_lower:
        posiciones.append(i)
        cantidad += 1

print("La palabra '", palabra, "' aparece ", cantidad, " veces.", sep="")

if cantidad > 0:
    print("Posiciones:", posiciones)
```

3. Conversión y Limpieza de Texto
Descripción:
Diseña una solución que pida al usuario una frase, la divida en palabras y elimine cualquier carácter no alfabético de cada una. Una vez procesada la información, la estructura resultante se transformará en una forma inmutable, y se presentará el contenido final al usuario de manera ordenada.

```python
print("Ingresa una frase:")
frase = input()

palabras = frase.split()
palabras_limpias = ["".join(letra for letra in palabra if ('a' <= letra <= 'z') or ('A' <= letra <= 'Z')) for palabra in palabras]

palabras_ordenadas = []
while palabras_limpias:
    menor = min(palabras_limpias)
    palabras_ordenadas += [menor]
    palabras_limpias = [p for p in palabras_limpias if p != menor]

print("Texto procesado:", tuple(palabras_ordenadas))
```

4. Registro y Consulta de Calificaciones de Exámenes
Descripción:
Implementa un sistema que gestione un registro de estudiantes y sus calificaciones. El programa debe permitir agregar nuevos registros, listar todos los existentes y consultar la calificación de un estudiante introduciendo su nombre. Se espera que la información se actualice y se presente de forma clara en cada paso.

```python
calificaciones = []

while True:
    print("\nMenú:")
    print("1. Agregar calificación")
    print("2. Listar calificaciones")
    print("3. Consultar calificación por nombre")
    print("4. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        nombre = input("Ingrese el nombre del estudiante: ")
        nota = float(input("Ingrese la calificación: "))
        calificaciones = calificaciones + [(nombre, nota)]
        print("Calificación agregada correctamente.")

    if opcion == "2":
        print("\nRegistro de calificaciones:")
        for estudiante in calificaciones:
            print(estudiante[0], "-", estudiante[1])

    if opcion == "3":
        buscar = input("Ingrese el nombre del estudiante a consultar: ")
        encontrado = False
        for estudiante in calificaciones:
            if estudiante[0].lower() == buscar.lower():
                print("Calificación de", estudiante[0] + ":", estudiante[1])
                encontrado = True
        if not encontrado:
            print("Estudiante no encontrado.")

    if opcion == "4":
        print("Saliendo del programa.")
        break
```

5. Calculadora de Estadísticas de Números
Descripción:
Desarrolla una aplicación que trabaje con una colección de números predefinidos. El sistema debe ofrecer al usuario un menú interactivo con opciones para calcular el promedio, identificar el valor máximo y mínimo, o eliminar un número. Tras cada operación, se mostrará tanto el resultado como el estado actual de la colección.

```python
numeros = [12, 45, 78, 23, 56, 89, 14, 67]

while True:
    print("\nMenú:")
    print("1. Calcular promedio")
    print("2. Encontrar valor máximo y mínimo")
    print("3. Eliminar un número")
    print("4. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        promedio = sum(numeros) / len(numeros)
        print("Promedio:", promedio)
        print("Estado actual de la colección:", numeros)

    if opcion == "2":
        maximo = max(numeros)
        minimo = min(numeros)
        print("Valor máximo:", maximo)
        print("Valor mínimo:", minimo)
        print("Estado actual de la colección:", numeros)

    if opcion == "3":
        eliminar = int(input("Ingrese el número a eliminar: "))
        if eliminar in numeros:
            numeros = [num for num in numeros if num != eliminar]
            print("Número eliminado correctamente.")
        else:
            print("El número no está en la colección.")
        print("Estado actual de la colección:", numeros)

    if opcion == "4":
        print("Saliendo del programa.")
        break
```
