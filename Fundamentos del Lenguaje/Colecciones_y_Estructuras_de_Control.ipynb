1. Gestión de Inventario de Productos
Descripción:
Se debe desarrollar un sistema que administre un inventario compuesto por productos y sus cantidades. El programa presentará el listado inicial y resaltará aquellos artículos cuyo stock esté por debajo de un umbral definido. Luego, permitirá actualizar la cantidad de un producto específico, mostrando el inventario antes y después de la modificación.

```python
productos = ["Manzana", "Pera", "Banano"]
unidades = [10, 5, 30]
stock=10

print("Inventario de producto:")
for i, producto in enumerate(productos, 1):  # Cambié 'productos' por 'producto'
    cantidad = unidades[i - 1]  # Obtener la cantidad correspondiente al producto
    
    if cantidad <= stock:
        print(f"{i}. {producto} - {cantidad} unidades (¡Stock bajo!)")
    else:
        print(f"{i}. {producto} - {cantidad} unidades")
        
        actualiza = int(input("\nIngresa el número del producto que deseas actualizar: "))
        if 1 <= actualiza <= len(productos):
            nueva_cantidad = int(input(f"Ingrese la nueva cantidad para {productos[actualiza - 1]}: "))
            unidades[actualiza - 1] = nueva_cantidad  # Actualizar la cantidad en la lista 'unidades'
            print("\nCantidad actualizada correctamente.")
        else:
            print("Número fuera de rango. Inténtalo de nuevo.")
            
print("Inventario actualizado:")
for i, producto in enumerate(productos, 1):
    cantidad = unidades[i - 1]
    if cantidad <= stock:
        print(f"{i}. {producto} - {cantidad} unidades (¡Stock bajo!)")
    else:
        print(f"{i}. {producto} - {cantidad} unidades")
               
            
            
            
        
     

```

2. Búsqueda de Palabras en un Texto
Descripción:
Crea una aplicación que solicite al usuario un párrafo y, a continuación, le permita buscar una palabra específica. El sistema debe contar cuántas veces aparece dicha palabra en el texto, mostrar las posiciones donde se encuentra (sin distinguir mayúsculas de minúsculas) y comunicar el resultado al usuario.



```python
parrafo = input("Ingresa un párrafo: ").lower()
palabra = input("Ingresa la palabra a buscar: ").lower()


conteo = parrafo.count(palabra)  # Cuenta las veces que aparece la palabra
posiciones = []  
indice = parrafo.find(palabra)

while indice != -1:
    posiciones.append(indice)
    indice = parrafo.find(palabra, indice + 1)  # Buscar la siguiente aparición

if posiciones:
    print(f"\nLa palabra '{palabra}' se encontró {len(posiciones)} veces en las posiciones: {posiciones}")
else:
    print(f"\nLa palabra '{palabra}' no se encuentra en el párrafo.")
```

3. Conversión y Limpieza de Texto
Descripción:
Diseña una solución que pida al usuario una frase, la divida en palabras y elimine cualquier carácter no alfabético de cada una. Una vez procesada la información, la estructura resultante se transformará en una forma inmutable, y se presentará el contenido final al usuario de manera ordenada.

```python
frase = input("Ingresa una frase: ")
palabras = frase.split()
palabras_limpias = [''.join(letra for letra in palabra if letra.isalpha()) for palabra in palabras]
resultado_final = tuple(sorted(palabras_limpias))
print("Resultado final:", resultado_final)
```

```python
registros = {}

cantidad = int(input("¿Cuántos estudiantes desea registrar? "))
for i in range(cantidad):
    nombre = input("Ingrese el nombre del estudiante: ")
    calificacion = float(input(f"Ingrese la calificación de {nombre}: "))
    registros[nombre] = calificacion  

# Mostrar todos los registros
print("\nRegistro de estudiantes:")
for nombre in registros:
    print(f"{nombre}: {registros[nombre]}")

# Consultar la calificación de un estudiante
nombre_buscar = input("\nIngrese el nombre del estudiante para consultar su calificación: ")
if nombre_buscar in registros:
    print(f"La calificación de {nombre_buscar} es {registros[nombre_buscar]}")
else:
    print("El estudiante no está registrado.")
```

```python
numeros = [12, 45, 7, 23, 89, 34, 56, 10]

for _ in range(5):  # El menú se mostrará hasta 5 veces
    print("\nLista actual:", numeros)
    print("1. Calcular promedio")
    print("2. Obtener máximo")
    print("3. Obtener mínimo")
    print("4. Eliminar un número")
    print("5. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        print(f"Promedio: {sum(numeros) / len(numeros)}")
    elif opcion == "2":
        print(f"Máximo: {max(numeros)}")
    elif opcion == "3":
        print(f"Mínimo: {min(numeros)}")
    elif opcion == "4":
        num = int(input("Ingrese el número a eliminar: "))
        if num in numeros:
            numeros.remove(num)
            print(f"Se eliminó {num}. Lista actualizada: {numeros}")
        else:
            print("Número no encontrado.")
    elif opcion == "5":
        print("Saliendo del programa...")
        break
    else:
        print("Opción no válida.")
```
