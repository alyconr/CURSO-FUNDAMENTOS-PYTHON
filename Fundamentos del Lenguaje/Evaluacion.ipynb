Procesamiento de Números del 1 al 20 Se requiere desarrollar un programa que trabaje con los números comprendidos entre el 1 y el 20, de forma que se identifiquen y muestren aquellos que sean divisibles por dos, acompañados de su posición en la secuencia. Además, el sistema debe recorrer la colección para determinar cuántos números no cumplen esa propiedad y calcular la suma de los que sí la cumplen. El objetivo es diseñar la lógica necesaria para examinar y procesar los números según las condiciones indicadas.

```python
numeros = list(range(1, 21))
contador_no_pares = 0
suma_pares = 0
numeros_pares = []

for i, numero in enumerate(numeros):
    if numero % 2 == 0:
        numeros_pares.append((i + 1, numero))
        suma_pares += numero
    else:
        contador_no_pares += 1

print("Números divisibles por 2 y su posición:")
for posicion, numero in numeros_pares:
    print(f"Posición: {posicion}, Número: {numero}")

print(f"\nCantidad de números no divisibles por 2: {contador_no_pares}")
print(f"Suma de los números divisibles por 2: {suma_pares}")
```

2. Administración de Actividades Diarias Imagina una aplicación para gestionar las actividades cotidianas de una persona, donde se disponga de una serie de tareas, algunas de las cuales indican que ya han sido concluidas. El desafío consiste en presentar cada actividad junto a su posición en la colección, filtrar aquellas que contengan una marca de finalización y actualizar el conjunto de actividades. Finalmente, se debe exhibir el estado del conjunto antes y después de aplicar el filtro, permitiendo al usuario observar los cambios realizados.

```python
def agregar_tarea(tareas, descripcion):
    tareas.append({"descripcion": descripcion, "completada": False})

def mostrar_tareas(tareas):
    for i, tarea in enumerate(tareas):
        estado = "✔" if tarea["completada"] else "X"
        print(f"{i + 1}. {estado} {tarea['descripcion']}")

def marcar_tarea_completada(tareas, indice):
    if 0 <= indice < len(tareas):
        tareas[indice]["completada"] = True

def filtrar_tareas_completadas(tareas):
    return [tarea for tarea in tareas if tarea["completada"]]


tareas = []

agregar_tarea(tareas, "Lavar la moto")
agregar_tarea(tareas, "Estudiar para el examen de JavaScript")
agregar_tarea(tareas, "Hacer Mercado")
agregar_tarea(tareas, "Comprar una gorra")

marcar_tarea_completada(tareas, 1)
marcar_tarea_completada(tareas, 2)

print("Estado de las tareas antes del filtro:")
mostrar_tareas(tareas)

tareas_completadas = filtrar_tareas_completadas(tareas)

print("\nTareas completadas:")
for tarea in tareas_completadas:
    print(f"✔ {tarea['descripcion']}")
```

3. Filtrado y Transformación de Texto Propón un sistema en el que se solicite al usuario una secuencia de caracteres. A partir de la entrada, se debe descomponer el texto en sus componentes individuales y extraer únicamente aquellos que correspondan a letras del alfabeto, omitiendo números y símbolos. Una vez realizado este filtrado, la colección resultante deberá ser convertida en una estructura que no permita modificaciones, y el resultado final se debe mostrar al usuario.



```python
def filtrar_y_transformar_texto():

    texto = input("Introduce una secuencia de caracteres: ")
    letras_filtradas = [caracter for caracter in texto if caracter.isalpha()]
    resultado_final = tuple(letras_filtradas)
    print("Resultado final:", resultado_final)

filtrar_y_transformar_texto()
```

4. Informe Interactivo de Alumnos Considera dos conjuntos de datos paralelos: uno con nombres de estudiantes y otro con sus respectivas calificaciones. El reto consiste en elaborar un informe que relacione cada alumno con su nota, presentando también la posición de cada registro. Posteriormente, el sistema debe permitir al usuario consultar la nota de un alumno específico introduciendo la posición correspondiente, finalizando la consulta si se ingresa un valor fuera del rango válido. La solución deberá vincular ambas colecciones de manera sincronizada y ofrecer una interacción dinámica.

```python
estudiantes = ["Juan", "Mariana", "Yesid", "Camila", "James"]
notas = [5.0, 3.8, 4.2, 2.0, 3.5]

for dato in enumerate(estudiantes):
  print(dato)

while True:
  indice = int(input("Ingrese el indice del estudiante; "))
  if indice < len(estudiantes) and indice >= 0:
    print(f"Estudiante: {estudiantes[indice]} tiene la siguiente nota: {notas[indice]}")
  else:
    print("El indice ingresado no es valido")
    break

print("Fin del programa")
```

5. Aplicación de Operaciones sobre Valores Numéricos Diseña una aplicación que inicie con una serie de valores numéricos predefinidos. El usuario interactuará mediante un menú que le permita realizar distintas operaciones: calcular la suma total de los valores, reordenarlos según un criterio específico o eliminar uno de ellos según una indicación. La interfaz deberá ofrecer la posibilidad de repetir estas acciones hasta que el usuario decida finalizar. El desafío es organizar el flujo de la interacción y manipular la colección de números de acuerdo a la opción elegida..

```python
valores = [10, 20, 345, 80, 50]
print(valores)
while True:
        opcion = input("seleccione una opcion: 1= suma, 2= reordenar, 3= eliminar, 4= salir ")
        if opcion == '1':
            print(f"Suma total: {sum(valores)}")
        elif opcion == '2':
            valores.sort()
            print(f"Valores reordenados: {valores}")
        elif opcion == '3':
            try:
                valores.remove(float(input("Valor a eliminar: ")))
                print(f"Valores actuales: {valores}")
            except ValueError:
                print("Valor no encontrado.")
        elif opcion == '4':
            print("¡Hasta luego!")
            break
        else:
            print("Opción no válida.")
```
