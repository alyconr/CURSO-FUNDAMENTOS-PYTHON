## Ejercicio 1: Calculadora Avanzada

**Descripción:**  
Crea una función que actúe como calculadora. La función recibirá dos números y un operador (por ejemplo, '+', '-', '*', '/', '**', '//') para realizar la operación correspondiente.  
- Utiliza estructuras de control (if/else) para seleccionar la operación a realizar.  
- Usa manejo de excepciones para controlar errores como la división por cero y entradas no numéricas.  
- Puedes importar el módulo `math` para operaciones adicionales si es necesario.

**Ejemplo de salida:**  

Ingrese el primer número: 10
Ingrese el segundo número: 5
Ingrese el operador (+, -, *, /, **, //): /
Resultado: 10 / 5 = 2.0

```python
def mi_calculadora():
    print("=== CALCULADORA SIMPLE ===")
    print("Operaciones disponibles: + (suma), - (resta), * (multiplicación), / (división)")

    try:
        num1 = int(input("Ingrese el primer número: "))
        num2 = int(input("Ingrese el segundo número: "))
        signo = input("Seleccione una operación (+, -, *, /): ")

        if signo == '+':
            resultado = num1 + num2
        elif signo == '-':
            resultado = num1 - num2
        elif signo == '*':
            resultado = num1 * num2
        elif signo == '/':
            resultado = num1 / num2
        else:
            print("Operación no válida.")
            return

        print(f"Resultado: {num1} {signo} {num2} = {resultado}")

    except ValueError:
        print("Por favor, ingrese solo números enteros.")
    except ZeroDivisionError:
        print("Error: No se puede dividir por cero.")

mi_calculadora()
```

## Ejercicio 2: Administrador de Inventarios

**Descripción:**  
Desarrolla un sistema para gestionar el inventario de productos usando un diccionario donde cada clave es el nombre del producto y su valor la cantidad en stock.  
- Implementa funciones para agregar, actualizar y eliminar productos.  
- Utiliza estructuras de control para recorrer y mostrar el inventario.  
- Incluye manejo de excepciones para capturar intentos de actualizar o eliminar un producto inexistente.

**Ejemplo de salida:**  
Inventario inicial: {'manzanas': 50, 'naranjas': 30, 'peras': 20}
Actualizando stock de 'peras' a 25...
Producto 'bananas' agregado con 40 unidades.
Eliminando 'naranjas'...
Inventario final: {'manzanas': 50, 'peras': 25, 'bananas': 40}

```python
def ver_inventario(stock):
    """Muestra los productos disponibles en el inventario"""
    print("\n📦  INVENTARIO ACTUAL")
    print("---------------------------")
    for item, cantidad in stock.items():
        print(f"🔸 {item.capitalize()}: {cantidad} unidades")
    print("---------------------------\n")

def nuevo_producto(stock):
    """Agrega un nuevo producto al inventario"""
    nombre = input("Nombre del nuevo producto: ").lower()
    if nombre in stock:
        print(f"⚠️ El producto '{nombre}' ya está registrado.")
        return

    try:
        cantidad = int(input(f"Ingrese la cantidad para '{nombre}': "))
        if cantidad < 0:
            print("❌ No se pueden agregar cantidades negativas.")
            return
        stock[nombre] = cantidad
        print(f"✅ Producto '{nombre.capitalize()}' añadido con {cantidad} unidades.")
    except ValueError:
        print("❌ Error: Ingrese una cantidad válida (número entero).")

def modificar_producto(stock):
    """Modifica la cantidad de un producto existente"""
    nombre = input("Producto a modificar: ").lower()

    if nombre not in stock:
        print(f"❌ El producto '{nombre}' no existe.")
        return

    try:
        cantidad = int(input(f"Nueva cantidad para '{nombre}' (actual: {stock[nombre]}): "))
        if cantidad < 0:
            print("❌ No se puede asignar una cantidad negativa.")
            return
        stock[nombre] = cantidad
        print(f"🔄 Producto '{nombre.capitalize()}' actualizado a {cantidad} unidades.")
    except ValueError:
        print("❌ Error: Debe ingresar un número válido.")

def quitar_producto(stock):
    """Elimina un producto del inventario"""
    nombre = input("Producto a eliminar: ").lower()

    if nombre in stock:
        confirmar = input(f"¿Seguro que desea eliminar '{nombre}'? (s/n): ").lower()
        if confirmar == 's':
            del stock[nombre]
            print(f"🗑️ Producto '{nombre.capitalize()}' eliminado.")
        else:
            print("❎ Operación cancelada.")
    else:
        print(f"❌ El producto '{nombre}' no está en el inventario.")

def mostrar_menu():
    """Muestra las opciones del sistema"""
    print("\n📋 MENÚ DE OPCIONES")
    print("1️⃣ Ver inventario")
    print("2️⃣ Agregar nuevo producto")
    print("3️⃣ Modificar producto existente")
    print("4️⃣ Eliminar producto")
    print("5️⃣ Salir")
    print("---------------------------")

# Inventario inicial
stock_inicial = {'manzanas': 50, 'naranjas': 30, 'peras': 20}

# Bucle principal
while True:
    ver_inventario(stock_inicial)
    mostrar_menu()

    try:
        eleccion = int(input("Seleccione una opción (1-5): "))

        if eleccion == 1:
            ver_inventario(stock_inicial)
        elif eleccion == 2:
            nuevo_producto(stock_inicial)
        elif eleccion == 3:
            modificar_producto(stock_inicial)
        elif eleccion == 4:
            quitar_producto(stock_inicial)
        elif eleccion == 5:
            print("👋 Cerrando sistema de inventario...")
            break
        else:
            print("❌ Opción no reconocida. Elija entre 1 y 5.")
    except ValueError:
        print("❌ Ingrese un número válido.")

# Inventario final
print("\n📦 INVENTARIO FINAL")
ver_inventario(stock_inicial)
```

## Ejercicio 3: Análisis de Texto

**Descripción:**  
Crea una función que reciba una cadena de texto y calcule la frecuencia de cada palabra en el mismo.  
- Convierte el texto a minúsculas y separa las palabras usando algún método (puedes utilizar expresiones regulares con el módulo `re`).  
- Emplea un diccionario para almacenar la palabra (clave) y su frecuencia (valor).  
- Controla excepciones en caso de que el usuario no ingrese ningún texto o ingrese caracteres inesperados.

**Ejemplo de salida:**  
Ingrese un texto: "Hola mundo, hola a todos en el mundo"
Frecuencia de palabras: hola: 2
mundo: 2
a: 1
todos: 1
en: 1
el: 1

```python
import re

def contar_palabras(texto):
    """Cuenta la frecuencia de cada palabra en un texto dado."""

    if not texto or not isinstance(texto, str):
        return "⚠️ Error: Ingresa un texto válido."

    # Pasar a minúsculas y extraer palabras con regex
    lista_palabras = re.findall(r'\b\w+\b', texto.lower())

    # Diccionario para almacenar las cantidades
    conteo = {}

    for palabra in lista_palabras:
        if palabra in conteo:
            conteo[palabra] += 1
        else:
            conteo[palabra] = 1

    return conteo

# Ejemplo interactivo
entrada_usuario = input("✏️ Escribe un texto para analizar: ")
frecuencia = contar_palabras(entrada_usuario)

# Mostrar el resultado
if isinstance(frecuencia, dict):
    print("\n📊 Frecuencia de palabras:")
    for palabra, veces in frecuencia.items():
        print(f"  - {palabra}: {veces}")
else:
    print(frecuencia)
```

## Ejercicio 4: Conversor de Unidades

**Descripción:**  
Desarrolla un conjunto de funciones para convertir unidades métricas. Por ejemplo:  
- De centímetros a pulgadas.  
- De kilómetros a millas.  
Utiliza estructuras de control para verificar que la entrada sea numérica, y maneja excepciones cuando el usuario ingrese datos inválidos.

**Ejemplo de salida:**  
Seleccione la conversión:

Centímetros a pulgadas

Kilómetros a millas
Opción: 1
Ingrese la cantidad en centímetros: 100
Resultado: 100 cm = 39.37 pulgadas

```python
# Conversor de unidades: cm a pulgadas y km a millas

def convertir_cm_a_pulgadas(centimetros):
    return centimetros * 0.393701

def convertir_km_a_millas(kilometros):
    return kilometros * 0.621371

print("\n🌐 CONVERSOR DE UNIDADES")
print("1️⃣ Centímetros a pulgadas")
print("2️⃣ Kilómetros a millas")

while True:
    opcion = input("\nSeleccione una opción (1/2) o escriba 'salir' para terminar: ").strip().lower()

    if opcion == 'salir':
        print("👋 Finalizando el conversor...")
        break

    if opcion not in ('1', '2'):
        print("❌ Opción no válida. Por favor, seleccione 1, 2 o 'salir'.")
        continue

    try:
        cantidad = float(input("🔢 Ingrese el valor que desea convertir: "))
        if cantidad < 0:
            print("⚠️ El número debe ser positivo.")
            continue
    except ValueError:
        print("❌ Error: Ingrese un número válido.")
        continue

    if opcion == '1':
        resultado = convertir_cm_a_pulgadas(cantidad)
        print(f"📏 {cantidad} cm = {resultado:.2f} pulgadas")
    else:
        resultado = convertir_km_a_millas(cantidad)
        print(f"🛣️ {cantidad} km = {resultado:.2f} millas")
```

## Ejercicio 5: Juego "Adivina el Número"

**Descripción:**  
Implementa un juego en el que la computadora selecciona aleatoriamente un número en un rango (por ejemplo, 1 a 50) y el usuario debe adivinarlo.  
- Emplea el módulo `random` para generar el número secreto.  
- Utiliza un bucle que permita múltiples intentos, con retroalimentación indicando si el número ingresado es mayor o menor que el secreto.  
- Gestiona excepciones para capturar entradas no numéricas.

**Ejemplo de salida:**  
¡Bienvenido al juego "Adivina el Número"!
Estoy pensando en un número entre 1 y 50.
Intento 1 - Ingresa tu número: 25
El número secreto es mayor.
Intento 2 - Ingresa tu número: 35
El número secreto es menor.
Intento 3 - Ingresa tu número: 30
¡Felicidades! Has adivinado el número secreto: 30

```python
import random

def adivina_el_numero():
    """Juego para adivinar un número aleatorio entre 1 y 20"""
    numero_aleatorio = random.randint(1, 20)
    intentos_realizados = 0
    intentos_maximos = 6

    print("🎲 ¡Estoy pensando en un número del 1 al 20!")

    while intentos_realizados < intentos_maximos:
        try:
            intento = int(input(f"\n🔍 Intento #{intentos_realizados + 1}: Adivina el número: "))
            intentos_realizados += 1

            if intento < numero_aleatorio:
                print("⬇️ Muy bajo.")
            elif intento > numero_aleatorio:
                print("⬆️ Muy alto.")
            else:
                print(f"🎉 ¡Felicidades! Adivinaste el número en {intentos_realizados} intentos.")
                return

        except ValueError:
            print("❌ Solo se permiten números enteros.")
            continue  # No se cuenta como intento válido

    print(f"\n❗ Lo siento, no lo lograste. El número correcto era {numero_aleatorio}.")

# Ejecutar el juego
adivina_el_numero()
```

## Ejercicio 6: Calculadora de Factorial

**Descripción:**  
Crea una función que calcule el factorial de un número entero ingresado por el usuario.  
- La función debe utilizar estructuras de control para iterar (o una solución recursiva) y calcular el factorial.  
- Implementa manejo de excepciones para asegurarte de que el número ingresado es entero y no negativo.

**Ejemplo de salida:**  
Ingrese un número entero para calcular su factorial: 5
El factorial de 5 es 120

```python
def factorial(n):
    """Calcula el factorial de un número de forma recursiva"""
    return 1 if n == 0 else n * factorial(n - 1)

def ejecutar_calculo_factorial():
    """Solicita al usuario un número y muestra su factorial"""
    while True:
        try:
            valor = int(input("🔢 Ingresa un número entero para calcular su factorial: "))
            if valor < 0:
                print("❌ El número no puede ser negativo.")
                continue
            resultado = factorial(valor)
            print(f"✅ El factorial de {valor} es: {resultado}")
            break
        except ValueError:
            print("⚠️ Error: Solo se permiten números enteros.")

# Iniciar proceso
ejecutar_calculo_factorial()
```

## Ejercicio 7: Manejo de Fechas y Horas

**Descripción:**  
Desarrolla una función que reciba una fecha ingresada por el usuario en formato "DD/MM/AAAA" y determine cuántos días faltan para esa fecha a partir de hoy.  
- Importa el módulo `datetime` para trabajar con fechas.  
- Utiliza estructuras de control para validar el formato y maneja excepciones en caso de error en el ingreso de la fecha.

**Ejemplo de salida:**  
Ingrese una fecha (DD/MM/AAAA): 25/12/2023
Faltan 150 días para el 25/12/2023.

```python
from datetime import datetime

def calcular_dias_restantes():
    """Calcula cuántos días faltan para una fecha específica ingresada por el usuario."""
    try:
        entrada = input("📅 Ingrese una fecha en formato DD/MM/AAAA: ")
        fecha_destino = datetime.strptime(entrada, "%d/%m/%Y")
        hoy = datetime.now()

        if fecha_destino < hoy:
            print("⏰ Esa fecha ya pasó.")
        else:
            dias_restantes = (fecha_destino - hoy).days
            print(f"📆 Faltan {dias_restantes} días para el {entrada}.")
    except ValueError:
        print("❌ Formato no válido. Usa el formato correcto: DD/MM/AAAA.")

# Ejecutar la función
calcular_dias_restantes()
```

## Ejercicio 8: Registro de Estudiantes

**Descripción:**  
Implementa un sistema que gestione un registro de estudiantes utilizando un diccionario, donde la clave sea el nombre y el valor la calificación.  
- Crea funciones para agregar un estudiante, actualizar la calificación, eliminar estudiantes y listar todo el registro.  
- Utiliza un bucle para mostrar un menú interactivo al usuario.  
- Maneja excepciones para gestionar errores en la selección de opciones o al intentar modificar un estudiante inexistente.

**Ejemplo de salida:**  
Registro de estudiantes inicial: {'Ana': 90, 'Luis': 78, 'Carlos': 85}

Menú:

Agregar estudiante

Actualizar calificación

Eliminar estudiante

Listar estudiantes

Salir

Opción: 2
Ingrese el nombre del estudiante: Luis
Ingrese la nueva calificación: 82
Estudiante actualizado: Luis - 82

Registro final: {'Ana': 90, 'Luis': 82, 'Carlos': 85}

```python
def gestion_notas():
    notas = {'Ana': 90, 'Luis': 78, 'Carlos': 85}

    while True:
        print("\n📚 MENÚ DE GESTIÓN DE NOTAS")
        print("1️⃣ Agregar estudiante")
        print("2️⃣ Actualizar calificación")
        print("3️⃣ Eliminar estudiante")
        print("4️⃣ Ver todos los registros")
        print("5️⃣ Salir")

        try:
            opcion = int(input("👉 Elige una opción (1-5): "))
            
            if opcion == 1:
                nombre = input("📝 Nombre del estudiante: ")
                try:
                    nota = int(input("📊 Calificación: "))
                    notas[nombre] = nota
                    print(f"✅ {nombre} agregado con calificación {nota}")
                except ValueError:
                    print("❌ Calificación inválida.")
                    
            elif opcion == 2:
                nombre = input("✏️ Nombre del estudiante a actualizar: ")
                if nombre in notas:
                    try:
                        nueva_nota = int(input("📈 Nueva calificación: "))
                        notas[nombre] = nueva_nota
                        print(f"🔄 Calificación de {nombre} actualizada a {nueva_nota}")
                    except ValueError:
                        print("❌ Calificación inválida.")
                else:
                    print("❌ Estudiante no encontrado.")
                    
            elif opcion == 3:
                nombre = input("🗑️ Nombre del estudiante a eliminar: ")
                if nombre in notas:
                    notas.pop(nombre)
                    print(f"✅ {nombre} ha sido eliminado del registro.")
                else:
                    print("❌ Estudiante no encontrado.")
                    
            elif opcion == 4:
                print("\n📋 LISTADO DE ESTUDIANTES:")
                for estudiante, nota in notas.items():
                    print(f"🔹 {estudiante}: {nota}")
                    
            elif opcion == 5:
                print("👋 Saliendo del sistema de registro...")
                break
            else:
                print("❌ Opción fuera de rango. Seleccione del 1 al 5.")
                
        except ValueError:
            print("❌ Entrada inválida. Debes ingresar un número del 1 al 5.")

# Iniciar el sistema
gestion_notas()
```

## Ejercicio 9: Generador de Contraseñas Aleatorias

**Descripción:**  
Crea una función que genere una contraseña aleatoria de una longitud especificada por el usuario.  
- Utiliza el módulo `random` junto con la biblioteca `string` para construir la contraseña a partir de letras (mayúsculas y minúsculas), dígitos y caracteres especiales.  
- Valida la entrada y maneja excepciones para asegurarte de que la longitud es un número entero positivo.

**Ejemplo de salida:**  
Ingrese la longitud deseada para la contraseña: 10
Contraseña generada: A8b#K3d!Qz

```python
import random
import string
import secrets

def crear_contraseña_segura():
    """Genera una contraseña segura con longitud definida por el usuario."""

    # Grupos de caracteres disponibles
    caracteres_disponibles = [
        string.ascii_lowercase,   # Letras minúsculas
        string.ascii_uppercase,   # Letras mayúsculas
        string.digits,            # Números
        string.punctuation        # Símbolos especiales
    ]

    # Solicitar al usuario la longitud de la contraseña
    while True:
        try:
            largo = int(input("🔐 Elige la longitud de tu contraseña (entre 8 y 64): "))
            if 8 <= largo <= 64:
                break
            print("⚠️ La longitud debe estar entre 8 y 64 caracteres.")
        except ValueError:
            print("❌ Por favor, ingresa un número válido.")

    # Garantizamos al menos un carácter de cada grupo
    clave = [secrets.choice(grupo) for grupo in caracteres_disponibles]

    # Rellenamos con caracteres aleatorios de todos los grupos combinados
    conjunto_total = ''.join(caracteres_disponibles)
    clave.extend(secrets.choice(conjunto_total) for _ in range(largo - 4))

    # Mezclamos para evitar patrones
    random.shuffle(clave)

    # Devolvemos la contraseña como string
    contraseña_final = ''.join(clave)
    return contraseña_final

# Prueba de uso
if __name__ == "__main__":
    print("\n✨ Contraseña generada:")
    print(crear_contraseña_segura())
```

## Ejercicio 10: Buscador de Archivos en un Directorio

**Descripción:**  
Desarrolla una función que reciba el nombre de un directorio y una subcadena para buscar archivos cuyo nombre la contenga.  
- Utiliza el módulo `os` para listar el contenido del directorio.  
- Aplica estructuras de control para filtrar los archivos según la subcadena.  
- Implementa manejo de excepciones para controlar errores como directorio no existente o permisos insuficientes.

**Ejemplo de salida:**  
Ingrese la ruta del directorio: /ruta/al/directorio
Ingrese la subcadena a buscar: reporte
Archivos encontrados que contienen "reporte":

reporte_enero.pdf

reporte_febrero.xlsx

resumen_reporte.txt

```python
import os  # Importamos el módulo que nos ayuda a trabajar con carpetas y archivos

# Esta función busca archivos que contengan una palabra en su nombre
def buscar_archivos(directorio, subcadena):
    try:
        # Lista todo lo que hay dentro del directorio
        archivos = os.listdir(directorio)
        
        # Aquí guardamos solo los archivos que contienen la subcadena que buscamos
        archivos_encontrados = []
        for archivo in archivos:
            if subcadena in archivo:
                archivos_encontrados.append(archivo)
        
        # Mostramos los archivos encontrados
        if archivos_encontrados:
            print("\nArchivos que contienen la palabra '{}':".format(subcadena))
            for archivo in archivos_encontrados:
                print(archivo)
        else:
            print(f"\nNo se encontraron archivos que contengan '{subcadena}'.")
    
    # Si el directorio no existe
    except FileNotFoundError:
        print(f"\nError: El directorio '{directorio}' no existe.")
    
    # Si no tenemos permiso para entrar al directorio
    except PermissionError:
        print(f"\nError: No tienes permisos para acceder al directorio '{directorio}'.")
    
    # Para cualquier otro error raro que pueda pasar
    except Exception as error:
        print(f"\nOcurrió un error: {error}")

# Código principal
if __name__ == "__main__":
    print("=== Buscador de archivos en un directorio ===")
    
    # Pedimos al usuario la ruta y la palabra a buscar
    ruta = input("Escribe la ruta del directorio: ")
    palabra = input("Escribe la palabra a buscar en los nombres de los archivos: ")
    
    # Llamamos a la función
    buscar_archivos(ruta, palabra)
```
