```python
def frecuencia(texto):
    try:
        if not texto.strip():
            print("Error, no se ingresó ningún texto. Vuelva a intentarlo")
            return

        texto_def = texto.lower()
        palabras = texto_def.split()

        if not palabras:
            print("Error el texto no tiene caracteres validos. Vuelva a intentar")
            return

        frecuencia = {}
        for palabra in palabras:
            if palabra in frecuencia:
                frecuencia[palabra] += 1
            else:
                frecuencia[palabra] = 1

        print("Frecuencia de palabras:")
        for palabra, conteo in frecuencia.items():
            print(f"{palabra}: {conteo}")

    except:
        print("Error en analisis de texto. Vuelva a intentar")

texto = input("Ingresa un texto: ")
frecuencia(texto)
```

```python
import os

def buscar_archivos(directorio, subcadena):
    try:

        if not os.path.isdir(directorio):
            print(f"El directorio '{directorio}' no existe.")
            return
        

        archivos_encontrados = []
        for archivo in os.listdir(directorio):
            ruta_completa = os.path.join(directorio, archivo)

            if os.path.isfile(ruta_completa) and subcadena in archivo:
                archivos_encontrados.append(archivo)


        if archivos_encontrados:
            print("Archivos encontrados:")
            for archivo in archivos_encontrados:
                print(f"- {archivo}")
        else:
            print(f"No se encontraron archivos que contengan '{subcadena}'.")
    
    except Exception as e:
        print(f"Ocurrió un error: {e}")

directorio_usuario = input("Ingrese la ruta del directorio: ")
subcadena_usuario = input("Ingrese la subcadena a buscar: ")

buscar_archivos(directorio_usuario, subcadena_usuario)
```
