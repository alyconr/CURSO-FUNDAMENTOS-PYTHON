```python
def order_list(numeros):

  pares = sorted([num for num in numeros if num % 2 == 0])
  impares = sorted([num for num in numeros if num % 2 != 0])
  return pares, impares

lista_numeros = [5, 2, 9, 1, 8, 4, 7, 0, 3, 6]
pares, impares = order_list(lista_numeros)
print(f"Números pares: {pares}")
print(f"Números impares: {impares}")
```

```python
def contador(palabras):
  conteo = {}
  for letra in palabras:
    conteo[letra] = conteo.get(letra, 0) + 1
  return conteo

palabra = "colombia"
conteo = contador(palabra)
print(conteo)
```

```python
def numero_a_palabras(numero):

  nombres = {
      "0": "cero", "1": "uno", "2": "dos", "3": "tres", "4": "cuatro",
      "5": "cinco", "6": "seis", "7": "siete", "8": "ocho", "9": "nueve"
  }
  cadena_numero = str(numero)
  nombres_digitos = [nombres[digito] for digito in cadena_numero]
  return " - ".join(nombres_digitos)

numero = 134
cadena_palabras = numero_a_palabras(numero)
print(f"El número {numero} en palabras es: '{cadena_palabras}'")
```
