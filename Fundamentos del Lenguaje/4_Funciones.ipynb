## Número de argumentos indeterminados

Algunas veces nos pueden llamar a una función con un número variable de argumentos.  En ese caso, debemos preparar a la función para que pueda tomarlos y gestionarlos en su interior.

````python
def suma(*args):
    print(type(args))
    return sum(args)/len(args)

print("La suma delos numeros es:", suma(3,5,7,9,10,8,6,4,2,1))
````

Si además de recibir valores variables, queremos recibir los nombres de las variables podemos definir nuestro parámetro como un diccionario variable. Veamos:

````python
def mostrar_info(**kargs):
    print(type(kargs))
    for k in kargs:
        print(f"{k} = {kargs[k]}")

mostrar_info(nombre="Justin", apellido="paz") 
mostrar_info(nombre="valentina", apellido="sierra", edad=20)

````

## Retorno de valores

Como en la mayoría de lenguajes de programación en Python empleamos la palabra reservada **return** para devolver un valor o resultado desde la función hacia la parte del programa donde fue invocada.

````python
def sumar(x,y):
    return x + y

print(sumar(5, 3))
````

En Python podemos tener funciones que retornen varios valores.  En ese caso los valores retornados se tratarán como una Tupla.  Miremos:

````python
def mi_funcion():
    return 1 , "a", 3.5, [3,5,7]

x = mi_funcion()

print(x)
print(type(x))
````

Al ser una tupla, podemos acceder a los valores devueltos a partir de su posición e incluso podemos realizar cualquier operación permitida para las tuplas..

## Ámbito de las variables en una función

Con relación a las variables, los cambios en las variables al interior de una función no tienen efecto por fuera de ella.  Veamos

````python
def cambio_valor():
    x = 98

x = 8
cambio_valor()
print(x)
x= 13
cambio_valor()
print(x)
````

Declaración de variables globales



```
global
```

Sin embargo, con la declaración de una variable al interior de una función como **global** su valor se conservará por fuera de la función.  Veamos

````python
def cambio_valor():
    global x
    x = 10

x= 8
cambio_valor()
print(x)
````

ELEMPLO ALACANCIA

````python
total = 0

def alcancia(moneda):
    global total
    total = total + moneda
    return total 

print(alcancia(10))
print(alcancia(20))
print(alcancia(5))
````

## Funciones recursivas

Una función recursiva es aquella que en su interior se invoca a sí misma.

````python
def factorial_recursivo(n):
    if  n == 1:
        return 1
    else:
        return n * factorial_recursivo(n-1) #  4 * 4-1= 3 => 4 * 3  
 
num = int(input("INgrese el numero:"))
print(f"EL factorial de {num} es {factorial_recursivo(num)}") 
````

````python
def atras(num):
    print(num)
    num -= 1
    if num > 0:
        atras(num)

atras(10)
````

1. Realiza una función que tome una lista de números enteros y devuelva dos listas ordenadas. La primera con los números pares y la segunda con los números impares..

````python
# Definimos una función llamada 'separar_pares_impares' que recibe una lista de números enteros
def separar_pares_impares(lista_numeros):
    # Creamos una lista vacía para guardar los números pares
    pares = []
    # Creamos otra lista vacía para guardar los números impares
    impares = []
    
    # Recorremos cada número de la lista que se recibe como parámetro
    for numero in lista_numeros:
        # Si el número es divisible entre 2 (es par)
        if numero % 2 == 0:
            # Lo agregamos a la lista de pares
            pares.append(numero)
        else:
            # Si no es par, lo agregamos a la lista de impares
            impares.append(numero)
    
    # Ordenamos la lista de pares de menor a mayor
    pares.sort()
    # Ordenamos la lista de impares de menor a mayor
    impares.sort()
    
    # Devolvemos las dos listas (pares e impares) ya ordenadas
    return pares, impares

# Creamos una lista de ejemplo con varios números
numeros = [10, 3, 4, 7, 8, 2, 5, 6, 1, 9]

# Llamamos la función con nuestra lista y guardamos el resultado en dos variables
lista_pares, lista_impares = separar_pares_impares(numeros)

# Mostramos la lista de números pares ya ordenada
print("Lista de números pares ordenados:", lista_pares)

# Mostramos la lista de números impares ya ordenada
print("Lista de números impares ordenados:", lista_impares)
````

2. Realiza una función que reciba una palabra y devuelva un diccionario donde las claves sean las diferentes letras que se encuentran en la palabra y los valores sean el número de veces que se repite cada letra en la palabra

#palabra = "colombia"    
#{"c":1, "o":2, "l":1, "m":1, "b":1, "i":1, "a":1}

````python
# Definimos una función llamada 'contar_letras' que recibe una palabra como parámetro
def contar_letras(palabra):
    # Creamos un diccionario vacío para guardar las letras y sus repeticiones
    letras_contadas = {}
    
    # Recorremos cada letra de la palabra
    for letra in palabra:
        # Verificamos si la letra ya está en el diccionario
        if letra in letras_contadas:
            # Si ya está, aumentamos su valor en 1
            letras_contadas[letra] += 1
        else:
            # Si no está, la agregamos con un valor de 1
            letras_contadas[letra] = 1
    
    # Devolvemos el diccionario con el conteo de letras
    return letras_contadas

# Ejemplo de uso con la palabra "colombia"
palabra = input("Ingrese una palabra: ")

# Llamamos la función y guardamos el resultado en una variable
resultado = contar_letras(palabra)

# Mostramos el resultado en pantalla
print(resultado)
````

3. Realiza una función que reciba un número y devuelva una cadena con los nombres de los números recibidos, separando cada número con un guión medio.  Por ejemplo, si el número recibido es 134, la función devolverá la cadena "uno - tres - cuatro"

````python
# Definimos una función llamada 'numero_a_palabras' que recibe un número entero
def numero_a_palabras(numero):
    # Creamos un diccionario donde cada dígito se asocia con su nombre en palabras
    numeros_texto = {
        "0": "cero",
        "1": "uno",
        "2": "dos",
        "3": "tres",
        "4": "cuatro",
        "5": "cinco",
        "6": "seis",
        "7": "siete",
        "8": "ocho",
        "9": "nueve"
    }

    # Convertimos el número recibido a cadena para poder recorrerlo dígito por dígito
    numero_str = str(numero)

    # Creamos una lista para guardar los nombres de cada dígito
    palabras = []

    # Recorremos cada carácter (dígito) en el número convertido a string
    for digito in numero_str:
        # Buscamos el nombre del dígito en el diccionario y lo agregamos a la lista
        palabras.append(numeros_texto[digito])

    # Unimos los nombres con un guión medio y espacios
    resultado = " - ".join(palabras)

    # Devolvemos el resultado final
    return resultado

# Ejemplo de uso
numero = int(input("Ingrese un número: "));

# Llamamos la función y guardamos el resultado
resultado = numero_a_palabras(numero)

# Mostramos el resultado
print(resultado)
````
