2. En una escuela se permite la inscripción de sus alumnos a las siguientes actividades deportivas: futbol (f), baloncesto (b)., ambos (a)., Realice un programa que permita ingresar el nombre de sus estudiantes e inscribir a cada uno en uno o los dos deportes.
La aplicación debe mostrar los siguientes listados:

Estudiantes inscritos en fulbol
Estudiantes inscritos en baloncesto
Estudiantes inscritos en ambos deportes
Todos los estudiantes inscritos en algún deporte
Estudiantes inscritos en solo un deporte.
Genere un menú con las opciones de la aplicación



```python
futbol = []
baloncesto = []

while True:

    print("1. Inscribir estudiante")
    print("2. Mostrar estudiantes inscritos en fútbol")
    print("3. Mostrar estudiantes inscritos en baloncesto")
    print("4. Mostrar estudiantes inscritos en ambos deportes")
    print("5. Mostrar todos los estudiantes inscritos")
    print("6. Mostrar estudiantes inscritos en solo un deporte")
    print("7. Salir del programa")

    opcion = input("Ingrese una opción: ")

    if opcion == '1':
        nombre = input("Ingrese el nombre del estudiante: ")
        deportes = input("Ingrese los deportes (f para fútbol, b para baloncesto, o ambas): ")
        if 'f' == deportes:
            futbol.append(nombre)
        if 'b' == deportes:
            baloncesto.append(nombre)
    elif opcion == '2':
        print("Estudiantes inscritos en fútbol:", futbol)
    elif opcion == '3':
        print("Estudiantes inscritos en baloncesto:", baloncesto)
    elif opcion == '4':
        print("Estudiantes inscritos en ambos deportes:", futbol.intersection(baloncesto))
    elif opcion == '5':
        print("Todos los estudiantes inscritos:", futbol.union(baloncesto))
    elif opcion == '6':
        solo_futbol = futbol.difference(baloncesto)
        solo_baloncesto = baloncesto.difference(futbol)
        print("Estudiantes inscritos en solo un deporte:", solo_futbol.union(solo_baloncesto))
    elif opcion == '7':
        break
    else:
        print("La opción es inválida.")
```

1. Realice un programa traductor de frutas del español al inglés. Para ello, solicite el nombre de una fruta al usuario y muestre su traducción al inglés. Si la fruta no se encuentra en el diccionario, permítale al usuario agregarla con su respectiva traducción. El programa solo termina cuando el usuario ingresa una fruta que no está en el diccionario y no desea agregarla.


```python
fruta = input("ingrese una fruta en español: ")
frutas = {"fresa":"strawberry","uva":"grape","cereza":"cherry"};
if fruta in frutas:
    print(f"la traducción de {fruta} es {frutas[fruta]}")
else:
  traduccion = input("ingrese la traducción en inglés: ")
  frutas[fruta] = traduccion
  print(f"la traducción de {fruta} es {frutas[fruta]}")
```
