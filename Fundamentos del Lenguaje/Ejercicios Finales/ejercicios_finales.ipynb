## Ejercicio 1: Calculadora Avanzada

**Descripción:**  
Crea una función que actúe como calculadora. La función recibirá dos números y un operador (por ejemplo, '+', '-', '*', '/', '**', '//') para realizar la operación correspondiente.  
- Utiliza estructuras de control (if/else) para seleccionar la operación a realizar.  
- Usa manejo de excepciones para controlar errores como la división por cero y entradas no numéricas.  
- Puedes importar el módulo `math` para operaciones adicionales si es necesario.

**Ejemplo de salida:**  

Ingrese el primer número: 10
Ingrese el segundo número: 5
Ingrese el operador (+, -, *, /, **, //): /
Resultado: 10 / 5 = 2.0

```python
def calculator():
    try:
        number1 = float(input("Ingrese el primer número: "))
        number2 = float(input("Ingrese el segundo número: "))
        operator = input("Ingrese el operador (+, -, *, /, **, //): ")


        if operator == '+':
            result = number1 + number2
            
        elif operator == '-':
            result = number1 - number2
            
        elif operator == '*':
            result = number1 * number2
            
        elif operator == '/':
            if number2 == 0:
                raise ZeroDivisionError
            result = number1 / number2
            
        elif operator == '**':
            result = number1 ** number2
            
        elif operator == '//':
            if number2 == 0:
                raise ZeroDivisionError
            result = number1 // number2
            
        else:
            print("Operación no válida")
            return

        print(f"Resultado: {number1} {operator} {number2} = {result}")

    except ValueError:
        print("Error: Número no válido")
        
    except ZeroDivisionError:
        print("Error: División por cero")

calculator()
```

## Ejercicio 2: Administrador de Inventarios

**Descripción:**  
Desarrolla un sistema para gestionar el inventario de productos usando un diccionario donde cada clave es el nombre del producto y su valor la cantidad en stock.  
- Implementa funciones para agregar, actualizar y eliminar productos.  
- Utiliza estructuras de control para recorrer y mostrar el inventario.  
- Incluye manejo de excepciones para capturar intentos de actualizar o eliminar un producto inexistente.

**Ejemplo de salida:**  
Inventario inicial: {'manzanas': 50, 'naranjas': 30, 'peras': 20}
Actualizando stock de 'peras' a 25...
Producto 'bananas' agregado con 40 unidades.
Eliminando 'naranjas'...
Inventario final: {'manzanas': 50, 'peras': 25, 'bananas': 40}

```python
initial_inventory = {"manzanas": 50, "naranjas": 30, "peras": 20}

def add_product(inventory, key, value):
    inventory[key] = value
    print(f"Producto '{key}' agregado con {value} unidades")

def update_product(inventory, key, value):
    try:
        if key in inventory:
            inventory[key] = value
            print(f"Stock de '{key}' actualizado a {value}")
        else:
            raise KeyError
    except KeyError:
        print(f"Error: El producto '{key}' no existe en el inventario")

def delete_product(inventory, key):
    try:
        del(inventory[key])
        print(f"Producto '{key}' eliminado")
    except KeyError:
        print(f"Error: El producto '{key}' no existe en el inventario")

while True:
    try:
        option = int(input("Elige una opción a realizar en un producto: 1. Agregar 2. Actualizar 3. Eliminar 4. Salir (1,2,3,4)"))
        
        if option == 1:
            key = input("Nombre del nuevo producto: ")
            value = int(input("Cantidad del nuevo producto: "))
            add_product(initial_inventory, key, value)
            print(f"Inventario final: {initial_inventory}")
        
        elif option == 2:
            key = input("Nombre del producto que será actualizado: ")
            value = int(input("Cantidad del producto que será actualizado "))
            update_product(initial_inventory, key, value)
            print(f"Inventario final: {initial_inventory}")
            
        elif option == 3:
            key = input("Nombre del producto que será eliminado: ")
            delete_product(initial_inventory, key)
            print(f"Inventario final: {initial_inventory}")
        
        elif option == 4:
            print(f"Inventario final: {initial_inventory}")
            print("Saliendo del programa")
            break
        
        else:
            print("Opción no válida.")
            
    except ValueError:
        print("Error: Ingresa un número válido.")

```

## Ejercicio 3: Análisis de Texto

**Descripción:**  
Crea una función que reciba una cadena de texto y calcule la frecuencia de cada palabra en el mismo.  
- Convierte el texto a minúsculas y separa las palabras usando algún método (puedes utilizar expresiones regulares con el módulo `re`).  
- Emplea un diccionario para almacenar la palabra (clave) y su frecuencia (valor).  
- Controla excepciones en caso de que el usuario no ingrese ningún texto o ingrese caracteres inesperados.

**Ejemplo de salida:**  
Ingrese un texto: "Hola mundo, hola a todos en el mundo"
Frecuencia de palabras: hola: 2
mundo: 2
a: 1
todos: 1
en: 1
el: 1

```python
import re

def count_word_frequency(text):
    try:
        if not text.strip():
            raise ValueError("No se ingresó ningún texto.")

        text = text.lower()
        word_list = re.findall(r'\b\w+\b', text)
        word_frequency = {}

        for word in word_list:
            if word in word_frequency:
                word_frequency[word] += 1
            else:
                word_frequency[word] = 1

        return word_frequency

    except ValueError as error:
        print(f"Error: {error}")
        
    except Exception as error:
        print(f"Ocurrió un error inesperado: {error}")

entered_text = input("Ingresa una cadena de texto: ")
frequency_result = count_word_frequency(entered_text)

if frequency_result:
    print("Frecuencia de palabras:")
    for word, frequency in frequency_result.items():
        print(f"{word}: {frequency}")

```

## Ejercicio 4: Conversor de Unidades

**Descripción:**  
Desarrolla un conjunto de funciones para convertir unidades métricas. Por ejemplo:  
- De centímetros a pulgadas.  
- De kilómetros a millas.  
Utiliza estructuras de control para verificar que la entrada sea numérica, y maneja excepciones cuando el usuario ingrese datos inválidos.

**Ejemplo de salida:**  
Seleccione la conversión:

Centímetros a pulgadas

Kilómetros a millas
Opción: 1
Ingrese la cantidad en centímetros: 100
Resultado: 100 cm = 39.37 pulgadas

```python
def centimeters_to_inches(centimeters):
    result1 = centimeters / 2.54
    return result1

def kilometers_to_miles(kilometers):
    result2 = kilometers * 0.621371
    return result2

while True:
    print("Conversión de unidades métricas")
    print("1. De centímetros a pulgadas 2. De kilómetros a millas 3. Salir")
    try:
        conversion = int(input("Seleccione la conversión: (1,2,3): "))
        
        if conversion == 1:
            centimeters = float(input("Ingrese la cantidad en centímetros: "))
            result = centimeters_to_inches(centimeters)
            print(f"Resultado: {centimeters} cm = {result:.2f} pulgadas")

        elif conversion == 2:
            kilometers = float(input("Ingrese la cantidad en kilómetros: "))
            result = kilometers_to_miles(kilometers)
            print(f"Resultado: {kilometers} km = {result:.2f} millas")

        elif conversion == 3:
            print("Saliendo del programa")
            break

        else:
            print("Opción inválida. Por favor selecciona 1, 2 o 3.")

    except ValueError:
        print("Error: Ingrese un número válido.")
```

## Ejercicio 6: Calculadora de Factorial

**Descripción:**  
Crea una función que calcule el factorial de un número entero ingresado por el usuario.  
- La función debe utilizar estructuras de control para iterar (o una solución recursiva) y calcular el factorial.  
- Implementa manejo de excepciones para asegurarte de que el número ingresado es entero y no negativo.

**Ejemplo de salida:**  
Ingrese un número entero para calcular su factorial: 5
El factorial de 5 es 120

```python
def factorial(number):
    factorial_result = 1
    for i in range(1, number + 1):
        factorial_result *= i
    return factorial_result

while True:
    try:
        chosen_number = int(input("Ingrese un número entero para calcular su factorial: "))
        if chosen_number < 0:
            print("Error: El número debe ser entero y no negativo.")
        else:
            break
    except ValueError:
        print("Error: Debes ingresar un número entero válido.")

result = factorial(chosen_number)
print(f"El factorial de {chosen_number} es {result}")
```

## Ejercicio 7: Manejo de Fechas y Horas

**Descripción:**  
Desarrolla una función que reciba una fecha ingresada por el usuario en formato "DD/MM/AAAA" y determine cuántos días faltan para esa fecha a partir de hoy.  
- Importa el módulo `datetime` para trabajar con fechas.  
- Utiliza estructuras de control para validar el formato y maneja excepciones en caso de error en el ingreso de la fecha.

**Ejemplo de salida:**  
Ingrese una fecha (DD/MM/AAAA): 25/12/2023
Faltan 150 días para el 25/12/2023.

```python
from datetime import *

def calcular_days_left(user_date):
    try:
        target_date = datetime.strptime(user_date, "%d/%m/%Y")
        current_date = datetime.today()
        
        if target_date < current_date:
            print(f"La fecha {user_date} ya ha pasado.")
        else:
            difference = target_date - current_date
            print(f"Faltan {difference.days} días para el {user_date}.")
    
    except ValueError:
        print("Error: Ingresa la fecha en el formato correcto (DD/MM/AAAA).")

entered_date = input("Ingrese una fecha (DD/MM/AAAA): ")
calcular_days_left(entered_date)

```

## Ejercicio 8: Registro de Estudiantes

**Descripción:**  
Implementa un sistema que gestione un registro de estudiantes utilizando un diccionario, donde la clave sea el nombre y el valor la calificación.  
- Crea funciones para agregar un estudiante, actualizar la calificación, eliminar estudiantes y listar todo el registro.  
- Utiliza un bucle para mostrar un menú interactivo al usuario.  
- Maneja excepciones para gestionar errores en la selección de opciones o al intentar modificar un estudiante inexistente.

**Ejemplo de salida:**  
Registro de estudiantes inicial: {'Ana': 90, 'Luis': 78, 'Carlos': 85}

Menú:

Agregar estudiante

Actualizar calificación

Eliminar estudiante

Listar estudiantes

Salir

Opción: 2
Ingrese el nombre del estudiante: Luis
Ingrese la nueva calificación: 82
Estudiante actualizado: Luis - 82

Registro final: {'Ana': 90, 'Luis': 82, 'Carlos': 85}

```python
record = {"Ana": 90, "Luis": 78, "Carlos": 85}

def add_student():
    student_name = input("Ingrese el nombre del estudiante: ")
    try:
        student_grade = int(input("Ingrese la calificación: "))
        record[student_name] = student_grade
        print(f"Estudiante agregado: {student_name} - {student_grade}")
    except ValueError:
        print("Error: Ingresa una calificación válida.")

def update_grade():
    student_name = input("Ingrese el nombre del estudiante: ")
    if student_name in record:
        try:
            student_grade = int(input("Ingrese la nueva calificación: "))
            record[student_name] = student_grade
            print(f"Estudiante actualizado: {student_name} - {student_grade}")
        except ValueError:
            print("Error: Ingresa una calificación válida.")
    else:
        print("El estudiante no existe.")

def delete_student():
    student_name = input("Ingrese el nombre del estudiante a eliminar: ")
    if student_name in record:
        del record[student_name]
        print(f"Estudiante eliminado: {student_name}")
    else:
        print("El estudiante no existe.")

def list_students():
    if record:
        print("Registro actual de estudiantes:")
        for student_name, student_grade in record.items():
            print(f"{student_name}: {student_grade}")
    else:
        print("No hay estudiantes registrados.")

while True:
    print("\nMenú:")
    print("1. Agregar estudiante")
    print("2. Actualizar calificación")
    print("3. Eliminar estudiante")
    print("4. Listar estudiantes")
    print("5. Salir")

    try:
        option = int(input("Opción: "))
        if option == 1:
            add_student()
        elif option == 2:
            update_grade()
        elif option == 3:
            delete_student()
        elif option == 4:
            list_students()
        elif option == 5:
            print("Registro final:", record)
            break
        else:
            print("Opción inválida.")
    except ValueError:
        print("Error: Ingresa un número válido.")
```

## Ejercicio 9: Generador de Contraseñas Aleatorias

**Descripción:**  
Crea una función que genere una contraseña aleatoria de una longitud especificada por el usuario.  
- Utiliza el módulo `random` junto con la biblioteca `string` para construir la contraseña a partir de letras (mayúsculas y minúsculas), dígitos y caracteres especiales.  
- Valida la entrada y maneja excepciones para asegurarte de que la longitud es un número entero positivo.

**Ejemplo de salida:**  
Ingrese la longitud deseada para la contraseña: 10
Contraseña generada: A8b#K3d!Qz

```python
import random, string

def generator(length):
    characters = list(string.ascii_letters) + list(string.digits) + list(string.punctuation)
    password = []
    for i in range(length):
        choice = random.choice(characters)
        password.append(choice)
    joined_password = "".join(password)
    return joined_password

while True:
    try:
        password_length = int(input("Ingrese la longitud deseada para la contraseña: "))
        if password_length > 0:
            new_password = generator(password_length)
            print(f"Contraseña generada: {new_password}")
            break
        else:
            print("La longitud debe ser un número entero positivo.")
    except ValueError:
        print("Por favor, ingrese un número entero válido.")
```

## Ejercicio 10: Buscador de Archivos en un Directorio

**Descripción:**  
Desarrolla una función que reciba el nombre de un directorio y una subcadena para buscar archivos cuyo nombre la contenga.  
- Utiliza el módulo `os` para listar el contenido del directorio.  
- Aplica estructuras de control para filtrar los archivos según la subcadena.  
- Implementa manejo de excepciones para controlar errores como directorio no existente o permisos insuficientes.

**Ejemplo de salida:**  
Ingrese la ruta del directorio: /ruta/al/directorio
Ingrese la subcadena a buscar: reporte
Archivos encontrados que contienen "reporte":

reporte_enero.pdf

reporte_febrero.xlsx

resumen_reporte.txt

```python
import os

def search_files(directory, substring):
    try:
        files = os.listdir(directory)
        found = []

        for file in files:
            if os.path.isfile(os.path.join(directory, file)) and substring in file:
                found.append(file)

        if found:
            print(f'Archivos encontrados que contienen "{substring}":')
            for file in found:
                print(file)
        else:
            print(f'No se encontraron archivos que contengan "{substring}".')

    except FileNotFoundError:
        print("Error: El directorio no existe.")
    except PermissionError:
        print("Error: No tienes permiso para acceder a este directorio.")
    except Exception as error:
        print(f"Ocurrió un error inesperado: {error}")

user_directory = input("Ingrese la ruta del directorio: ")
substring_to_search = input("Ingrese la subcadena a buscar: ")

search_files(user_directory, substring_to_search)
```
