**Descripción:**

## Ejercicio 2: Administrador de Inventarios

**Descripción:**  
Desarrolla un sistema para gestionar el inventario de productos usando un diccionario donde cada clave es el nombre del producto y su valor la cantidad en stock.  
- Implementa funciones para agregar, actualizar y eliminar productos.  
- Utiliza estructuras de control para recorrer y mostrar el inventario.  
- Incluye manejo de excepciones para capturar intentos de actualizar o eliminar un producto inexistente.

**Ejemplo de salida:**  
Inventario inicial: {'manzanas': 50, 'naranjas': 30, 'peras': 20}
Actualizando stock de 'peras' a 25...
Producto 'bananas' agregado con 40 unidades.
Eliminando 'naranjas'...
Inventario final: {'manzanas': 50, 'peras': 25, 'bananas': 40}

```python
#Descripción:
#Desarrolla un sistema para gestionar el inventario de productos usando un diccionario donde cada clave es el nombre del producto y su valor la cantidad en stock.

#Implementa funciones para agregar, actualizar y eliminar productos.
#Utiliza estructuras de control para recorrer y mostrar el inventario.
#Incluye manejo de excepciones para capturar intentos de actualizar o eliminar un producto inexistente.
#Ejemplo de salida:
#Inventario inicial: {'manzanas': 50, 'naranjas': 30, 'peras': 20} Actualizando stock de 'peras' a 25... Producto 'bananas' agregado con 40 unidades. Eliminando 'naranjas'... Inventario final: {'manzanas': 50, 'peras': 25, 'bananas': 40}

Inventario_inicial = {'manzanas': 50, 'naranjas': 30, 'peras': 20}

def inventario():
    print("Inventario inicial:")
    print(Inventario_inicial)

    while True:
        print("\n¿Qué desea realizar?\n 1. Agregar Producto\n 2. Actualizar un producto\n 3. Eliminar un producto\n 4. Salir")
        try:
            numero = int(input("Ingrese una opción: "))
        except ValueError:
            print("Por favor, ingrese un número válido.")
            continue

        if numero == 1:
            producto = input("Nombre del producto a agregar: ")
            try:
                cantidad = int(input("Cantidad del producto: "))
                Inventario_inicial[producto] = cantidad
                print(f"Producto '{producto}' agregado con {cantidad} unidades.")
            except ValueError:
                print("La cantidad debe ser un número entero.")

        elif numero == 2:
            producto_actualizar = input("Nombre del producto a actualizar: ")
            if producto_actualizar in Inventario_inicial:
                try:
                    cantidad_nueva = int(input("Nueva cantidad: "))
                    Inventario_inicial[producto_actualizar] = cantidad_nueva
                    print(f"Stock de '{producto_actualizar}' actualizado a {cantidad_nueva}.")
                except ValueError:
                    print("La cantidad debe ser un número entero.")
            else:
                print(f"Error: El producto '{producto_actualizar}' no existe en el inventario.")

        elif numero == 3:
            producto_eliminar = input("Nombre del producto a eliminar: ")
            try:
                del Inventario_inicial[producto_eliminar]
                print(f"Producto '{producto_eliminar}' eliminado del inventario.")
            except KeyError:
                print(f"Error: El producto '{producto_eliminar}' no existe en el inventario.")

        elif numero == 4:
            print("Inventario final:", Inventario_inicial)
            break

        else:
            print("Opción no válida. Intente de nuevo.")


inventario()
```

```python
#Ejercicio 3: Análisis de Texto
#Descripción:
#Crea una función que reciba una cadena de texto y calcule la frecuencia de cada palabra en el mismo.

#Convierte el texto a minúsculas y separa las palabras usando algún método (puedes utilizar expresiones regulares con el módulo re).
#Emplea un diccionario para almacenar la palabra (clave) y su frecuencia (valor).
#Controla excepciones en caso de que el usuario no ingrese ningún texto o ingrese caracteres inesperados.
#Ejemplo de salida:
#Ingrese un texto: "Hola mundo, hola a todos en el mundo" Frecuencia de palabras: hola: 2 mundo: 2 a: 1 todos: 1 en: 1 el: 1
import re 

def conteo_palabras ():
    texto = input("Deme la cadena de texto").lower()
    busqueda = re.findall(r'\w+', texto)
    
    diccionario = {}
    
    for i in busqueda:
        if i in diccionario:
            diccionario[i] += 1
        else:
            diccionario[i] = 1
            
    print("Frecuencia de palabras:")
    for palabra, frecuencia in diccionario.items():  
        print(f"{palabra}: {frecuencia}")
    
conteo_palabras()
```

## Ejercicio 7: Manejo de Fechas y Horas

**Descripción:**  
Desarrolla una función que reciba una fecha ingresada por el usuario en formato "DD/MM/AAAA" y determine cuántos días faltan para esa fecha a partir de hoy.  
- Importa el módulo `datetime` para trabajar con fechas.  
- Utiliza estructuras de control para validar el formato y maneja excepciones en caso de error en el ingreso de la fecha.

**Ejemplo de salida:**  
Ingrese una fecha (DD/MM/AAAA): 25/12/2023
Faltan 150 días para el 25/12/2023.

```python
#Ejercicio 7: Manejo de Fechas y Horas
#Descripción:
#Desarrolla una función que reciba una fecha ingresada por el usuario en formato "DD/MM/AAAA" y determine cuántos días faltan para esa fecha a partir de hoy.

#Importa el módulo datetime para trabajar con fechas.s
#Utiliza estructuras de control para validar el formato y maneja excepciones en caso de error en el ingreso de la fecha.
#Ejemplo de salida:
#Ingrese una fecha (DD/MM/AAAA): 25/12/2023 Faltan 150 días para el 25/12/2023.

import datetime

def calcular_dias_faltantes():
    try:
        fecha_usuario = input("Ingrese una fecha (DD/MM/AAAA): ")
        fecha_objetivo = datetime.datetime.strptime(fecha_usuario, "%d/%m/%Y")
        
        hoy = datetime.datetime.now()
        
        diferencia = fecha_objetivo - hoy
        
    
        dias_faltantes = diferencia.days
        
        if dias_faltantes > 0:
            print(f"Faltan {dias_faltantes} días para el {fecha_usuario}.")
        elif dias_faltantes == 0:
            print(f"¡Hoy es {fecha_usuario}!")
        else:
            print(f"La fecha {fecha_usuario} ya pasó hace {-dias_faltantes} días.")

    except ValueError:
        print("Formato de fecha incorrecto. Por favor ingrese la fecha como DD/MM/AAAA.")


calcular_dias_faltantes()
```

## Ejercicio 9: Generador de Contraseñas Aleatorias

**Descripción:**  
Crea una función que genere una contraseña aleatoria de una longitud especificada por el usuario.  
- Utiliza el módulo `random` junto con la biblioteca `string` para construir la contraseña a partir de letras (mayúsculas y minúsculas), dígitos y caracteres especiales.  
- Valida la entrada y maneja excepciones para asegurarte de que la longitud es un número entero positivo.

**Ejemplo de salida:**  
Ingrese la longitud deseada para la contraseña: 10
Contraseña generada: A8b#K3d!Qz

```python
import random
import string

def generar_contraseña(longitud):
    if not isinstance(longitud, int) or longitud <= 0:
        raise ValueError("La longitud debe ser un número entero positivo.")

    caracteres = string.ascii_letters + string.digits + string.punctuation
    for i in range(longitud):
        contraseña += random.choice(caracteres)
    return contraseña

def contraseña():
    try:
        entrada = input("Ingrese la longitud deseada para la contraseña: ")
        longitud = int(entrada)
        contraseña = generar_contraseña(longitud)
        print(f"Contraseña generada: {contraseña}")
    except ValueError:
        print("Por favor, ingrese un número entero positivo válido.")
    except Exception as error:
        print(f"Ocurrió un error: {error}")

contraseña()
```

## Ejercicio 10: Buscador de Archivos en un Directorio

**Descripción:**  
Desarrolla una función que reciba el nombre de un directorio y una subcadena para buscar archivos cuyo nombre la contenga.  
- Utiliza el módulo `os` para listar el contenido del directorio.  
- Aplica estructuras de control para filtrar los archivos según la subcadena.  
- Implementa manejo de excepciones para controlar errores como directorio no existente o permisos insuficientes.

**Ejemplo de salida:**  
Ingrese la ruta del directorio: /ruta/al/directorio
Ingrese la subcadena a buscar: reporte
Archivos encontrados que contienen "reporte":

reporte_enero.pdf

reporte_febrero.xlsx

resumen_reporte.txt

```python
import os

def buscar_archivos(directorio, subcadena):
    try:
        archivos_encontrados = []

        # Verifica que el directorio exista y sea accesible
        if not os.path.isdir(directorio):
            raise FileNotFoundError(f"El directorio '{directorio}' no existe.")

        for archivo in os.listdir(directorio):
            ruta_completa = os.path.join(directorio, archivo)
            if os.path.isfile(ruta_completa) and subcadena in archivo:
                archivos_encontrados.append(archivo)

        return archivos_encontrados

    except PermissionError:
        print("Permiso denegado para acceder al directorio.")
    except FileNotFoundError as e:
        print(e)
    except Exception as e:
        print(f"Ocurrió un error inesperado: {e}")

def ruta():
    directorio = input("Ingrese la ruta del directorio: ")
    subcadena = input("Ingrese la subcadena a buscar: ")

    archivos = buscar_archivos(directorio, subcadena)

    if archivos:
        print(f'\nArchivos encontrados que contienen "{subcadena}":\n')
        for archivo in archivos:
            print(archivo)
    else:
        print(f'\nNo se encontraron archivos que contengan "{subcadena}".')


ruta()
```
