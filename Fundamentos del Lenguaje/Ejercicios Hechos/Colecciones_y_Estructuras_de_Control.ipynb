1. Gestión de Inventario de Productos
Descripción:
Se debe desarrollar un sistema que administre un inventario compuesto por productos y sus cantidades. El programa presentará el listado inicial y resaltará aquellos artículos cuyo stock esté por debajo de un umbral definido. Luego, permitirá actualizar la cantidad de un producto específico, mostrando el inventario antes y después de la modificación.

```python
inventario = {
    "producto1": 20,
    "producto2": 5,
    "producto3": 15,
    "producto4": 3
}

umbral = 10

print("Inventario inicial:")
for producto, cantidad in inventario.items():
    estado = "STOCK BAJO" if cantidad < umbral else ""
    print(f"{producto}: {cantidad} {estado}")

producto_a_actualizar = "producto2"
nueva_cantidad = 12
print(f"Inventario antes de la actualización: {producto_a_actualizar} = {inventario[producto_a_actualizar]}")
inventario[producto_a_actualizar] = nueva_cantidad
print(f"Inventario después de la actualización: {producto_a_actualizar} = {inventario[producto_a_actualizar]}")
```

2. Búsqueda de Palabras en un Texto
Descripción:
Crea una aplicación que solicite al usuario un párrafo y, a continuación, le permita buscar una palabra específica. El sistema debe contar cuántas veces aparece dicha palabra en el texto, mostrar las posiciones donde se encuentra (sin distinguir mayúsculas de minúsculas) y comunicar el resultado al usuario.



```python
texto = input("Ingrese un párrafo: ")
palabra_a_buscar = input("Ingrese la palabra a buscar: ").lower()

contador = 0
posiciones = []
for i, palabra in enumerate(texto.lower().split()):
    if palabra == palabra_a_buscar:
        contador += 1
        posiciones.append(i + 1)

print(f"La palabra '{palabra_a_buscar}' se encuentra {contador} veces en el texto.")
if contador > 0:
    print(f"Posiciones: {', '.join(map(str, posiciones))}")
```

3. Conversión y Limpieza de Texto
Descripción:
Diseña una solución que pida al usuario una frase, la divida en palabras y elimine cualquier carácter no alfabético de cada una. Una vez procesada la información, la estructura resultante se transformará en una forma inmutable, y se presentará el contenido final al usuario de manera ordenada.

```python
frase = input("Ingrese una frase: ")

palabras_limpias = [''.join(filter(str.isalpha, palabra)) for palabra in frase.split()]

palabras_inmutables = tuple(sorted(palabras_limpias))

print("Contenido final ordenado:", palabras_inmutables)
```

4. Registro y Consulta de Calificaciones de Exámenes
Descripción:
Implementa un sistema que gestione un registro de estudiantes y sus calificaciones. El programa debe permitir agregar nuevos registros, listar todos los existentes y consultar la calificación de un estudiante introduciendo su nombre. Se espera que la información se actualice y se presente de forma clara en cada paso.

```python
calificaciones = {}

print("\nMenú:")
print("1. Agregar estudiante")
print("2. Listar estudiantes")
print("3. Consultar calificación")
print("4. Salir")

while True:
    opcion = int(input("Ingrese una opción: "))

    if opcion == 1:
        estudiante = input("Ingrese el nombre del estudiante: ")
        calificacion = int(input("Ingrese la calificación del estudiante: "))
        calificaciones[estudiante] = calificacion
    elif opcion == 2:
        print("\nLista de estudiantes:")
        for estudiante, calificacion in calificaciones.items():
            print(f"{estudiante}: {calificacion}")
    elif opcion == 3:
        estudiante = input("Ingrese el nombre del estudiante: ")
        if estudiante in calificaciones:
            print(f"La calificación de {estudiante} es {calificaciones[estudiante]}")
        else:
            print(f"{estudiante} no existe en la lista")
    elif opcion == 4:
        break
    else:
        print("Opción inválida. Intente nuevamente.")
```

5. Calculadora de Estadísticas de Números
Descripción:
Desarrolla una aplicación que trabaje con una colección de números predefinidos. El sistema debe ofrecer al usuario un menú interactivo con opciones para calcular el promedio, identificar el valor máximo y mínimo, o eliminar un número. Tras cada operación, se mostrará tanto el resultado como el estado actual de la colección.

```python
numeros = [10, 20, 30, 40, 50]

print("Menú:")
print("1. Calcular promedio de los numeros")
print("2. Valor max y min")
print("3. Eliminar algun numero")
print("4. Salir")

while True:
    opcion = int(input("Ingrese la opcion deseada: "))

    if opcion == 1:
        promedio = sum(numeros) / len(numeros)
        print (f"\nEl promedio de los numeros es: {promedio}")
    elif opcion == 2:
        maximo = max(numeros)
        print(f"\nEl numero maximo es: {maximo}")
        minimo = min(numeros)
        print(f"\nEl numero minimo es: {minimo}")
    elif opcion == 3:
        numero_eliminado = int(input("Escoge algun numero de la lista para eliminar: "))
        numeros.remove(numero_eliminado)
        print(f"\nEl numero {numero_eliminado} fue eliminado de la lista: {numeros}")
    elif opcion == 4:
        print(f"\nPrograma finalizado...")
        break
```
