```python
# Primer ejercicio: Ingresar una lista de elementos y mostrarla junto con su versión invertida
def ejercicio_1():
    n = int(input("¿Cuántos elementos deseas ingresar en la lista? "))
    lista = []
    for i in range(n):
        elemento = input(f"Ingrese el elemento {i+1}: ")
        lista.append(elemento)

    print("Contenido de la lista:", lista)
    print("Contenido invertido de la lista:", lista[::-1])

# Segundo ejercicio: Crear una lista de números, solicitar un número y eliminarlo de todas las posiciones
def ejercicio_2():
    lista_numeros = list(map(int, input("Ingresa los números separados por espacio: ").split()))
    numero_a_eliminar = int(input("Ingresa el número que deseas eliminar de la lista: "))

    # Eliminar el número de todas las posiciones
    lista_numeros = [num for num in lista_numeros if num != numero_a_eliminar]

    print("Lista final después de eliminar el número:", lista_numeros)

# Tercer ejercicio: Solicitar dos frases y devolver las letras repetidas en la misma posición
def ejercicio_3():
    frase1 = input("Ingresa la primera frase: ")
    frase2 = input("Ingresa la segunda frase: ")

    # Encontrar las letras que se repiten en la misma posición
    letras_comunes = [frase1[i] for i in range(min(len(frase1), len(frase2))) if frase1[i] == frase2[i]]

    print("Letras que se repiten en la misma posición:", letras_comunes)
```
