1. Realiza una función que tome una lista de números enteros y devuelva dos listas ordenadas. La primera con los números pares y la segunda con los números impares.


```python
def paresImpares(numeros):

    pares = []
    impares = []
    
    for numero in numeros:
        if numero % 2 == 0:
            pares.append(numero)
        else:
            impares.append(numero)
    
    pares.sort()
    impares.sort()
    
    return pares, impares


lista_numeros = [10, 3, 5, 8, 2, 7, 4, 1, 6, 9]
pares, impares = paresImpares(lista_numeros)
    
print("Números pares:", pares)
print("Números impares:", impares)
```

2. Realiza una función que reciba una palabra y devuelva un diccionario donde las claves sean las diferentes letras que se encuentran en la palabra y los valores sean el número de veces que se repite cada letra en la palabra

```python
def contador_letras(palabra):

    contador = {}
    
    for letra in palabra:
        if letra in contador:
            contador[letra] += 1
        else:
            contador[letra] = 1
            
    return contador

palabra = "Mundo"
resultado = contador_letras(palabra)
    
print("Conteo de letras:", resultado)
```

3. Realiza una función que reciba un número y devuelva una cadena con los nombres de los números recibidos, separando cada número con un guión medio. Por ejemplo, si el número recibido es 134, la función devolverá la cadena "uno - tres - cuatro"


```python
def numeroPalabra(numero):

    nombres_digitos = { '0': 'cero', '1': 'uno', '2': 'dos', '3': 'tres', '4': 'cuatro', '5': 'cinco', '6': 'seis', '7': 'siete', '8': 'ocho', '9': 'nueve'}

    numero_str = str(numero)
    
    # Crear una lista de nombres de los dígitos
    nombres = [nombres_digitos[digit] for digit in numero_str]

    resultado = ' - '.join(nombres)
    
    return resultado

numero = 568
resultado = numeroPalabra(numero)
    
print("Número en palabras:", resultado)
```
