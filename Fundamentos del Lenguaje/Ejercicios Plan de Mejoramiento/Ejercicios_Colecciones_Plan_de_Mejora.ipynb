Ejercicio 1: Registro de Calificaciones y Estadísticas
Descripción:
Se requiere desarrollar un programa que administre el desempeño académico de un grupo de estudiantes. Utiliza una colección para almacenar los nombres de los alumnos y sus calificaciones (cada alumno puede tener varias notas). La solución deberá calcular el promedio de cada estudiante, determinar si aprueba (por ejemplo, con un promedio mayor o igual a cierto valor) y mostrar un resumen detallado con el nombre, el promedio y el estado de cada uno.

```bash
Registro de calificaciones:
- Ana: [70, 85, 90] -> Promedio: 81.67 -> Aprobado
- Luis: [55, 60, 58] -> Promedio: 57.67 -> Reprobado
- Carlos: [80, 78, 85] -> Promedio: 81.00 -> Aprobado
```

````python
def calcular_promedio(calificaciones):
    return sum(calificaciones) / len(calificaciones)

def determinar_estado(promedio, nota_aprobatoria=6.0):
    return "Aprobado" if promedio >= nota_aprobatoria else "Reprobado"

def registrar_estudiantes():
    estudiantes = {}
    while True:
        nombre = input("Ingrese el nombre del estudiante (o 'salir' para finalizar): ")
        if nombre.lower() == 'salir':
            break
        
        calificaciones = []
        while True:
            try:
                nota = float(input(f"Ingrese una calificación para {nombre} (o 0 para terminar): "))
                if nota == 0:
                    break
                calificaciones.append(nota)
            except ValueError:
                print("Por favor ingrese una calificación válida.")
        
        estudiantes[nombre] = calificaciones

    return estudiantes

def mostrar_resumen(estudiantes):
    print("\nResumen de calificaciones:")
    for nombre, calificaciones in estudiantes.items():
        promedio = calcular_promedio(calificaciones)
        estado = determinar_estado(promedio)
        print(f"{nombre}: Promedio = {promedio:.2f} | Estado = {estado}")

def main():
    estudiantes = registrar_estudiantes()
    mostrar_resumen(estudiantes)

main()
````

Ejercicio 2: Análisis de Inventario con Diccionarios
Descripción:
Crea un sistema de inventario que administre productos y sus cantidades utilizando una colección que relacione cada producto con su stock. La solución debe permitir actualizar la cantidad de un producto, eliminar aquellos que se encuentren agotados (por ejemplo, cantidad igual a cero) y mostrar el inventario ordenado alfabéticamente, reflejando los cambios realizados.

```bash
Inventario inicial:
- Manzanas: 50
- Naranjas: 20
- Peras: 0
- Plátanos: 30

Inventario actualizado:
- Manzanas: 45
- Naranjas: 20
- Plátanos: 30

```

````python
def mostrar_inventario(inventario):
    print("\nInventario actual:")
    for producto, cantidad in sorted(inventario.items()):
        print(f"{producto}: {cantidad}")

def actualizar_producto(inventario):
    producto = input("Ingrese el nombre del producto a actualizar: ")
    if producto in inventario:
        try:
            nueva_cantidad = int(input(f"Ingrese la nueva cantidad para {producto}: "))
            if nueva_cantidad < 0:
                print("La cantidad no puede ser negativa.")
            else:
                inventario[producto] = nueva_cantidad
                print(f"Cantidad de {producto} actualizada a {nueva_cantidad}.")
        except ValueError:
            print("Por favor ingrese un número entero válido para la cantidad.")
    else:
        print(f"El producto {producto} no se encuentra en el inventario.")

def eliminar_producto(inventario):
    producto = input("Ingrese el nombre del producto a eliminar: ")
    if producto in inventario:
        if inventario[producto] == 0:
            del inventario[producto]
            print(f"El producto {producto} ha sido eliminado del inventario.")
        else:
            print(f"No se puede eliminar {producto} porque tiene cantidad disponible.")
    else:
        print(f"El producto {producto} no se encuentra en el inventario.")

def agregar_producto(inventario):
    producto = input("Ingrese el nombre del producto a agregar: ")
    if producto in inventario:
        print(f"El producto {producto} ya existe en el inventario.")
    else:
        try:
            cantidad = int(input(f"Ingrese la cantidad de {producto}: "))
            if cantidad < 0:
                print("La cantidad no puede ser negativa.")
            else:
                inventario[producto] = cantidad
                print(f"Producto {producto} agregado con cantidad {cantidad}.")
        except ValueError:
            print("Por favor ingrese un número entero válido para la cantidad.")

def main():
    inventario = {
        "manzanas": 50,
        "naranjas": 30,
        "peras": 20
    }
    
    while True:
        print("\nMenú de opciones:")
        print("1. Mostrar inventario")
        print("2. Actualizar producto")
        print("3. Eliminar producto")
        print("4. Agregar producto")
        print("5. Salir")
        
        try:
            opcion = int(input("Seleccione una opción (1-5): "))
            if opcion == 1:
                mostrar_inventario(inventario)
            elif opcion == 2:
                actualizar_producto(inventario)
            elif opcion == 3:
                eliminar_producto(inventario)
            elif opcion == 4:
                agregar_producto(inventario)
            elif opcion == 5:
                print("Saliendo del sistema de inventario.")
                break
            else:
                print("Opción no válida. Por favor ingrese un número entre 1 y 5.")
        except ValueError:
            print("Por favor ingrese un número entero válido para la opción.")

main()
````

Ejercicio 3: Fusión y Diferencia de Conjuntos de Hobbies
Descripción:
Se dispone de dos colecciones que contienen los hobbies de dos grupos de personas. El desafío consiste en determinar qué actividades comparten ambos grupos, identificar los hobbies exclusivos de cada uno y crear una colección que unifique todos los hobbies sin repeticiones. Esta tarea ayudará a practicar operaciones de unión, intersección y diferencia.

```bash
Hobbies del Grupo A: {'leer', 'correr', 'pintar', 'viajar'}
Hobbies del Grupo B: {'correr', 'cocinar', 'viajar', 'bailar'}

Hobbies en común: {'correr', 'viajar'}
Hobbies exclusivos del Grupo A: {'leer', 'pintar'}
Hobbies exclusivos del Grupo B: {'cocinar', 'bailar'}
Todos los hobbies: {'leer', 'correr', 'pintar', 'viajar', 'cocinar', 'bailar'}
```

````python
def analizar_hobbies(hobbies_grupo_1, hobbies_grupo_2):
    union_hobbies = hobbies_grupo_1 | hobbies_grupo_2
    interseccion_hobbies = hobbies_grupo_1 & hobbies_grupo_2
    diferencia_grupo_1 = hobbies_grupo_1 - hobbies_grupo_2
    diferencia_grupo_2 = hobbies_grupo_2 - hobbies_grupo_1

    print("Hobbies en común entre ambos grupos:", interseccion_hobbies)
    print("Hobbies exclusivos del grupo 1:", diferencia_grupo_1)
    print("Hobbies exclusivos del grupo 2:", diferencia_grupo_2)
    print("Todos los hobbies sin repeticiones:", union_hobbies)

hobbies_grupo_1 = {"leer", "fútbol", "ciclismo", "viajar", "pintura"}
hobbies_grupo_2 = {"pintura", "fútbol", "correr", "videojuegos"}

analizar_hobbies(hobbies_grupo_1, hobbies_grupo_2)
````

Ejercicio 4: Manipulación de Tuplas de Datos de Empleados
Descripción:
Diseña un programa que maneje la información de un empleado representada en una colección inmutable (por ejemplo, una tupla) que contenga datos como el nombre, la edad, el departamento y el salario. La tarea consiste en transformar esta colección en una estructura modificable, aplicar un aumento salarial (por ejemplo, del 10%) y luego volver a convertirla en una colección inmutable para mostrar los datos actualizados.

```bash
Datos iniciales del empleado: ('Carlos', 30, 'Ventas', 3000)
Datos actualizados del empleado: ('Carlos', 30, 'Ventas', 3300)
```

````python
def actualizar_salario(empleado):
    empleado_lista = list(empleado)
    salario_actualizado = empleado_lista[3] * 1.10
    empleado_lista[3] = salario_actualizado
    empleado_actualizado = tuple(empleado_lista)
    return empleado_actualizado

empleado = ("Juan Pérez", 30, "Recursos Humanos", 2500)

empleado_actualizado = actualizar_salario(empleado)

print("Empleado original:", empleado)
print("Empleado con salario actualizado:", empleado_actualizado)
````

Ejercicio 5: Filtrado de Palabras en una Frase
Descripción:
El objetivo es desarrollar una solución que solicite al usuario una oración, la divida en palabras y luego filtre aquellas que tengan una longitud mayor a 4 caracteres. La colección resultante se debe presentar de manera ordenada, demostrando la capacidad de transformar, filtrar y visualizar los elementos relevantes del texto.

```bash
Frase ingresada: "La programación en Python es muy divertida"
Palabras con más de 4 caracteres: ['programación', 'Python', 'divertida']
```

````python
def filtrar_palabras(frase):
    palabras = frase.split()
    palabras_filtradas = [palabra for palabra in palabras if len(palabra) > 4]
    return sorted(palabras_filtradas)

frase = input("Ingrese una frase: ")

resultado = filtrar_palabras(frase)

print("Palabras con más de 4 caracteres:", resultado)
````
