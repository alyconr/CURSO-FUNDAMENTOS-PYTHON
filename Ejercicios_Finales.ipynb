***Ejercicio 1: Calculadora Avanzada***



1. Crea una función que actúe como calculadora. La función recibirá dos números y un operador (por ejemplo, '+', '-', '*', '/', '**', '//') para realizar la operación correspondiente.

```python
try:
    x = float(input("Ingresa el primer número: "))
    y = float(input("Ingresa el segundo número: "))
    operacion = input("Ingresa el operador que necesitas: (+, -, *, /, **, //) ")

    if operacion == '+':
        r = x + y
    elif operacion == '-':
        r = x - y
    elif operacion == '*':
        r = x * y
    elif operacion == '/':
        r = x / y
    elif operacion == '**':
        r = x ** y
    elif operacion == '//':
        r = x // y
    else:
        print("Operacion no válida")
        exit()

    print(f"Resultado: {x} {operacion} {y} = {r}")

except ZeroDivisionError:
    print("Error: La División por cero.")
except ValueError:
    print("Error: Solo Ingrese números válidos.")
```

***Ejercicio 2: Administrador de Inventarios***

2. Desarrolla un sistema para gestionar el inventario de productos usando un diccionario donde cada clave es el nombre del producto y su valor la cantidad en stock.

```python
inventario = {'Manzanas': 50, 'Naranjas': 30, 'Peras': 20}
print("Inventario inicial:", inventario)

def agregar_producto(nombre, cantidad):
    if nombre in inventario:
        print(f"El producto '{nombre}' ya existe. Dale en actualizar para cambiar la cantidad.")
    else:
        inventario[nombre] = cantidad
        print(f"Producto '{nombre}' fue agregado con {cantidad} unidades.")

def actualizar_producto(nombre, nueva_cantidad):
    try:
        if nombre in inventario:
            inventario[nombre] = nueva_cantidad
            print(f"Actualizando inventario de '{nombre}' a {nueva_cantidad}")
        else:
            raise KeyError(f"El producto '{nombre}' no existe en el inventario.")
    except KeyError as e:
        print(f"Error: {e}")

def eliminar_producto(nombre):
    try:
        if nombre in inventario:
            del inventario[nombre]
            print(f"Eliminando '{nombre}'...")
        else:
            raise KeyError(f"El producto '{nombre}' no existe en el inventario.")
    except KeyError as e:
        print(f"Error: {e}")

actualizar_producto('Peras', 25)
agregar_producto('Bananas', 40)
eliminar_producto('Naranjas')

print("Inventario final:", inventario)
```

***Ejercicio 3: Analisis De Texto***


3. Crea una función que reciba una cadena de texto y calcule la frecuencia de cada palabra en el mismo

```python
import re

def frecuencia_palabras():
    try:
        texto = input("Ingrese un texto con palabras repetidas: ").strip()

        if not texto:
            raise ValueError("No se ingresó ningún texto.")

        palabras = re.findall(r'\b\w+\b', texto.lower())

        Frecuencia = {}

        for palabra in palabras:
            if palabra in Frecuencia:
                Frecuencia[palabra] += 1
            else:
                Frecuencia[palabra] = 1

        print("Frecuencia de palabras:")
        for palabra, cantidad in Frecuencia.items():
            print(f"{palabra}: {cantidad}")

    except ValueError as ve:
        print(f"Error: {ve}")
    except Exception as e:
        print(f"Ocurrió un error: {e}")

frecuencia_palabras()
```

***Ejercicio 4: Conversor De Unidades***

4. Desarrolla un conjunto de funciones para convertir unidades métricas

```python
def cm_a_pulgadas(cm):
    return cm / 2.54

def km_a_millas(km):
    return km * 0.621371

print("Selecciona la conversión:")
print("1. Centímetros a pulgadas:")
print("2. Kilómetros a millas:")

try:
    opcion = int(input("Opción: "))

    if opcion == 1:
        cantidad = float(input("Ingresa la cantidad en centímetros: "))
        resultado = cm_a_pulgadas(cantidad)
        print(f"Resultado: {cantidad} cm = {round(resultado, 2)} pulgadas")

    elif opcion == 2:
        cantidad = float(input("Ingresa la cantidad en kilómetros: "))
        resultado = km_a_millas(cantidad)
        print(f"Resultado: {cantidad} km = {round(resultado, 2)} millas")

    else:
        print("Opción no válida. Seleccione las opciones 1 o 2.")

except ValueError:
    print("Error: Ingrese solo números válidos.")
except Exception as e:
    print(f"Ocurrió un error: {e}")
```

***Ejercicio 5: Adivina El Numero***

5. Implementa un juego en el que la computadora selecciona aleatoriamente un número en un rango (por ejemplo, 1 a 50) y el usuario debe adivinarlo.

```python
import random
print('¡Bienvenido al juego "Adivina el Número"!')
print("Estoy pensando en un número entre 1 y 50.")

numero_secreto = random.randint(1, 50)
intento = 1

while True:
    try:
        guess = int(input(f"Intento {intento} - Ingresa tu número: "))

        if guess < 1 or guess > 50:
            print("Por favor, ingresa un número dentro del rango (1-50).")
            continue

        if guess < numero_secreto:
            print("El número secreto es mayor.")
        elif guess > numero_secreto:
            print("El número secreto es menor.")
        else:
            print(f"¡Felicidades! Has adivinado el número secreto: {numero_secreto}")
            break

        intento += 1

    except ValueError:
        print("Error: Debes ingresar un número válido.")
```

***Ejercicio 6: Calculadora Factorial***

6. Crea una función que calcule el factorial de un número entero ingresado por el usuario.

```python
def calculadora_factorial():
    try:
        num = int(input("Ingrese un número entero para calcular su factorial: "))

        if num < 0:
            raise ValueError("No, El número no puede ser negativo")

        Factorial = 1
        for i in range(1, num + 1):
            Factorial *= i

        print(f"El factorial del {num} es: {Factorial}")

    except ValueError as ve:
        print(f"Error: {ve}")
    except Exception as e:
        print(f"Ocurrió un error: {e}")

calculadora_factorial()
```

***Ejercicio 7: Manejo De Fechas Y Horas***

7. Desarrolla una función que reciba una fecha ingresada por el usuario en formato "DD/MM/AAAA" y determine cuántos días faltan para esa fecha a partir de hoy.

```python
from datetime import datetime

def dias_fecha():
    try:
        Pregunta = input("Ingrese alguna fecha de esta manera: (DD/MM/AAAA) ").strip()

        fecha_1 = datetime.strptime(Pregunta, "%d/%m/%Y").date()
        fecha_actual = datetime.today().date()

        diferencia = (fecha_1 - fecha_actual).days

        if diferencia > 0:
            print(f"Faltan {diferencia} días para el {Pregunta}.")
        elif diferencia == 0:
            print(f"¡Hoy es {Pregunta}!")
        else:
            print(f"La fecha {Pregunta} ya pasó hace {-diferencia} días.")

    except ValueError:
        print("Error: Fecha inválida.")
    except Exception as e:
        print(f"Ocurrió un error: {e}")

dias_fecha()
```

***EJERCICIO 8: Registro De Estudiantes***

8. Implementa un sistema que gestione un registro de estudiantes utilizando un diccionario, donde la clave sea el nombre y el valor la calificación.

```python
registro_estudiantes = { 'Ana': 90, 'Luis': 78, 'Carlos': 85, 'Josue': 100}

def agregar_estudiante():
    nombre = input("Ingrese el nombre del estudiante: ").strip()
    if nombre in registro_estudiantes:
        print(f"El estudiante '{nombre}' ya existe.")
        return
    try:
        calificacion = float(input("Ingrese la calificación: "))
        registro_estudiantes[nombre] = calificacion
        print(f"Estudiante agregado: {nombre} - {calificacion}")
    except ValueError:
        print("Error: La calificación debe ser un número.")

def actualizar_calificacion():
    nombre = input("Ingrese el nombre del estudiante: ").strip()
    if nombre not in registro_estudiantes:
        print(f"Error: El estudiante '{nombre}' no está registrado.")
        return
    try:
        nueva_calificacion = float(input("Ingrese la nueva calificación: "))
        registro_estudiantes[nombre] = nueva_calificacion
        print(f"Estudiante actualizado: {nombre} - {nueva_calificacion}")
    except ValueError:
        print("Error: La calificación debe ser un número.")

def eliminar_estudiante():
    nombre = input("Ingrese el nombre del estudiante a eliminar: ").strip()
    if nombre in registro_estudiantes:
        del registro_estudiantes[nombre]
        print(f"Estudiante '{nombre}' eliminado.")
    else:
        print(f"Error: El estudiante '{nombre}' no existe.")

def listar_estudiantes():
    if not registro_estudiantes:
        print("El registro está vacío.")
    else:
        print("\nRegistro actual de estudiantes:")
        for nombre, calificacion in registro_estudiantes.items():
            print(f"- {nombre}: {calificacion}")

print("Registro de estudiantes inicial:", registro_estudiantes)

while True:
    print("\nMenú:\n")
    print("1. Agregar estudiante")
    print("2. Actualizar calificación")
    print("3. Eliminar estudiante")
    print("4. Listar estudiantes")
    print("5. Salir")

    try:
        opcion = input("Opción: ").strip()

        if opcion == '1':
            agregar_estudiante()
        elif opcion == '2':
            actualizar_calificacion()
        elif opcion == '3':
            eliminar_estudiante()
        elif opcion == '4':
            listar_estudiantes()
        elif opcion == '5':
            print("Gracias, Hasta Pronto.")
            break
        else:
            print("Opción no válida. Solo números del 1 al 5.")
    except Exception as e:
        print(f"Ocurrió un error: {e}")

print("\nRegistro final:", registro_estudiantes)
```

***Ejercicio 9: Generador De Contraseñas Aleatoreas***

9. Crea una función que genere una contraseña aleatoria de una longitud especificada por el usuario.

```python

import random
import string

def Generacion_de_contraseña():
    try:
        longitud = int(input("Ingrese la longitud deseada para la contraseña: "))

        if longitud <= 0:
            raise ValueError("El número debe ser entero positivo")

        caracteres = string.ascii_letters + string.digits + string.punctuation
        contraseña = ''.join(random.choice(caracteres) for _ in range(longitud))

        print(f"Contraseña generada: {contraseña}")

    except ValueError as e:
        print(f"Error: {e}")
    except Exception as e:
        print(f"Ocurrió un error: {e}")

Generacion_de_contraseña()
```
