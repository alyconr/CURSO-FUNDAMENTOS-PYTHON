1. Gestión de Inventario de Productos Descripción: Se debe desarrollar un sistema que administre un inventario compuesto por productos y sus cantidades. El programa presentará el listado inicial y resaltará aquellos artículos cuyo stock esté por debajo de un umbral definido. Luego, permitirá actualizar la cantidad de un producto específico, mostrando el inventario antes y después de la modificación.

```python
# Inventario inicial
inventario = {"Laptop": 10, "Smartphone": 15, "Tablet": 8, "Monitor": 4, "Teclado": 20}
umbral = 5

# Mostrar inventario inicial
print("=== INVENTARIO INICIAL ===")
for item in inventario:
    estado = "BAJO" if inventario[item] < umbral else "OK"
    print(f"{item}: {inventario[item]} - {estado}")

# Actualizar producto
producto = input("\nProducto a actualizar: ")
if producto in inventario:
    antes = inventario[producto]
    inventario[producto] = int(input("Nueva cantidad: "))
    
    # Mostrar resultado
    print("\n=== INVENTARIO ACTUALIZADO ===")
    for item in inventario:
        estado = "BAJO" if inventario[item] < umbral else "OK"
        print(f"{item}: {inventario[item]} - {estado}")
    print(f"\nCambio: {producto} {antes} -> {inventario[producto]}")
else:
    print("Producto no encontrado")
```

2. Búsqueda de Palabras en un Texto Descripción: Crea una aplicación que solicite al usuario un párrafo y, a continuación, le permita buscar una palabra específica. El sistema debe contar cuántas veces aparece dicha palabra en el texto, mostrar las posiciones donde se encuentra (sin distinguir mayúsculas de minúsculas) y comunicar el resultado al usuario.

```python
# Pedir el texto y la palabra
texto = input("Escribe un párrafo: ")
palabra = input("¿Qué palabra buscas? ")

# Convertir a minúsculas
texto = texto.lower()
palabra = palabra.lower()

# Buscar apariciones
posicion = 0
posiciones = []

while posicion != -1:
    posicion = texto.find(palabra, posicion)
    if posicion != -1:
        posiciones.append(posicion)
        posicion += 1

# Mostrar los resultados
contador = len(posiciones)
print(f"\nLa palabra '{palabra}' aparece {contador} veces")
if contador > 0:
    print(f"Está en las posiciones: {posiciones}")
else:
    print("No se encontró la palabra")
```

3. Conversión y Limpieza de Texto Descripción: Diseña una solución que pida al usuario una frase, la divida en palabras y elimine cualquier carácter no alfabético de cada una. Una vez procesada la información, la estructura resultante se transformará en una forma inmutable, y se presentará el contenido final al usuario de manera ordenada.

```python
procesar_frase_lambda = lambda: (
    lambda frase_input: tuple(
        "".join(c.lower() for c in palabra if c.isalpha())
        for palabra in frase_input.split()
        if any(c.isalpha() for c in palabra)
    )
)(input("Ingrese una frase: "))

if __name__ == "__main__":
    resultado = procesar_frase_lambda()
    print("\nFrase procesada :")
    print(" ".join(resultado))
```

4. Registro y Consulta de Calificaciones de Exámenes Descripción: Implementa un sistema que gestione un registro de estudiantes y sus calificaciones. El programa debe permitir agregar nuevos registros, listar todos los existentes y consultar la calificación de un estudiante introduciendo su nombre. Se espera que la información se actualice y se presente de forma clara en cada paso.

```python
calificaciones = {}

while True:
    print("\n--- Calificaciones ---")
    print("1. Agregar")
    print("2. Ver todas")
    print("3. Buscar")
    print("4. Salir")

    opcion = input("Elige: ")

    if opcion == '1':
        nombre = input("Nombre: ")
        while True:
            try:
                nota = float(input(f"Nota de {nombre} (10-50): "))
                if 10 <= nota <= 50:
                    calificaciones[nombre] = nota
                    print(f"{nombre} guardado.")
                    break
                else:
                    print("Nota debe ser 0-10.")
            except ValueError:
                print("Escribe un número.")

    elif opcion == '2':
        print("\n--- Todas ---")
        [print(f"{n}: {c}") for n, c in calificaciones.items()] if calificaciones else print("No hay notas.")

    elif opcion == '3':
        buscar = input("Buscar nombre: ")
        print(f"Nota de {buscar}: {calificaciones.get(buscar, 'No encontrado')}")

    elif opcion == '4':
        print("¡Adiós!")
        break

    else:
        print("Opción inválida.")
```

5. Calculadora de Estadísticas de Números Descripción: Desarrolla una aplicación que trabaje con una colección de números predefinidos. El sistema debe ofrecer al usuario un menú interactivo con opciones para calcular el promedio, identificar el valor máximo y mínimo, o eliminar un número. Tras cada operación, se mostrará tanto el resultado como el estado actual de la colección.

```python
numeros = [10, 5, 20, 15, 5, 30]
while True:
    o = input("1 Promedio, 2 Máximo, 3 Mínimo, 4 Eliminar, 5 Ver, 6 Salir: ").strip()
    if o == '1': print("Promedio:", sum(numeros) / len(numeros) if numeros else "Vacío")
    elif o == '2': print("Máximo:", max(numeros) if numeros else "Vacío")
    elif o == '3': print("Mínimo:", min(numeros) if numeros else "Vacío")
    elif o == '4':
        if numeros:
            try:
                e = float(input("Eliminar: "))
                if e in numeros: numeros.remove(e); print(f"{e} eliminado.")
                else: print(f"{e} no está.")
            except ValueError: print("Inválido.")
        else: print("Vacío.")
    elif o == '5': print("Números:", numeros)
    elif o == '6': break
    else: print("Inválido.")
```
