EJERCICIO 1 Procesamiento de Números del 1 al 20:

Se requiere desarrollar un programa que trabaje con los números comprendidos entre el 1 y el 20, de forma que se identifiquen y muestren aquellos que sean divisibles por dos, acompañados de su posición en la secuencia. Además, el sistema debe recorrer la colección para determinar cuántos números no cumplen esa propiedad y calcular la suma de los que sí la cumplen. El objetivo es diseñar la lógica necesaria para examinar y procesar los números según las condiciones indicadas.

```python
numeros = list(range(1, 20))
divisibles = []
no_divisibles = 0
suma_divisibles = 0

for i in range(len(numeros)):
    if numeros[i] % 2 == 0:
        divisibles.append((numeros[i], i + 1))
        suma_divisibles += numeros[i]
    else:
        no_divisibles += 1

print("Números divisibles por dos y su posición:", divisibles)
print("Cantidad de números no divisibles por dos:", no_divisibles)
print("Suma de números divisibles por dos:", suma_divisibles)
```

EJERCICIO 2 Administración de Actividades Diarias:

Imagina una aplicación para gestionar las actividades cotidianas de una persona, donde se disponga de una serie de tareas, algunas de las cuales indican que ya han sido concluidas. El desafío consiste en presentar cada actividad junto a su posición en la colección, filtrar aquellas que contengan una marca de finalización y actualizar el conjunto de actividades. Finalmente, se debe exhibir el estado del conjunto antes y después de aplicar el filtro, permitiendo al usuario observar los cambios realizados.

```python
actividades = [
    "Desayunar [ ]",
    "Hacer ejercicio [X]",
    "Trabajar [ ]",
    "Almorzar [X]",
    "Estudiar [ ]",
    "Cenar [ ]",
    "Dormir [X]"
]

numeros_originales = list(range(1, len(actividades) + 1))

print("Todas las actividades:")
for i in range(len(actividades)):
     print(f"{numeros_originales[i]}. {actividades[i]}")

actividades_filtradas = []
numeros_filtrados = []

for i in range(len(actividades)):
    if "[X]" not in actividades[i]:
        actividades_filtradas.append(actividades[i])
        numeros_filtrados.append(numeros_originales[i])

actividades = actividades_filtradas
numeros_originales = numeros_filtrados

print("Actividades pendientes:")
for i in range(len(actividades)):
    print(f"{numeros_originales[i]}. {actividades[i]}")
```

EJERCICIO 3 Filtrado y Transformación de Texto:

Propón un sistema en el que se solicite al usuario una secuencia de caracteres. A partir de la entrada, se debe descomponer el texto en sus componentes individuales y extraer únicamente aquellos que correspondan a letras del alfabeto, omitiendo números y símbolos. Una vez realizado este filtrado, la colección resultante deberá ser convertida en una estructura que no permita modificaciones, y el resultado final se debe mostrar al usuario.

```python
entrada_usuario = input("Ingresa una secuencia de caracteres: ")

letras_filtradas = []
for caracter in entrada_usuario:
    if 'a' <= caracter <= 'z' or 'A' <= caracter <= 'Z':
        letras_filtradas.append(caracter)

tupla_letras = tuple(letras_filtradas)

print("Letras filtradas ineditables son: ", tupla_letras)
```

EJERCICIO 4 Informe Interactivo de Alumnos:

Considera dos conjuntos de datos paralelos: uno con nombres de estudiantes y otro con sus respectivas calificaciones. El reto consiste en elaborar un informe que relacione cada alumno con su nota, presentando también la posición de cada registro. Posteriormente, el sistema debe permitir al usuario consultar la nota de un alumno específico introduciendo la posición correspondiente, finalizando la consulta si se ingresa un valor fuera del rango válido. La solución deberá vincular ambas colecciones de manera sincronizada y ofrecer una interacción dinámica.

```python
nombres_estudiantes = ["pedro", "carlos", "sofía", "jose", "laura"]
calificaciones = [50, 40, 35, 20, 30]

print("Informe de calificaciones:")
for i in range(len(nombres_estudiantes)):
    print(f"Posición: {i + 1}, Alumno: {nombres_estudiantes[i]}, Calificación: {calificaciones[i]}")

while True:
    try:
        posicion = int(input("Ingrese la posición del alumno o 0 para salir: "))
        if posicion == (0):
            break
        if 1 <= posicion <= len(nombres_estudiantes):
            print(f"La calificación de {nombres_estudiantes[posicion - 1]} es: {calificaciones[posicion - 1]}")
        else:
            print("Posición fuera de rango.")
    except ValueError:
        print("Entrada inválida. Ingrese un número entero.")
```

EJERCICIO 5 Aplicación de Operaciones sobre Valores Numéricos:

Diseña una aplicación que inicie con una serie de valores numéricos predefinidos. El usuario interactuará mediante un menú que le permita realizar distintas operaciones: calcular la suma total de los valores, reordenarlos según un criterio específico o eliminar uno de ellos según una indicación. La interfaz deberá ofrecer la posibilidad de repetir estas acciones hasta que el usuario decida finalizar. El desafío es organizar el flujo de la interacción y manipular la colección de números de acuerdo a la opción elegida.

```python
valores_numericos = [10, 5, 20, 15, 25]

while True:
    print("\nMenú de opciones:")
    print("1. Calcular suma total")
    print("2. Reordenar valores")
    print("3. Eliminar valor")
    print("4. Salir")

    opcion = input("Ingresa el número de la opción deseada: ")

    if opcion == "1":
        suma_total = 0
        for valor in valores_numericos:
            suma_total += valor
        print("La suma total es:", suma_total)

    elif opcion == "2":
        print("Opciones de reordenamiento:")
        print("a. Menor a mayor")
        print("b. Mayor a menor")
        criterio = input("Ingresa el criterio de reordenamiento (a/b): ")
        if criterio == "a":
            for i in range(len(valores_numericos)):
                for j in range(i + 1, len(valores_numericos)):
                    if valores_numericos[i] > valores_numericos[j]:
                        valores_numericos[i], valores_numericos[j] = valores_numericos[j], valores_numericos[i]
            print("Los valores reordenados de menor a mayor son:", valores_numericos)
        elif criterio == "b":
            for i in range(len(valores_numericos)):
                for j in range(i + 1, len(valores_numericos)):
                    if valores_numericos[i] < valores_numericos[j]:
                        valores_numericos[i], valores_numericos[j] = valores_numericos[j], valores_numericos[i]
            print("Los valores reordenados de mayor a menor:", valores_numericos)
        else:
            print("Criterio inválido.")

    elif opcion == "3":
        try:
            valor_eliminar = int(input("Ingresa el valor a eliminar: "))
            if valor_eliminar in valores_numericos:
                nueva_lista = []
                for valor in valores_numericos:
                    if valor != valor_eliminar:
                        nueva_lista.append(valor)
                valores_numericos = nueva_lista
                print("Valor eliminado.")
            else:
                print("El valor no se encuentra en la lista.")
        except ValueError:
            print("Entrada inválida. Ingresa un número entero.")

    elif opcion == "4":
        print("Saliendo de la aplicación.")
        break

    else:
        print("Opción inválida.")
```
