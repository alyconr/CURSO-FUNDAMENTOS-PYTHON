EJERCICIO 2: Crear una lista de números y solicitar un número. Borrar de la lista ese número en todas las posiciones donde se encuentre. Presentar la lista final

```python
lista_inicial = list('1266285377198')
numero_usuario = int(input("Ingrese el número que desea eliminar: "))

lista_numerica = [int(x) for x in lista_inicial]
print("Lista inicial:", lista_numerica)


while numero_usuario in lista_numerica:
    lista_numerica.remove(numero_usuario)

print("Lista final:", lista_numerica)
```

EJERCICIO 3: Solicite al usuario dos(2) frases y devuelva una lista con todas las letras que se repiten en la misma posición de ambas listas

ejemplo:
frase1:  "holasena"
frase2:  "todogana"
salida: ["o","n", "a"]

```python
frase1 = input("Ingrese la primera frase: ")
frase2 = input("Ingrese la segunda frase: ")

letras_repetidas = []
longitud_minima = min(len(frase1), len(frase2))

for i in range(longitud_minima):
    if frase1[i] == frase2[i]:
        letras_repetidas.append(frase1[i])

print("Letras repetidas en la misma posición:", letras_repetidas)
```
