EJERCICIO 1: Se dispone de una lista de códigos de productos que puede contener elementos duplicados. El reto consiste en convertir la lista en una colección que elimine los duplicados, agregar un nuevo producto al conjunto y luego descartar un producto específico. Posteriormente, se compararán dos colecciones (por ejemplo, inventarios de dos almacenes) para identificar los productos comunes, aquellos exclusivos de cada inventario y los que no se repiten entre ambos.

SALIDA:

Inventario original: {101, 102, 103, 104, 105}

Inventario tras agregar producto 110: {101, 102, 103, 104, 105, 110}

Inventario tras descartar producto 103: {101, 102, 104, 105, 110}

 .

Inventario Almacén A: {101, 102, 104, 110}

Inventario Almacén B: {102, 104, 107, 110}

Productos comunes: {102, 104, 110}

Productos exclusivos en Almacén A: {101}

Productos exclusivos en Almacén B: {107}

```python
codigos_productos = [101, 102, 103, 104, 105, 102, 103, 110]
inventario_almacen_a = {101, 102, 104, 110}
inventario_almacen_b = {102, 104, 107, 110}

inventario_original = set(codigos_productos)
print("Inventario original:", inventario_original)

inventario_original.add(110)
print("Inventario tras agregar producto 110:", inventario_original)

inventario_original.discard(103)
print("Inventario tras descartar producto 103:", inventario_original)

productos_comunes = inventario_almacen_a.intersection(inventario_almacen_b)
productos_exclusivos_a = inventario_almacen_a.difference(inventario_almacen_b)
productos_exclusivos_b = inventario_almacen_b.difference(inventario_almacen_a)

print("\nInventario Almacén A:", inventario_almacen_a)
print("Inventario Almacén B:", inventario_almacen_b)
print("Productos comunes:", productos_comunes)
print("Productos exclusivos en Almacén A:", productos_exclusivos_a)
print("Productos exclusivos en Almacén B:", productos_exclusivos_b)
```

EJERCICIO 2: Dos grupos de amigos tienen colecciones con sus géneros musicales favoritos. Se requiere determinar si ambos grupos comparten al menos un género (evaluar si las colecciones son disjuntas o no). Además, se debe presentar la intersección de géneros, la unión de ambos conjuntos y la diferencia simétrica (los géneros que no coinciden entre los grupos).

SALIDA:

Preferencias Grupo A: {'rock', 'pop', 'jazz'}

Preferencias Grupo B: {'pop', 'reggae', 'electrónica'}

.

¿Tienen géneros en común? Sí

Géneros comunes: {'pop'}

Unión de géneros: {'rock', 'pop', 'jazz', 'reggae', 'electrónica'}

Géneros exclusivos (diferencia simétrica): {'rock', 'jazz', 'reggae', 'electrónica'}


```python
preferencias_grupo_a = {'rock', 'pop', 'jazz'}
preferencias_grupo_b = {'pop', 'reggae', 'electrónica'}

generos_comunes = preferencias_grupo_a.intersection(preferencias_grupo_b)

tienen_comun = len(generos_comunes) > 0
union_generos = preferencias_grupo_a.union(preferencias_grupo_b)
diferencia_simetrica = preferencias_grupo_a.symmetric_difference(preferencias_grupo_b)

print("Preferencias Grupo A:", preferencias_grupo_a)
print("Preferencias Grupo B:", preferencias_grupo_b)
print("\n¿Tienen géneros en común?", "Sí")
print("Géneros comunes:", generos_comunes)
print("Unión de géneros:", union_generos)
print("Géneros exclusivos (diferencia simétrica):", diferencia_simetrica)
```

EJERCICIO 1: Desarrolla un traductor sencillo que utilice un diccionario donde las claves sean palabras en español y los valores sus traducciones en inglés. El usuario ingresará una palabra en español; si la palabra se encuentra en el diccionario, se mostrará su traducción. En caso contrario, se le solicitará al usuario que proporcione la traducción, la cual se agregará al diccionario. El proceso se repetirá hasta que el usuario decida finalizar la consulta.

SALIDA:

Diccionario inicial: {"hola": "hello", "mundo": "world", "casa": "house"}

.

Ingrese una palabra en español: sol

La palabra 'sol' no se encuentra en el diccionario.

Ingrese la traducción de 'sol': sun

Nueva entrada agregada: 'sol': 'sun'

Ingrese una palabra en español: casa

La traducción de 'casa' es: house

.

Ingrese una palabra en español: salir

Terminando el programa.

```python
diccionario = {"hola": "hello", "mundo": "world", "casa": "house"}

print("Diccionario inicial:", diccionario)

while True:
    palabra_espanol = input("\nIngrese una palabra en español: ")
    if palabra_espanol == "salir":
        print("Terminando el programa.")
        break

    if palabra_espanol in diccionario:
        print(f"La traducción de '{palabra_espanol}' es: {diccionario[palabra_espanol]}")
    else:
        traduccion_ingles = input(f"La palabra '{palabra_espanol}' no se encuentra en el diccionario.\nIngrese la traducción de '{palabra_espanol}': ")
        diccionario[palabra_espanol] = traduccion_ingles
        print(f"Nueva entrada agregada: '{palabra_espanol}': '{traduccion_ingles}'")
```

EJERCICIO 2: Crea una aplicación que administre el inventario de una tienda usando un diccionario, donde cada clave representa el nombre de un producto y cada valor la cantidad en stock. La aplicación debe permitir consultar la cantidad de un producto, actualizar el stock de uno existente, agregar nuevos productos y eliminar aquellos que ya no estén disponibles. Al final se mostrará el inventario actualizado.

SALIDA:
 
Inventario inicial:

{"manzanas": 50, "naranjas": 30, "peras": 20}

.

Consulta: ¿Cuántas naranjas hay? -> 30

.

Actualizando stock de peras a 25.

Agregando producto: "bananas" con stock 40.

Eliminando producto: "naranjas".

.

Inventario final:

{"manzanas": 50, "peras": 25, "bananas": 40}

```python
inventario = {"manzanas": 50, "naranjas": 30, "peras": 20}

print("Inventario inicial:", inventario)

producto_consulta = "naranjas"
cantidad_consulta = inventario.get(producto_consulta)
print(f"\nConsulta: ¿Cuántas {producto_consulta} hay? -> {cantidad_consulta}")

producto_actualizar = "peras"
nueva_cantidad = 25
inventario[producto_actualizar] = nueva_cantidad
print(f"\nActualizando stock de {producto_actualizar} a {nueva_cantidad}.")

producto_agregar = "bananas"
cantidad_agregar = 40
inventario[producto_agregar] = cantidad_agregar
print(f"Agregando producto: \"{producto_agregar}\" con stock {cantidad_agregar}.")

producto_eliminar = "naranjas"
del inventario[producto_eliminar]
print(f"Eliminando producto: \"{producto_eliminar}\".")

print("\nInventario final:", inventario)
```
