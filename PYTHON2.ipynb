2. En un partido de fútbol, se ofrece un descuento a los aficionadies que depende del estrato y la edad. Si el estrato es 1 y su edad es menor a 18 el descuento será del 20% sobre el valor de la boleta. Si el estrato es 1 y el alumno tiene 18 o mas años, el descuento será del 15% Si el estrato es 2 y la edad es menor a 18 años, el descuento será del 10% y si el estrato es 2 y la edad es 18 años o más, el descuento será del 5%. Determinar el total del dinero recaudado y descontado por las últimas N personas que ingresan al partido.


```python

total_recaudado = 0
num_personas = int(input("Ingrese el número de personas: "))
valor_entrada = int(input("Ingrese el precio  de la entrada: $ "))

for i in range(num_personas):
    estrato = int(input(f"Ingrese el estrato de la persona {i+1} (1 o 2): "))
    edad = int(input(f"Ingrese la edad de la persona {i+1}: "))

    if estrato == 1:
        if edad < 18:
            descuento = 0.20
        else:
            descuento = 0.15
    elif estrato == 2:
        if edad < 18:
            descuento = 0.10
        else:
            descuento = 0.05
    else:
        descuento = 0  # Sin descuento si el estrato no es válido

    pago = valor_entrada * (1 - descuento)
    total_recaudado += pago


print(f"El total recaudado es: ${total_recaudado:.0f}")
```
