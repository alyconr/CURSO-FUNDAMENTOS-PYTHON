En un sistema de control de calidad, se deben inspeccionar las piezas de un producto para determinar si cumplen con los estándares de calidad. Si la pieza es defectuosa, se debe marcar como rechazada y enviar una alerta al operador. Si la pieza cumple con los estándares de calidad, se debe marcar como aprobada y continuar con la producción.

```python
pieza = input("Ingrese los estándares de calidad (1 si se cumple, 0 si no) separados por espacios: ")

estandares = list(map(int, pieza.split()))

if 0 in estandares:
    print("Pieza rechazada: al menos un estándar no cumplido.")
else:
    print("Pieza aprobada: cumple con todos los estándares.")
```

Leer un número y presentar la tabla de multiplicar de ese número entre 1 y 10. Utilizar el siguiente formato de ejemplo:
1 x 1 = 1
1 x 2 = 2
1 x 3 = 3
1 x 4 = 4
1 x 5 = 5

```python
numero = int(input("Ingrese un número: "))

print(f"Tabla de multiplicar del {numero}:")
contador = 1
while contador <= 10:
    resultado = numero * contador
    print(f"{numero} x {contador} = {resultado}")
    contador += 1
```

En un partido de fútbol, se ofrece un descuento a los aficionados que depende del estrato y la edad. Si el estrato es 1 y su edad es menor a 18 el descuento será del 20% sobre el valor de la boleta. Si el estrato es 1 y el alumno tiene 18 o mas años, el descuento será del 15%. Si el estrato es 2 y la edad es menor a 18 años, el descuento será del 10% y si el estrato es 2 y la edad es 18 años o más, el descuento será del 5%. Determinar el total del dinero recaudado y descontado por las últimas N personas que ingresan al partido.

```python
total_recaudado = 0
total_descuento = 0

N = int(input("Ingrese el número de personas: "))
recaudado_total = 0
monto_descuento_total = 0

for _ in range(N):
    edad = int(input("Ingrese la edad: "))
    estrato = input("Ingrese el estrato (1/2): ")
    valor_boleta = float(input("Ingrese el valor de la boleta: "))

    descuento = 0
    if estrato == "1":
        if edad < 18:
            descuento = 0.20
        else:
            descuento = 0.15
    elif estrato == "2":
        if edad < 18:
            descuento = 0.10
        else:
            descuento = 0.05

    monto_descuento = valor_boleta * descuento
    valor_final = valor_boleta - monto_descuento

    recaudado_total += valor_final
    monto_descuento_total += monto_descuento

    print(f"Descuento aplicado: {descuento * 100}%")
    print(f"Valor del descuento: {monto_descuento}")
    print(f"Valor a pagar: {valor_final}")

print(f"Total recaudado: {recaudado_total}")
print(f"Total descuentos aplicados: {monto_descuento_total}")
```

Leer un password de ingreso a un programa y mostrar el mensaje de bienvenida si es correcto. Mientras no lo sea, debe mostrar el mensaje de Password incorrecto. El programa debe terminar automáticamente al quinto intento fallido.


```python
PASSWORD_CORRECTO = "12345"  # Puedes cambiar esta contraseña
intentos = 0

while intentos < 5:
    password = input("Ingrese su password: ")
    if password == PASSWORD_CORRECTO:
        print("¡Bienvenido al programa!")
        break
    else:
        intentos += 1
        print("Password incorrecto. Intente de nuevo.")

    if intentos == 5:
        print("Ha excedido el número máximo de intentos. Acceso bloqueado.")
```
