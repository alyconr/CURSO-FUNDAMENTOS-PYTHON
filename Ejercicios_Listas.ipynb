1. Pregunte al usuario cuántos elementos desea ingresar en una lista, luego solicite cada uno de ellos y presente el contenido de la lista y su contenido invertido.


```python
cuantos_elementos = int(input("¿Cuántos elementos desea ingresar en la lista?: "))

# Esto es una lista vacia, para que los datos ingresados por el usuario sen escritos mas adelante
elementos = []
for i in range(cuantos_elementos):
    elemento = input(f"Ingrese el elemento {i + 1}: ")
    elementos.append(elemento)

print("Lista ingresada:", elementos)
print("Lista Invertida:", elementos[::-1])
```

2. Crear una lista de números y solicitar un número. Borrar de la lista ese número en todas las posiciones donde se encuentre. Presentar la lista final


```python
lista = [1, 2, 3, 4, 2, 5, 6, 7, 7, 8, 9, 2, 10, 5, 8, 7]
print(lista)

num_eliminar = int(input("Ingrese un numero de la lista anterior, que quieras eliminar: "))

nueva_lista = []
for i in range(len(lista)):
    if lista[i] != num_eliminar:
        nueva_lista.append(lista[i])


lista = nueva_lista

print("Lista final:", lista)
```

3. Solicite al usuario dos(2) frases y devuelva una lista con todas las letras que se repiten en la misma posición de ambas listas
ejemplo: frase1: "holasena" frase2: "todogana" salida: ["o","n", "a"]

```python
frase1 = input("ingrese una frase: ")
frase2 = input("ingrese otra frase: ")

letras_min = min(len(frase1), len(frase2))

lista_letras = []

for i in range(letras_min):
    if frase1[i] == frase2[i]:
      lista_letras.append(frase1[i])
print(lista_letras)
```
