1-Apropiacion de Conjuntos

Se dispone de una lista de códigos de productos que puede contener elementos duplicados. El reto consiste en convertir la lista en una colección que elimine los duplicados, agregar un nuevo producto al conjunto y luego descartar un producto específico. Posteriormente, se compararán dos colecciones (por ejemplo, inventarios de dos almacenes) para identificar los productos comunes, aquellos exclusivos de cada inventario y los que no se repiten entre ambos.

```python
inventario_Duplicado = {101, 102, 103, 104, 105, 103, 104, 105, 104, 103}
print("Inventario_original:", inventario_Duplicado)

inventario_Duplicado.add(100)
print("Inventario tras agregar producto: 100" ,inventario_Duplicado)

inventario_Duplicado.discard(101)
print("Inventario tras descartar el producto: 101",inventario_Duplicado)


Inventario_Almacén_A = {101, 102, 104, 110}
print("Inventario Almacén A:",Inventario_Almacén_A)
Inventario_Almacén_B = {102, 104, 107, 110}
print("Inventario Almacén B:",Inventario_Almacén_B)

print(f"Productos en comun: {Inventario_Almacén_A.intersection(Inventario_Almacén_B)}")

print(f"Productos exclusivos en Almacén A:  {Inventario_Almacén_A.difference(Inventario_Almacén_B)}")
print(f"Productos exclusivos en Almacén B:  {Inventario_Almacén_B.difference(Inventario_Almacén_A)}")
```

2-Apropiacion de Conjuntos

Dos grupos de amigos tienen colecciones con sus géneros musicales favoritos. Se requiere determinar si ambos grupos comparten al menos un género (evaluar si las colecciones son disjuntas o no). Además, se debe presentar la intersección de géneros, la unión de ambos conjuntos y la diferencia simétrica (los géneros que no coinciden entre los grupos).

```python
Preferencias_Grupo_A = {'rock', 'pop', 'jazz', 'Popular', 'Meregue', 'Cumbia'}
print("Preferencias Grupo A:",Preferencias_Grupo_A)
Preferencias_Grupo_B = {'pop', 'reggae', 'electrónica', 'Salsa', 'Champeta', 'Cumbia'}
print("Preferencias Grupo B:",Preferencias_Grupo_B)

Pregunta = input("¿Tienen géneros en común?: ")



print(f"Generos en comun: {Preferencias_Grupo_A.intersection(Preferencias_Grupo_B)}")

print(f"Unión de géneros: {Preferencias_Grupo_A.union(Preferencias_Grupo_B)}")


print(f"Géneros exclusivos (diferencia simétrica):: {Preferencias_Grupo_A.symmetric_difference(Preferencias_Grupo_B)}")
```

1-Apropiacion de Diccionario

Desarrolla un traductor sencillo que utilice un diccionario donde las claves sean palabras en español y los valores sus traducciones en inglés. El usuario ingresará una palabra en español; si la palabra se encuentra en el diccionario, se mostrará su traducción. En caso contrario, se le solicitará al usuario que proporcione la traducción, la cual se agregará al diccionario. El proceso se repetirá hasta que el usuario decida finalizar la consulta.

```python
Diccionario_inicial = {"hola": "hello", "mundo": "world", "casa": "house","Feliz": "Happy", "Amarrillo": "Yellow"}
print("Diccionario inicial: ", Diccionario_inicial)


while True:
    palabra = input("\nIngrese una palabra en español: ").lower()

    if palabra == "salir":
        print("Terminando el programa.")
        break

    if palabra in Diccionario_inicial:
        print(f"La traducción de '{palabra}' es: {Diccionario_inicial[palabra]}")
    else:
        print(f"La palabra '{palabra}' no se encuentra en el diccionario.")
        traduccion = input(f"Ingrese la traducción de '{palabra}': ").lower()
        Diccionario_inicial[palabra] = traduccion
        print(f"Nueva palabra agregada: '{palabra}': '{traduccion}'")
```

2-Apropiacion de Diccionario

Crea una aplicación que administre el inventario de una tienda usando un diccionario, donde cada clave representa el nombre de un producto y cada valor la cantidad en stock. La aplicación debe permitir consultar la cantidad de un producto, actualizar el stock de uno existente, agregar nuevos productos y eliminar aquellos que ya no estén disponibles. Al final se mostrará el inventario actualizado.

```python
Inventario_inicial ={"manzanas": 50, "naranjas": 30, "peras": 20, "bananos":40, "piñas":10}
print( "inventario_inicial : ",Inventario_inicial )
producto  = input("Ingrese el nombre del producto que quiere consultar : ")
if producto in Inventario_inicial:
            print("Cantidad de", producto, ":", Inventario_inicial[producto])
else:
            print("El producto", producto, "no está en el inventario.")

producto = input("Ingrese el nombre del producto a actualizar: ")
if producto in Inventario_inicial:
        cantidad = int(input("Ingrese la cantidad que quiere agregar o quitar (puede ser negativa): "))
        Inventario_inicial[producto] += cantidad
        print("Stock actualizado.")
else:
        print("El producto no existe en el inventario.")

producto = input("Ingrese el nombre del nuevo producto: ")
if producto in Inventario_inicial:
        print("Ese producto ya existe.")
else:
        cantidad = int(input("Ingrese la cantidad inicial del producto: "))
        Inventario_inicial[producto] = cantidad
        print("Producto agregado.")

producto = input("Ingrese el nombre del producto a eliminar: ")
if producto in Inventario_inicial:
        del Inventario_inicial[producto]
        print("Producto eliminado.")
else:
        print("El producto no existe en el inventario.")

print("Inventario actualizado:", Inventario_inicial)
for producto, cantidad in Inventario_inicial.items():
    print(f"- {producto}: {cantidad}")
```
