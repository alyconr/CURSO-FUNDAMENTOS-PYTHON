Recorrido de listas

Como las listas son oibjetos iterales las podemos recorrer con ciclo **for**

```python
numeros = [1,2,3,4,5]
print(numeros)
for n in numeros:
  print(n)
```

Podemos recorrer la lista utilizando el metodo **len**

```python
mi_lista = [4,5,6,7]
for indice in range (0, len(mi_lista)):
  print(mi_lista[indice])
```

Podemos acceder al indice de cada elemento a traves de la funvion **enumerate**

```python
numeros = [3,29,42,34,54]
for x,y in enumerate(numeros):
  print(f"En la posicion x {x} Encontramos el valor : {y}")
```

```python
letra = ["A", "B", "C","D", "E"]
for indice, valor in enumerate(letra):
  print(f"En la posicion {indice} Encontramos el valor {valor}")
```

```python
lista = [2,36,76,23]
ind = 0
while ind < len(lista):
  print(lista[ind])
  ind += 1
```

1. Pregunte al usuario cuántos elementos desea ingresar en una lista, luego solicite cada uno de ellos y presente el contenido de la lista y su contenido invertido.

```python
cantidad = int(input("Ingrese la cantidad de elementos que desea ingresar en la lista: "))
lista = []
for i in range(cantidad):
  elemento = input(f"Ingrese el elemento {i+1}: ")
  lista.append(elemento)
print("Contenido de la lista:")
print(lista)
print("Contenido invertido de la lista:")
print(lista[::-1])
```

2. Crear una lista de números y solicitar un número. Borrar de la lista ese número en todas las posiciones donde se encuentre. Presentar la lista final

```python
lista_n = (1, 2, 3, 4, 5, 6, 7, 8,)
numero = int(input("Ingrese un numero: "))
while numero in lista_n:
  lista_n = list(lista_n)
  lista_n.remove(numero)
  lista_n = tuple(lista_n)
print(lista_n)
```

ejemplo: frase1: "holasena" frase2: "todogana" salida: ["o","n", "a"]



```python
frase1 = input("Ingrese la primera frase: ")
frase2 = input("Ingrese la segunda frase: ")
letras_repetidas = []
for i in range(len(frase1)):
  if frase1[i] == frase2[i]:
      letras_repetidas.append(frase1[i])
print( letras_repetidas)
```
