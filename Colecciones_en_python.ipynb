#Recorridos de listas

Como las listas son objetos interables las podemos recorrer con una ciclo **for**

```python
numeros = [1,2,3,4,5]
print(numeros)
for n in numeros:
  print(n)
```

Podemos recorrer la lista utilizando el modelo **len**

```python
mi_lista = [4,5,6,7]
for indice in range (0, len(mi_lista)):
  print(mi_lista[indice])
```

Podemos acceder al indice de cada elemento a través de la función **enumerate**

```python
numeros = [3,29,42,34,54]
for x,y in enumerate(numeros):
  print(f"En la posición x {x} ENCONTRAMOS EL Valor: {y}")
```

```python
letras = ["A", "B", "C", "D", "E"]
for indice, valor in enumerate(letras):
  print(f"EN la posición x {indice} ENCONTRAMOS El Valor: {valor}")
```

```python
lista = [2,36, 76,23]
ind = 0
while ind < len(lista):
  print(lista[ind])
  ind += 1
```

Se cuenta con dos listas y se desean recorrer de manera simultánea mostrando el valor de las dos en cada posición. ¿Ayudarias a terminar el código de tal manera que se obtenga el siguiente resultado?

ma - pa

me - pe

mi - pi

mo - po

mu - pu


```python
lista1 = ["ma", "me", "mi", "mo", "mu"]
lista2 = ["pa", "pe", "pi", "po", "pu"]
for i in range(len(lista1)):
   print(f"{lista1[i]} - {lista2[i]}")
```

#Metodos Aplicables a Listas

El metodo **sort()** que nos permite ordenar los elementos

```python
mi_lista = [5,10,8,3,1,52]
mi_lista.sort()
print(mi_lista)
```

Podemos pasarle el siguiente método **reverse=True** al método **sort** para que ordene los elementos inversamente

```python
mi_lista = [5,10,8,3,1,52]
mi_lista.sort(reverse=True)
print(mi_lista)
```

El metodo **remove()** recibe como argumento un objeto y lo borra de la lista

```python
lenguajes = ["Python", "Java", "JavaScript"]
lenguajes.remove("Java")
print(lenguajes)
```

Si utilizo **del** debo utilizar la posición para eliminar el elemento, si utilizo **remove** puedo utilizar directamente el elemento que voy a borrar

```python
lenguajes = ["Python", "Java", "JavaScript"]
```

```python
del lenguajes [1]
print(lenguajes)
```

El metodo **reverse()** invierte el contenido de la lista

```python
lenguajes = ["Python", "Java", "JavaScript"]
lenguajes.reverse()
print(lenguajes)
```

```python
cadena = "HOLA MUNDO SENA"
print (cadena[::-1])
```

```python
cadena = "HOLA MUNDO SENA"
lista = list(cadena)
print(lista)
lista.reverse()
print(lista)
cadena = "".join(lista)
print(cadena)
```

2. Crear una lista de números y solicitar un número. Borrar de la lista ese número en todas las posiciones donde se encuentre. Presentar la lista final

```python
numeros = [1, 2, 3, 4, 5, 2, 3, 6]
numero_borrar = int(input("Ingrese el numero que desea borrar: "))
numeros = [n for n in numeros if n != numero_borrar]
print("Lista final:")
print(numeros)
```

3. Solicite al usuario dos (2) frases y devuelva una lista con todas las letras que se repiten en la misma posición de ambas listas.

ejemplo: frase1: "holasena" frase2: "todogana" salida: ["o","n", "a"]




```python
frase1 = input("Ingrese la primera frase: ")
frase2 = input("Ingrese la segunda frase: ")
if len(frase1) != len(frase2):
  print("Las frases deben tener la misma longitud")
else:
  letras_repetidas = []
  for i in range(len(frase1)):
    if frase1[i] == frase2[i]:
      letras_repetidas.append(frase1[i])

  print("Letras que se repiten en la misma posicin:")
  print(letras_repetidas)
```
