 1.Adivina el Número Secreto Imagina un juego en el que se establece un valor numérico oculto y el usuario debe descubrirlo mediante intentos sucesivos. Después de cada intento, el sistema indicará si el valor propuesto es mayor o menor que el número guardado, de manera que el usuario ajuste su respuesta. El proceso continuará hasta que se alcance el valor correcto, momento en el cual se mostrará un mensaje de felicitación.


```python
numero_secreto = 6
intentos = 0

print("Bienvenido al juego de adivinar el número secreto. ¡Puedes adivinarlo!")

while True:
    numero_del_usuario = int(input("Ingresa un número entre 1 y 10: "))
    intentos += 1
    if numero_del_usuario < numero_secreto:
        print("El número secreto es mayor.")
    elif numero_del_usuario > numero_secreto:
        print("El número secreto es menor.")
    else:

        print(f"¡Felicidades! Adivinaste el número en {intentos} intentos.")
        break

```

2.Validación de Contraseña Segura Se requiere crear un mecanismo en el que el usuario introduzca una clave y el sistema verifique que cumpla ciertos requisitos, como una longitud mínima, la inclusión de letras tanto en mayúsculas como en minúsculas y la presencia de números. Si la entrada no satisface estos criterios, el programa deberá solicitar una nueva clave hasta que la condición se cumpla. El resultado final mostrará que la clave proporcionada es segura.

```python
print("Bienvenido al sistema de validación de contraseñas seguras.")

while True:
    contraseña = input("Por favor, ingrese una contraseña segura: ")

    if len(contraseña) < 8:
        print("La contraseña debe tener al menos 8 caracteres.")
        continue
    tiene_mayuscula = any(caracter.isupper() for caracter in contraseña)
    if not tiene_mayuscula:
        print("La contraseña debe contener al menos una letra mayúscula.")
        continue

    tiene_minuscula = any(caracter.islower() for caracter in contraseña)
    if not tiene_minuscula:
        print("La contraseña debe contener al menos una letra minúscula.")
        continue

    tiene_numero = any(caracter.isdigit() for caracter in contraseña)
    if not tiene_numero:
        print("La contraseña debe contener al menos un número.")
        continue

    print("¡Contraseña segura!")
    break
```

3.Control de Acceso a un Evento Exclusivo Desarrolla una solución en la que se pida al usuario ingresar ciertos datos personales, como su edad y si dispone de una invitación especial. Con base en los valores ingresados, el sistema determinará si el usuario puede acceder al evento o no, mostrando un mensaje que indique su admisión o rechazo. La lógica debe considerar las diferentes combinaciones posibles de respuestas para garantizar una correcta evaluación.

```python
nombre = input("Ingresa tu nombre: ")
edad = int(input("Ingresa tu edad: "))
invitado_especial = input("¿Tienes una invitación especial? (Sí/No): ").lower()

if edad >=18 and invitado_especial == "si":
   print(f"{nombre}, eres mayor de edad y tienes una invitación especial. ¡Puedes acceder al evento!")
else:
    print(f"{nombre}, no cumples con los requisitos para acceder al evento.")
```

4.Cálculo del Factorial de un Número El desafío consiste en permitir que el usuario introduzca un número entero y, a partir de ese valor, obtener el producto de todos los números enteros positivos hasta alcanzarlo. La idea es acumular el resultado paso a paso, multiplicando cada valor intermedio, para finalmente mostrar el valor resultante que representa el factorial del número dado. Este ejercicio ayudará a comprender el proceso de acumulación mediante iteraciones.

```python
numero= int(input("Ingrese un numero entero"))
if numero <0:
  print("Ingrese un numero entero positivo")
else:
  factorial=1
  for i in range(1,numero+1):
    factorial*=i

    print(f"El factorial del numero  {numero}es {factorial}")

```
