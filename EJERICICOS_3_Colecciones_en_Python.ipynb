```python
divisibles_2 = []
no_divisibles_2 = []

for i in range(1, 20):
    if i % 2 == 0:
        divisibles_2.append(i)
    else:
        no_divisibles_2.append(i)

suma_no_divisibles = 0
for num in no_divisibles_2:
    suma_no_divisibles += num

print("Números divisibles por 2:", divisibles_2)
print("Números no divisibles por 2:", no_divisibles_2)
print("Suma de los no divisibles por 2:", suma_no_divisibles)
```

```python
tareas = ["cocinar", "Hacer ejercicio", "Leer", "estudiar para el examen", "salir con amigos"]
completadas = ["Leer", "estudiar para el examen"]
nuevas_tareas = []

indice = 0
while indice < len(tareas):
    tarea_actual = tareas[indice]
    es_completada = False
    j = 0
    while j < len(completadas):
        if tarea_actual == completadas[j]:
            es_completada = True
        j += 1
    if es_completada == False:
        nuevas_tareas.append(tarea_actual)
    indice += 1
print("Lista de tareas después del filtro:", nuevas_tareas)
```

```python
estudiantes = ["beatriz", "Carmen", "tomas", "Daniel", "Esperanza"]
calificaciones = [75, 68, 92, 50, 84]
print("Lista de estudiantes y calificaciones:")
i = 0
while i < len(estudiantes):
    print(i, estudiantes[i], "- Nota:", calificaciones[i])
    i += 1

while True:
    indice = int(input("Ingrese el número del estudiante para ver sus calificaciones(o -2 para salir): "))
    if indice == -2:
        break
    if 0 <= indice < len(estudiantes):
        print("Nota de", estudiantes[indice], ":", calificaciones[indice])
    else:
        print("Índice fuera de rango, intente nuevamente.")
```

```python
valores = [10, 5, 8, 20, 15]

while True:
    print("\nMenú de opciones:")
    print("1. Sumar todos los valores")
    print("2. Filtrar valores mayores a 10")
    print("3. Eliminar un valor")
    print("4. Salir")
    opcion = int(input("Seleccione una opción: "))

    if opcion == 1:
        suma_total = 0
        i = 0
        while i < len(valores):
            suma_total += valores[i]
            i += 1
        print("Suma total:", suma_total)
    elif opcion == 2:
        filtrados = []
        i = 0
        while i < len(valores):
            if valores[i] > 10:
                filtrados.append(valores[i])
            i += 1
        print("Valores mayores a 10:", filtrados)
    elif opcion == 3:
        eliminar = int(input("Ingrese el valor a eliminar: "))
        nueva_lista = []
        i = 0
        while i < len(valores):
            if valores[i] != eliminar:
                nueva_lista.append(valores[i])
            i += 1
        valores = nueva_lista
        print("Lista después de eliminar el valor:", valores)
    elif opcion == 4:
        print("Saliendo del programa.")
        break
    else:
        print("Opción no válida, intente de nuevo.")
```
