2. Administración de Actividades Diarias
Imagina una aplicación para gestionar las actividades cotidianas de una persona, donde se disponga de una serie de tareas, algunas de las cuales indican que ya han sido concluidas. El desafío consiste en presentar cada actividad junto a su posición en la colección, filtrar aquellas que contengan una marca de finalización y actualizar el conjunto de actividades. Finalmente, se debe exhibir el estado del conjunto antes y después de aplicar el filtro, permitiendo al usuario observar los cambios realizados.




```python
actividades = input("Ingresa las actividades separadas por comas: ")
print("Sus actividades son: ", actividades)
actividades_finalizadas = input("Ingresa las actividades finalizadas separadas por comas: ")
print("Sus actividades finalizadas son: ", actividades_finalizadas)
actividades = actividades.split(",")
actividades_finalizadas = actividades_finalizadas.split(",")
actividades = [actividad.strip() for actividad in actividades]
actividades_finalizadas = [actividad.strip() for actividad in actividades_finalizadas]
print("Actividades sin finalizar:")
for i, actividad in enumerate(actividades):
    if actividad not in actividades_finalizadas:
        print(f"{i + 1}. {actividad}")
        actividades_finalizadas.append(actividad)
print("Actividades finalizadas:")
for i, actividad in enumerate(actividades_finalizadas):
    print(f"{i + 1}. {actividad}")
    actividades.remove(actividad)
    actividades_finalizadas.remove(actividad)
print("Actividades restantes:")
for i, actividad in enumerate(actividades):
    print(f"{i + 1}. {actividad}")
```
