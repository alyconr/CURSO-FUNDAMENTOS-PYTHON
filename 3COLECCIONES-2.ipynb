2. En una escuela se permite la inscripción de sus alumnos a las siguientes actividades deportivas: futbol (f), baloncesto (b)., ambos (a)., Realice un programa que permita ingresar el nombre de sus estudiantes e inscribir a cada uno en uno o los dos deportes.
La aplicación debe mostrar los siguientes listados:

- Estudiantes inscritos en fulbol
- Estudiantes inscritos en baloncesto
- Estudiantes inscritos en ambos deportes
- Todos los estudiantes inscritos en algún deporte
- Estudiantes inscritos en solo un deporte.

Genere un menú con las opciones de la aplicación *(hecho instructor)*

```python
baloncesto = set()
futbol = set()

numero_estudiantes = int(input("INGRESE EL NUMERO DE ESTUDIANTES: "))


for estudiantes in range(numero_estudiantes):
  nombre = input("Ingrese Su Nombre: ")
  seleccion = input("A que deporte se desea Inscribir:  Futbol(f), Baloncesto(b), Ambos(a)").lower()
  if seleccion == 'f':
    futbol.add(nombre)
  elif seleccion == 'b':
    baloncesto.add(nombre)
  elif seleccion == 'a':
    futbol.add(nombre)
    baloncesto.add(nombre)
  else:
    print("No se realizó ninguna inscripción")

print("Futbol:\n", futbol)
print("Baloncesto:\n", baloncesto)
print("Ambos:\n", futbol.intersection(baloncesto)) ## Intersection is 
print("Algunos:\n", futbol.union(baloncesto)) ## etudiantes inscritps en algún deporte
print("Solo Uno:\n", futbol.symmetric_difference(baloncesto))
```

3. Se dispone de una lista de códigos de productos que puede contener elementos duplicados. El reto consiste en convertir la lista en una colección que elimine los duplicados, agregar un nuevo producto al conjunto y luego descartar un producto específico. Posteriormente, se compararán dos colecciones (por ejemplo, inventarios de dos almacenes) para identificar los productos comunes, aquellos exclusivos de cada inventario y los que no se repiten entre ambos. 

```python
inventario_original = {101, 102, 103, 104, 105}
print("Inventario original:", inventario_original)

inventario_original.add(110)
print("Inventario tras agregar producto 110:", inventario_original)

inventario_original.discard(103)
print("Inventario tras descartar producto 103:", inventario_original)

print("\n--- Comparación de Inventarios de Almacenes ---")

inventario_almacen_a = {101, 102, 104, 110}
inventario_almacen_b = {102, 104, 107, 110}

print("Inventario Almacén A:", inventario_almacen_a)
print("Inventario Almacén B:", inventario_almacen_b)

productos_comunes = inventario_almacen_a.intersection(inventario_almacen_b)
print("Productos comunes:", productos_comunes)

productos_exclusivos_a = inventario_almacen_a.difference(inventario_almacen_b)
print("Productos exclusivos en Almacén A:", productos_exclusivos_a)

productos_exclusivos_b = inventario_almacen_b.difference(inventario_almacen_a)
print("Productos exclusivos en Almacén B:", productos_exclusivos_b)
```

4. Dos grupos de amigos tienen colecciones con sus géneros musicales favoritos. Se requiere determinar si ambos grupos comparten al menos un género (evaluar si las colecciones son disjuntas o no). Además, se debe presentar la intersección de géneros, la unión de ambos conjuntos y la diferencia simétrica (los géneros que no coinciden entre los grupos).

```python
# Preferencias de géneros musicales de cada grupo
preferencias_grupo_a = {"rock", "pop", "jazz"}
preferencias_grupo_b = {"pop", "reggae", "electrónica"}

print("Preferencias Grupo A:", preferencias_grupo_a)
print("Preferencias Grupo B:", preferencias_grupo_b)

print("\n--- Análisis de preferencias ---")

tienen_en_comun = not preferencias_grupo_a.isdisjoint(preferencias_grupo_b)

print("¿Tienen géneros en común?", "Sí" if tienen_en_comun else "No")

generos_comunes = preferencias_grupo_a.intersection(preferencias_grupo_b)
print("Géneros comunes:", generos_comunes)

union_generos = preferencias_grupo_a.union(preferencias_grupo_b)
print("Unión de géneros:", union_generos)

generos_exclusivos = preferencias_grupo_a.symmetric_difference(preferencias_grupo_b)
print("Géneros exclusivos (diferencia simétrica):", generos_exclusivos)
```

```python
diccionario_de_frutas = {"manzana": "apple", "banana": "banana", "naranja": "orange", "pera": "pear", "uva": "grape"}

fruta = input("Ingrese el nombre de la fruta en español: ")

while True:
    if fruta in diccionario_de_frutas:
        print(f"La traducción de {fruta} es: {diccionario_de_frutas[fruta]}")
        break
    else:
        print("La fruta no se encuentra en el diccionario.")
        fruta = input("Ingrese el nombre de la fruta en español: ")
        if fruta == "salir":
            break

print("¡Hasta luego!")
```

2. Desarrolla un traductor sencillo que utilice un diccionario donde las claves sean palabras en español y los valores sus traducciones en inglés. El usuario ingresará una palabra en español; si la palabra se encuentra en el diccionario, se mostrará su traducción. En caso contrario, se le solicitará al usuario que proporcione la traducción, la cual se agregará al diccionario. El proceso se repetirá hasta que el usuario decida finalizar la consulta.

```python
def traductor_consola():
    """
    Un traductor sencillo español-inglés basado en colecciones de Python,
    simulando la interacción de consola proporcionada.
    """
    diccionario = {"hola": "hello", "mundo": "world", "casa": "house"}
    print(f"Diccionario inicial: {diccionario}")

    while True:
        palabra_espanol = input("Ingrese una palabra en español: ")
        palabra_espanol = palabra_espanol.lower()  # Convertir a minúsculas para ser consistente

        if palabra_espanol == "salir":
            print("Terminando el programa.")
            break
        elif palabra_espanol in diccionario:
            traduccion = diccionario[palabra_espanol]
            print(f"La traducción de '{palabra_espanol}' es: {traduccion}")
        else:
            print(f"La palabra '{palabra_espanol}' no se encuentra en el diccionario.")
            nueva_traduccion = input(f"Ingrese la traducción de '{palabra_espanol}': ")
            diccionario[palabra_espanol] = nueva_traduccion.lower()
            print(f"Nueva entrada agregada: '{palabra_espanol}': '{nueva_traduccion.lower()}'")

if __name__ == "__main__":
    traductor_consola()
```

3. Crea una aplicación que administre el inventario de una tienda usando un diccionario, donde cada clave representa el nombre de un producto y cada valor la cantidad en stock. La aplicación debe permitir consultar la cantidad de un producto, actualizar el stock de uno existente, agregar nuevos productos y eliminar aquellos que ya no estén disponibles. Al final se mostrará el inventario actualizado.

```python
inventario = {"manzanas": 50, "naranjas": 30, "peras": 20}
print("Inventario inicial:", inventario)

consulta = "naranjas"
print(f"\nConsulta: ¿Cuántas {consulta} hay? -> {inventario.get(consulta, 0)}")

actualizar = "peras"
inventario[actualizar] = 25
print(f"Actualizando stock de {actualizar} a {inventario[actualizar]}.")

agregar = "bananas"
inventario[agregar] = 40
print(f"Agregando producto: \"{agregar}\" con stock {inventario[agregar]}.")

eliminar = "naranjas"
if eliminar in inventario:
    del inventario[eliminar]
    print(f"Eliminando producto: \"{eliminar}\".")

print("\nInventario final:", inventario)
```
