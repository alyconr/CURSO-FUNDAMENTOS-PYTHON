RECORRIDO DE LISTAS

como las listas son objetos literables las podemos recorrer con un ciclo **for**

```python
numeros = [1,2,3,4,5]
print(numeros)
for n in numeros:
    print(n)
```

podemos recorrer las listas utilizando el metodo **len**

```python
mi_ista = [4,5,6,7]
for indice in range (0, len(mi_ista)):
    print(mi_ista[indice])
```

podemos acceder al indice de cada elemnto a traves de la funcion **enumerante**

```python
numeros = [3,29,42,34,54]
for x,y in enumerate(numeros):
    print(f"En la posicion x {x} encontramos el valor: {y}")
```

```python
letras = ["A","B","C","D","E"]
for indice,valor in enumerate(letras):
    print(f"en la posicion x {indice} encontramos el valor: {valor}")
```

```python
lista = [2,36,76,23]
ind = 0
while ind < len(lista):
      print(lista[ind])
      ind += 1
```

e cuenta con dos listas y se desean recorer de manera simultanea mostrando el valor de las dos en cada posicion ¿ayudarias a termianr el codigo de tal manera que se obtenga el siguiente resultado?

ma - pa
me - pe
mi - pi
mo - po
mu - pu

```python
lista_1 = ["ma","me","mi","mo","mu"]
lista_2 = ["pa","pe","pi","po","pu"]
for i in range(min(len (lista_1), len (lista_2))):
    print(lista_1[i] + " - " + lista_2[i])
```

```python
lista_1 = ["ma","me","mi","mo","mu"]
lista_2 = ["pa","pe","pi","po"]
lista_3 = ["pu","pu","pu","pu","pu"]
lista_4 = ["ta","te","ti","to","tu"]

for x in zip(lista_1,lista_2,lista_3,lista_4):
    print(x)
```

Pregunte al usuario cuántos elementos desea ingresar en una lista, luego solicite cada uno de ellos y presente el contenido de la lista y su contenido invertido.

```python
usuario = int (input("¿Cuanto elementos desea ingresar? "))

for x in range(usuario):
  elementos = input("ingresa el elemento")
  elementos = list(elementos)
  print(elementos)
  elementos = elementos[::-1]
  print(elementos)



```

Crear una lista de números y solicitar un número. Borrar de la lista ese número en todas las posiciones donde se encuentre. Presentar la lista final

```python
numeros = [int(x) for x in input("Ingrese una lista de números separados por espacio: ").split()]
num_a_eliminar = int(input("Ingrese el número a eliminar: "))
numeros = [num for num in numeros if num != num_a_eliminar]
print("Lista final:", numeros)
```

Solicite al usuario dos(2) frases y devuelva una lista con todas las letras que se repiten en la misma posición de ambas listas
ejemplo: frase1: "holasena" frase2: "todogana" salida: ["o","n", "a"]

```python
frase1 = input("Ingrese la primera frase: ")
frase2 = input("Ingrese la segunda frase: ")
letras_repetidas = [letra1 for letra1, letra2 in zip(frase1, frase2) if letra1 == letra2]
print("Letras repetidas en la misma posición:", letras_repetidas)
```
