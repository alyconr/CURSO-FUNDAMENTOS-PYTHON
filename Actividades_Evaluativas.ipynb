1. Gestión de Inventario de Productos Descripción: Se debe desarrollar un sistema que administre un inventario compuesto por productos y sus cantidades. El programa presentará el listado inicial y resaltará aquellos artículos cuyo stock esté por debajo de un umbral definido. Luego, permitirá actualizar la cantidad de un producto específico, mostrando el inventario antes y después de la modificación.

```python
inventario = {  # Diccionario para almacenar el inventario
    "Producto Zapato": 10,
    "Producto Busos": 5,
    "Producto Gorras": 20,
    "Producto Dados": 7,
    "Producto Estrellas": 15,
}
umbral = 8  # Umbral para stock bajo

print("Inventario inicial:")  # Mostrar el inventario inicial
for producto, cantidad in inventario.items():  # Recorrer el inventario
    print(f"{producto}: {cantidad}")  # Mostrar el producto y su cantidad

print("\nProductos con stock bajo:")  # Mostrar productos con stock bajo
for producto, cantidad in inventario.items():  # Recorrer el inventario
    if cantidad < umbral:  # Verificar si la cantidad está por debajo del umbral
        print(f"{producto}: {cantidad}")  # Mostrar el producto con stock bajo


producto_actualizar = input("\nIngresa el producto a actualizar: ")  # Solicitar al usuario el producto a actualizar
nueva_cantidad = int(input("Ingresa la nueva cantidad: "))  # Solicitar al usuario la nueva cantidad

if producto_actualizar in inventario:  # Verificar si el producto existe en el inventario
    inventario[producto_actualizar] = nueva_cantidad  # Actualizar la cantidad del producto
    print("\nInventario actualizado:")  # Mostrar el inventario actualizado
    for producto, cantidad in inventario.items():  # Recorrer el inventario
        print(f"{producto}: {cantidad}")  # Mostrar el producto y su cantidad
else:
    print("El producto ingresado no existe en el inventario.")  # Mostrar mensaje de error
```

2. Búsqueda de Palabras en un Texto Descripción: Crea una aplicación que solicite al usuario un párrafo y, a continuación, le permita buscar una palabra específica. El sistema debe contar cuántas veces aparece dicha palabra en el texto, mostrar las posiciones donde se encuentra (sin distinguir mayúsculas de minúsculas) y comunicar el resultado al usuario.

```python
Parrafo = input("Ingresa Un parrafo De Texto: ")  # Solicitar al usuario el parrafo de texto
Palabra_1 = input("Ingrese la palabara que deseas buscar: ")

Parrafo_miniscula = Parrafo.lower()
print(Parrafo_miniscula)
Palabra_1_miniscula = Palabra_1.lower()
print(Palabra_1_miniscula)

Palabra_1 = Parrafo_miniscula.split()  # Dividir el párrafo en palabras

conteo = 0  # Inicializar el contador de coincidencias
posiciones = []  # Lista para almacenar las posiciones

for i, p in enumerate(Palabra_1):  # Recorrer la lista de palabras
    if p == Palabra_1_miniscula:  # Verificar si la palabra coincide
        conteo += 1  # Incrementar el contador
        posiciones.append(i)  # Agregar la posición a la lista

print(f"La palabra '{Palabra_1_miniscula}' aparece {conteo} veces en las posiciones {posiciones}")  # Mostrar el resultado
```

3. Conversión y Limpieza de Texto Descripción: Diseña una solución que pida al usuario una frase, la divida en palabras y elimine cualquier carácter no alfabético de cada una. Una vez procesada la información, la estructura resultante se transformará en una forma inmutable, y se presentará el contenido final al usuario de manera ordenada.


```python
Frase = input("Ingresa Un parrafo De Texto con caracteres especiales intercalados: ") # Solicitar al usuario el parrafo de texto

Frase = Frase.split() # Dividir el párrafo en palabras
print(Frase)

letras = [] # Lista para almacenar las letras filtradas

for caracter in Frase: # Recorrer cada carácter de la entrada
  if caracter.isalpha(): # Verificar si el carácter es una letra
    letras.append(caracter) # Agregar la letra a la lista
tupla_letras = tuple(letras) # Convertir la lista en una tupla

print(tupla_letras) # Mostrar la tupla resultante
```

4. Registro y Consulta de Calificaciones de Exámenes Descripción: Implementa un sistema que gestione un registro de estudiantes y sus calificaciones. El programa debe permitir agregar nuevos registros, listar todos los existentes y consultar la calificación de un estudiante introduciendo su nombre. Se espera que la información se actualice y se presente de forma clara en cada paso.

```python
lista_1 = ["Ana", "Julian", "Camilo", "Samuel", "Matias", "Carlos"]
lista_2 = [100, 90, 78, 50, 80, 60]

def agregar_estudiante(nombre, calificacion):
    lista_1.append(nombre)
    lista_2.append(calificacion)
    print(f"Estudiante {nombre} agregado con calificación {calificacion}.")

def listar_estudiantes():
    print("\nLista de estudiantes:")
    for i in range(len(lista_1)):
        print(f"{lista_1[i]} - {lista_2[i]}")

def consultar_calificacion(nombre):
    if nombre in lista_1:
        indice = lista_1.index(nombre)
        print(f"La calificación de {nombre} es: {lista_2[indice]}")
    else:
        print(f"No se encontró al estudiante {nombre}.")

while True:
    print("\nMenú:")
    print("1. Agregar estudiante")
    print("2. Listar estudiantes")
    print("3. Consultar calificación")
    print("4. Salir")

    opcion = input("Elige una opción: ")

    if opcion == "1":
        nombre = input("Nombre del estudiante: ")
        calificacion = int(input("Calificación: "))
        agregar_estudiante(nombre, calificacion)
    elif opcion == "2":
        listar_estudiantes()
    elif opcion == "3":
        nombre = input("Nombre del estudiante: ")
        consultar_calificacion(nombre)
    elif opcion == "4":
        print("Programa finalizado.")
        break
    else:
        print("Opción no válida.")
```

5. Calculadora de Estadísticas de Números Descripción: Desarrolla una aplicación que trabaje con una colección de números predefinidos. El sistema debe ofrecer al usuario un menú interactivo con opciones para calcular el promedio, identificar el valor máximo y mínimo, o eliminar un número. Tras cada operación, se mostrará tanto el resultado como el estado actual de la colección.

```python
numeros = [10, 5, 20, 15, 25, 40, 60, 70]  # Lista inicial de números

while True:  # Bucle principal del programa
    print("\nMenú:")  # Mostrar el menú
    print("1. Calcular promedio")
    print("2. Encontrar máximo y mínimo")
    print("3. Eliminar número")
    print("4. Mostrar lista actual")
    print("5. Salir")

    opcion = input("Elige una opción: ")  # Solicitar al usuario que elija una opción

    if opcion == "1":  # Calcular promedio
        promedio = sum(numeros) / len(numeros)  # Calcular el promedio
        print(f"El promedio es: {promedio}")  # Mostrar el promedio

    elif opcion == "2":  # Encontrar máximo y mínimo
        maximo = max(numeros)  # Encontrar el valor máximo
        minimo = min(numeros)  # Encontrar el valor mínimo
        print(f"El valor máximo es: {maximo} y el valor mínimo es: {minimo}")  # Mostrar los valores máximo y mínimo

    elif opcion == "3":  # Eliminar número
        numero_eliminar = int(input("Ingresa el número a eliminar: "))  # Solicitar al usuario que ingrese el número a eliminar
        if numero_eliminar in numeros:  # Verificar si el número está en la lista
            numeros.remove(numero_eliminar)  # Eliminar el número de la lista
            print(f"Número {numero_eliminar} eliminado.")  # Mostrar mensaje de confirmación

        else:
            print(f"El número {numero_eliminar} no está en la lista.")  # Mostrar mensaje de error
    elif opcion == "4":  # Mostrar lista actual
        print(f"Lista actual: {numeros}")  # Mostrar la lista actual
    elif opcion == "5":  # Salir
        print("Programa finalizado.")  # Mostrar mensaje de finalización
        break  # Salir del bucle principal
    else:
        print("Opción no válida.")  # Mostrar mensaje de error si la opción no es válida
```
