#Primero


```python
motor = input("Digite si esta prendido o apagado(on ,of)  ")


if motor == "of":
    print("El motor se ha apagado está apagado")
    print("Fin del programa")

elif motor == "on":
    print("El motor está encendido")
    temperatura = int(input("Digite la temperatura actual de su motor: "))
    if temperatura >= 80:
        print("El motor está caliente, se ha apagado")
    else:
        print("El motor está bien")
else:
    print("Entrada no válida. Por favor, digite 'of' o 'on'.")
    print ("fin del programa")


```

#Segundo

```python
matricula = int(input("Ingrese el valor de su matricula para calcular su descuento: "))
estrato = int(input("Ingrese su estrato (1 o 2): "))
edad = int(input("Ingrese su edad: "))

if estrato == 1:
    if edad < 18:
        descuento = 0.20
    else:
        descuento = 0.15
elif estrato == 2:
    if edad < 18:
        descuento = 0.10
    else:
        descuento = 0.05
else:
    descuento = 0

valor_descuento = matricula * descuento
precio_final = matricula - valor_descuento

print(f"El valor del descuento es: {valor_descuento}")
print(f"El precio final de la matrícula es: {precio_final}")

```

#Quinto

```python
N = int(input("Ingrese el número de personas: "))
valor_boleta = float(input("Ingrese el valor de la boleta: "))

total_recaudado = 0
total_descontado = 0

for i in range(N):
    print(f"\nPersona {i + 1}:")
    estrato = int(input("Ingrese el estrato (1 o 2): "))
    edad = int(input("Ingrese la edad: "))
    if estrato == 1:
        if edad < 18:
            descuento = 0.20
        else:
            descuento = 0.15
    elif estrato == 2:
        if edad < 18:
            descuento = 0.10
        else:
            descuento = 0.05
    else:
        descuento = 0

    pago_final = valor_boleta * (1 - descuento)
    total_recaudado += pago_final
    total_descontado += valor_boleta * descuento

print(f"\nTotal recaudado: ${total_recaudado:.2f}")
print(f"Total descontado: ${total_descontado:.2f}")
```
