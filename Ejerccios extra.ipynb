1.Calculadora Multiformato Desarrolla un programa que permita al usuario ingresar dos números y, a partir de ellos, realice diversas operaciones matemáticas. El sistema debe procesar la suma, la diferencia, el producto, la división, el residuo y la potencia, mostrando cada resultado junto a su respectivo formato numérico. Además, la solución debe encargarse de transformar la información recibida para asegurar que los cálculos se efectúen con precisión y se presenten de manera ordenada.

```python
print("Bienvenido a la calculadora multiformato")
numero1 = float(input("Ingrese el primer número: "))
numero2 = float(input("Ingrese el segundo número: "))
suma =numero1 +numero2
print(f"la suma de los dos numeros es {suma}")
resta=numero1-numero2
print(f"la resta de los dos numeros es {resta}")
multiplicacion=numero1*numero2
print(f"La multiplicion de los dos numeros es {multiplicacion}")
division=numero1/numero2
print(f"La division de los dos numeros es {division}")
residuo=numero1%numero2
print(f"El residuo de la division entre los dos numeros es {residuo}")
potencia=numero1**numero2
print(f"La potencia de los dos numeros es {potencia}")
```

2.Conversión entre Sistemas Numéricos El objetivo es construir una aplicación que solicite un número en formato decimal y lo convierta a sus equivalentes en otros sistemas, como el binario, el octal y el hexadecimal. Asimismo, el programa deberá permitir realizar la conversión inversa, de modo que el usuario pueda ingresar un número en cualquiera de estos formatos y obtener su valor en el sistema decimal. Se espera que la solución organice y formatee los resultados de manera clara para facilitar la comparación entre las distintas representaciones.

```python
def decimal_a_otros():
    decimal = int(input("Ingrese un número decimal: "))
    binario = bin(decimal)
    octal = oct(decimal)
    hexadecimal = hex(decimal)
    print("\nResultados:")
    print(f"Decimal: {decimal}")
    print(f"Binario: {binario}")
    print(f"Octal: {octal}")
    print(f"Hexadecimal: {hexadecimal}\n")

def otros_a_decimal():
    sistema = input("Ingrese el sistema numérico (binario, octal, hexadecimal): ").lower()
    numero = input(f"Ingrese el número en {sistema}: ")
    if sistema == "binario":
        decimal = int(numero, 2)
    elif sistema == "octal":
        decimal = int(numero, 8)
    elif sistema == "hexadecimal":
        decimal = int(numero, 16)
    else:
        print("Sistema numérico no válido.")
        return
    print(f"\nEl número {numero} en {sistema} es {decimal} en decimal.\n")


    while True:
        print("1. Convertir de decimal a binario, octal y hexadecimal")
        print("2. Convertir de binario, octal o hexadecimal a decimal")
        print("3. Salir")
        opcion = input("Seleccione una opción: ")
        if opcion == "1":
            decimal_a_otros()
        elif opcion == "2":
            otros_a_decimal()
        elif opcion == "3":
            print("¡Hasta luego!")
            break
        else:
            print("Opción no válida. Intente de nuevo.\n")


```

3.Control de Precios y Descuentos Imagina una solución que pida al usuario el nombre de un producto junto con su precio original. El reto consiste en calcular un descuento fijo, mostrar el valor descontado y determinar el precio final tras aplicar la reducción. La información se debe presentar de forma legible, utilizando formatos que permitan distinguir claramente los montos involucrados y garantizando la exactitud en los cálculos mediante operaciones matemáticas adecuadas.

```python
print("Bienvenido a el programa de control de precios y descuentos")
nombre_producto =input("Ingrese el nombre del producto:")
precio_producto =float(input("Ingrese el precio del producto:"))
descuento= 0.20
monto_descuento =precio_producto*descuento
precio_final=precio_producto-monto_descuento
print(f"Producto {nombre_producto}")
print(f"Precio original: ${precio_producto}")
print(f"Descuento aplicado: ${monto_descuento}")
print(f"Precio final: ${precio_final}")
```

4.Análisis Detallado de una Frase Crea una aplicación que invite al usuario a introducir una frase y que, a partir de ella, extraiga distintos elementos: la longitud total, la letra inicial y final, y una sección intermedia del texto. El sistema debe además transformar el contenido a diferentes presentaciones, como mayúsculas y minúsculas, y contar cuántas veces aparece un carácter específico. Este ejercicio pone a prueba la capacidad de manipular y transformar secuencias textuales, así como de realizar comparaciones y conteos precisos.

```python
print("Bienvenido al programa de analisis de frases")
frase=input("Ingrese una frase:")
longitud_frase=len(frase)
print(f"La longitud de la frase es {longitud_frase}")
letra_inicial=frase[0]
print(f"La letra inicial de la frase es {letra_inicial}")
letra_final=frase[-1]
print(f"La letra final de la frase es {letra_final}")

inicio = longitud_frase // 4
fin = 3 * (longitud_frase // 4)
seccion_intermedia = frase[inicio:fin]
print(f"Sección intermedia: {seccion_intermedia}")

frase_mayuscula=frase.upper()
frase_minuscula=frase.lower()
print(f"Frase en mayusculas: {frase_mayuscula}")
print(f"Frase en minusculas: {frase_minuscula}")

caracter_contado= input("Ingrese un caracter para contar en la frase:")
contador=frase.count(caracter_contado)
print(f"El caracter {caracter_contado} aparece {contador} veces en la frase")
```

5.Cálculo de Bonificaciones Laborales Diseña una solución que reciba datos de un trabajador, tales como su remuneración actual y el tiempo que ha estado en la empresa, para calcular una bonificación basada en un porcentaje determinado. El programa debe computar el incremento correspondiente, mostrar el monto adicional y el nuevo total, presentando la información de forma clara y formateada. El desafío consiste en aplicar operaciones aritméticas precisas y transformar los datos de entrada para lograr un resultado correcto y comprensible.

```python
print("Bienvenido al programa que te permitirá calcular tus bonificaciones laborales")
sueldo = float(input("Ingrese su sueldo actual: "))

años_antiguedad = int(input("Ingrese los años de antigüedad en la empresa: "))
porcentaje_bonificacion = 0.05
bonificacion = sueldo * porcentaje_bonificacion * años_antiguedad
nuevo_sueldo = sueldo + bonificacion
print("Resultados:")
print(f"Sueldo actual: ${sueldo:} COP")
print(f"Bonificación por {años_antiguedad} años de antigüedad: ${bonificacion:} COP")
print(f"Nuevo sueldo total: ${nuevo_sueldo:} COP")
```
