Pregunte al usuario cuántos elementos desea ingresar en una lista, luego solicite cada uno de ellos y presente el contenido de la lista y su contenido invertido.


```python
n = int(input("Cuantos elementos desea ingresar en la lista? "))
lista = []
for _ in range(n):
    elemento = input("Ingrese un elemento: ")
    lista.append(elemento)
print("Lista ingresada:", lista)
print("Lista invertida:", list(reversed(lista)))
```

Crear una lista de números y solicitar un número. Borrar de la lista ese número en todas las posiciones donde se encuentre. Presentar la lista fina

```python
lista_numeros = list(map(int, input("Ingrese una lista de numeros separados por espacio: ").split()))
num_a_borrar = int(input("Ingrese el numero a eliminar: "))
lista_numeros = [num for num in lista_numeros if num != num_a_borrar]
print("Lista final:", lista_numeros)
```

Solicite al usuario dos(2) frases y devuelva una lista con todas las letras que se repiten en la misma posición de ambas listas
ejemplo: frase1: "holasena" frase2: "todogana" salida: ["o","n", "a"]

```python
frase1 = input("Ingrese la primera frase: ")
frase2 = input("Ingrese la segunda frase: ")
repetidas = [frase1[i] for i in range(min(len(frase1), len(frase2))) if frase1[i] == frase2[i]]
print("Letras repetidas en la misma posicion:", repetidas)
```
