1.Procesamiento de Números del 1 al 20 Se requiere desarrollar un programa que trabaje con los números comprendidos entre el 1 y el 20, de forma que se identifiquen y muestren aquellos que sean divisibles por dos, acompañados de su posición en la secuencia. Además, el sistema debe recorrer la colección para determinar cuántos números no cumplen esa propiedad y calcular la suma de los que sí la cumplen. El objetivo es diseñar la lógica necesaria para examinar y procesar los números según las condiciones indicadas.

```python
numeros = list(range(1,21))
numeros_pares =[]
contador_impares = 0
suma_numeros_pares = 0

for i, numero in enumerate(numeros):
    if numero % 2 == 0:
        numeros_pares.append(numero)
        suma_numeros_pares += numero
    else:
        contador_impares += 1

print("numeros pares:", numeros_pares)
print("cantidad de numeros impares:", contador_impares)
print("suma de los numeros pares:", suma_numeros_pares)





```

2.Administración de Actividades Diarias Imagina una aplicación para gestionar las actividades cotidianas de una persona, donde se disponga de una serie de tareas, algunas de las cuales indican que ya han sido concluidas. El desafío consiste en presentar cada actividad junto a su posición en la colección, filtrar aquellas que contengan una marca de finalización y actualizar el conjunto de actividades. Finalmente, se debe exhibir el estado del conjunto antes y después de aplicar el filtro, permitiendo al usuario observar los cambios realizados.

```python
actividades = [
    "Hacer ejercicio",
    "hacer mercado SE COMPLETO",
    "ver peliculas",
    "arreglar la casa SE COMPLETO",
    "estudiar para un examen",
    "hacer la cena SE COMPLETO",
    "Leer un libro"
]

print("Lista de actividades iniciales:")
for i, actividad in enumerate(actividades, start=1):
    print(f"{i}. {actividad}")

actividades_pendientes = [actividad for actividad in actividades if "SE COMPLETO" not in actividad]

print("lista de actividades pendientes:")
for i, actividad in enumerate(actividades_pendientes, start=1):
    print(f"{i}. {actividad}")

total_actividades = len(actividades)
total_pendientes = len(actividades_pendientes)

if total_pendientes == 0:
    print("Todas las actividades han sido completadas")
else:
    print(f"Quedan {total_pendientes} actividades pendientes de las{total_actividades} completadas. ")





```

3.Filtrado y Transformación de Texto Propón un sistema en el que se solicite al usuario una secuencia de caracteres. A partir de la entrada, se debe descomponer el texto en sus componentes individuales y extraer únicamente aquellos que correspondan a letras del alfabeto, omitiendo números y símbolos. Una vez realizado este filtrado, la colección resultante deberá ser convertida en una estructura que no permita modificaciones, y el resultado final se debe mostrar al usuario.

```python
def filtrar_letras(texto):

    letras = [caracter for caracter in texto if caracter.isalpha()]

    resultado_inmutable = tuple(letras)

    return resultado_inmutable


entrada_usuario = input("Ingrese una secuencia de caracteres: ")


resultado = filtrar_letras(entrada_usuario)

print("Resultado (solo letras en una estructura inmutable):", resultado)

```

4.Informe Interactivo de Alumnos Considera dos conjuntos de datos paralelos: uno con nombres de estudiantes y otro con sus respectivas calificaciones. El reto consiste en elaborar un informe que relacione cada alumno con su nota, presentando también la posición de cada registro. Posteriormente, el sistema debe permitir al usuario consultar la nota de un alumno específico introduciendo la posición correspondiente, finalizando la consulta si se ingresa un valor fuera del rango válido. La solución deberá vincular ambas colecciones de manera sincronizada y ofrecer una interacción dinámica.

```python
# Definición de las listas de nombres y calificaciones
lista_nombre_estudiantes = ["Julian", "Luis", "Ana", "Marcos"]
lista_calificaciones = [90, 80, 50, 30]

# Mostrar el informe académico de estudiantes
print("Informe académico de estudiantes:")
for i, (nombre, nota) in enumerate(zip(lista_nombre_estudiantes, lista_calificaciones), start=1):
    print(f"{i}. {nombre} - {nota}")

# Bucle para consultas interactivas
while True:
    try:
        # Solicitar al usuario que ingrese una posición
        posicion = int(input("Ingrese un número de posición para consultar la nota del estudiante (fuera del rango para salir): "))

        # Verificar si la posición está dentro del rango válido
        if 1 <= posicion <= len(lista_nombre_estudiantes):
            print(f"{lista_nombre_estudiantes[posicion - 1]} tiene una calificación de {lista_calificaciones[posicion - 1]}")
        else:
            print("Posición no válida. Fin de la consulta.")
            break  # Salir del bucle si la posición es inválida
    except ValueError:
        print("Por favor, ingrese un número válido.")



```

5.Aplicación de Operaciones sobre Valores Numéricos Diseña una aplicación que inicie con una serie de valores numéricos predefinidos. El usuario interactuará mediante un menú que le permita realizar distintas operaciones: calcular la suma total de los valores, reordenarlos según un criterio específico o eliminar uno de ellos según una indicación. La interfaz deberá ofrecer la posibilidad de repetir estas acciones hasta que el usuario decida finalizar. El desafío es organizar el flujo de la interacción y manipular la colección de números de acuerdo a la opción elegida.

```python
def main():
    valores = [10, 20, 30, 40, 50]
    while True:
        print("\nMenú de opciones:\n"
              "1. Calcular la suma total de los valores\n"
              "2. Ordenar los valores en orden ascendente\n"
              "3. Ordenar los valores en orden descendente\n"
              "4. Eliminar un valor por su índice\n"
              "5. Mostrar la lista actual de valores\n"
              "6. Salir")
        opcion = input("Seleccione una opción (1-6): ")
        if opcion == '1':
            print(f"Suma total: {sum(valores)}")
        elif opcion == '2':
            valores.sort()
            print("Valores ordenados en orden ascendente.")
        elif opcion == '3':
            valores.sort(reverse=True)
            print("Valores ordenados en orden descendente.")
        elif opcion == '4':
            try:
                indice = int(input(f"Ingrese el índice del valor a eliminar (0 a {len(valores)-1}): "))
                if 0 <= indice < len(valores):
                    print(f"Valor {valores.pop(indice)} eliminado.")
                else:
                    print("Índice fuera de rango.")
            except ValueError:
                print("Entrada no válida. Por favor, ingrese un número entero.")
        elif opcion == '5':
            print(f"Valores actuales: {valores}")
        elif opcion == '6':
            print("Finalizando la aplicación.")
            break
        else:
            print("Opción no válida. Por favor, seleccione una opción del 1 al 6.")

if __name__ == "__main__":
    main()

```
