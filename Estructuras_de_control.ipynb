2.Una universidad ofrece un descuento a los estudiantes que depende del estrato y la edad. Si el estrato es 1 y su edad es menor a 18 el descuento será del 20% sobre el valor de la matricula. Si el estrato es 1 y el alumno tiene 18 o mas años, el descuento será del 15%. Si el estrato es 2 y la edad es menor a 18 años, el descuento será del 10% y si el estrato es 2 y la edad es 18 años o mas, el descuento será del 5%.

Escribir el precio que debera pagar un estudiante por su matrícula y el valor del descuento.

```python
valor_matricula = int(input("Ingrese el valor de la matrícula: "))
estrato = int(input("Ingrese el estrato del estudiante (1 o 2): "))
edad = int(input("Ingrese la edad del estudiante: "))

if estrato == 1:
    if edad >= 18:
        descuento = valor_matricula * 20 // 100
    else:
        descuento = valor_matricula * 15 // 100
else:
    if edad <= 18:
        descuento = valor_matricula * 10 // 100
    else:
        descuento = valor_matricula * 5 // 100

total_a_pagar = valor_matricula - descuento

print("El valor del descuento es: $", descuento)
print("El total a pagar por la matrícula es: $", total_a_pagar)
```

2. En un partido de fútbol, se ofrece un descuento a los aficionados que depende del estrato y la edad. Si el estrato es 1 y su edad es menor a 18 el descuento será del 20% sobre el valor de la boleta. Si el estrato es 1 y el alumno tiene 18 o mas años, el descuento será del 15%. Si el estrato es 2 y la edad es menor a 18 años, el descuento será del 10% y si el estrato es 2 y la edad es 18 años o más, el descuento será del 5%. Determinar el total del dinero recaudado y descontado por las últimas N personas que ingresan al partido.

```python
num_personas = int(input("Ingrese la cantidad de personas que ingresaron: "))
precio_boleta = int(input("Ingrese el precio de la boleta: "))

total_recaudado = 0
total_descuento = 0

for i in range(num_personas):
    print(f"\nPersona {i + 1}:")
    estrato = int(input("Ingrese el estrato (1 o 2): "))
    edad = int(input("Ingrese la edad: "))

    if estrato == 1:
        if edad < 18:
            descuento = 0.20
        else:
            descuento = 0.15
    elif estrato == 2:
        if edad < 18:
            descuento = 0.10
        else:
            descuento = 0.05
    else:
        print("Estrato no válido. Se asume sin descuento.")
        descuento = 0.0

    monto_descuento = precio_boleta * descuento
    precio_final = precio_boleta - monto_descuento

    total_recaudado += precio_final
    total_descuento += monto_descuento

    print(f"Descuento aplicado: {descuento * 100}%")
    print(f"Valor pagado por la boleta: {precio_final:.2f}")

print("\n--- Resumen del partido ---")
print(f"Total recaudado: {total_recaudado:.2f}")
print(f"Total descontado: {total_descuento:.2f}")
```

3.Leer un password de ingreso a un programa y mostrar el mensaje de bienvenida si es correcto. Mientras no lo sea, debe mostrar el mensaje de Password incorrecto. El programa debe terminar automáticamente al quinto intento fallido.

```python
password_correcto = "clave123"
intentos = 0

while intentos < 5:
    password_ingresado = input("Ingrese el password: ")

    if password_ingresado == password_correcto:
        print("Bienvenido al programa.")
        break
    else:
        print("Password incorrecto.")
        intentos += 1

if intentos == 5:
    print("Se han agotado los intentos. Acceso bloqueado.")
```
