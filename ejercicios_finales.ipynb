```python
#Ejercicio 1
import math 

def calculadora():
    try:
        num1 = float(input("Ingrese su primer numero"))
        num2 = float(input("Ingrese su segundo numero"))
        operador = input("Ingrese el operador: '+', '-', '/', '*', '**', '//', '%'")

        if operador == "+":
            resultado = num1 + num2
        elif operador == "-":
            resultado = num1 - num2
        elif operador == "/":
            resultado = num1 / num2
        elif operador == "*":
            resultado = num1 * num2
        elif operador == "**":
            resultado = num1 ** num2
        elif operador == "//":
            resultado = num1 // num2
        elif operador == "%":
            resultado = num1 % num2
        else:
            print("Operador no válido.")
            return
        
        print(f"{num1} {operador} {num2} = {resultado}")
    
    except ValueError as numero:
        print(f"Error: Ingrese un numero valido")
    except ZeroDivisionError:
        print(f"Error: no se puede dividir por 0")
    except Exception:
        print("Error inesperado")


calculadora()
```

```python
#Ejercicio 25

def administrador_inventario():
    inventario={'manzanas':40, 'mandarinas':30, 'pera':50, 'aguacate':30,'lechuga':20, 'tomate':40, 'durazno':60}

    def mostrar_inventario():
        print("INVENTARIO ACTUAL")
        for producto, cantidad in inventario.items():
            print(f"{producto.capitalize()}: {cantidad} unidades")

    while True:
        print("Opciones")  
        print("1.Mostrar inventario")
        print("2.Agregar producto")
        print("3.Actualizar producto")  
        print("4.Eliminar producto") 
        print("5.Salir")   

        opcion= input("Seleccione una opcion del (1-5):")
        
        try:
            if opcion == "1":
                mostrar_inventario()
            elif opcion == "2":
                producto=input("Ingrese el nombre del produto que desea agregar: ").lower()
                if producto in inventario:
                    print(f"ERROR:'{producto} YA ESTA EN EL INVVENTARIO")
                    continue
                cantidad_producto=int(input(f"Ingrese la cantidad de '{producto}"))
                inventario[producto]=cantidad_producto
                print("Producto agregado con exito")
            elif opcion == "3":
                producto=input("Ingrese el nombre del producto que desea actualizar").lower()
                if producto not in inventario:
                     raise KeyError(f"El producto '{producto}' no existe en el inventario")
                nueva_cantidad = int(input(f"Nueva cantidad para {producto}: "))
                inventario[producto] = nueva_cantidad
                print(f" Stock de '{producto}' actualizado a {nueva_cantidad} unidades.")
            elif opcion == "4":
                producto = input("Escriba el nombre del producto que desea eliminar: ").lower()
                if producto not in inventario:
                    raise KeyError(f"El producto {producto} no existe en el inventario")
                del inventario[producto]
                print(f"El producto {producto} ha sido eliminado del inventario")
            elif opcion == "5":
                print("Finalizando sistema")
                break
            else:
                print("Opción no valida")

        except ValueError:
            print("Error: la cantidad del producto debe ser un numero entero")
        except KeyError as pri:
            print(f"Error {pri}")
        except Exception as seg:
            print(f"Error inesperado {seg}")


administrador_inventario()
```

```python
#Ejercicio 3
import re

def analizar_texto():
    try:
        texto = input("Ingrese un texto: ").strip()
        if not texto:
            raise ValueError("No se ingresó ningún texto")
        
        palabras = re.findall(r'\b\w+\b', texto.lower())
        frecuencia = {}
        
        for palabra in palabras:
            frecuencia[palabra] = frecuencia.get(palabra, 0) + 1
        
        print("Frecuencia de palabras:")
        for palabra, count in frecuencia.items():
            print(f"{palabra}: {count}")
    
    except Exception as e:
        print(f"Error: {e}")


analizar_texto()
```

```python
# Ejercicio 4
def conversor_unidades():
    try:
        print("Seleccione la conversión:")
        print("1. Centímetros a pulgadas")
        print("2. Kilómetros a millas")
        opcion = int(input("Opción: "))
        
        if opcion == 1:
            cm = float(input("Ingrese la cantidad en centímetros: "))
            pulgadas = cm / 2.54
            print(f"Resultado: {cm} cm = {pulgadas:.2f} pulgadas")
        elif opcion == 2:
            km = float(input("Ingrese la cantidad en kilómetros: "))
            millas = km / 1.60934
            print(f"Resultado: {km} km = {millas:.2f} millas")
        else:
            print("Opción no válida")
    
    except ValueError:
        print("Error: Ingrese un valor numérico válido")

conversor_unidades()
```

```python
# Ejercicio 5
import random

def adivina_numero():
    print('¡Bienvenido al juego "Adivina el Número"!')
    secreto = random.randint(1, 50)
    intentos = 0
    
    while True:
        try:
            intentos += 1
            num = int(input(f"Intento {intentos} - Ingresa tu número: "))
            
            if num < secreto:
                print("El número secreto es mayor.")
            elif num > secreto:
                print("El número secreto es menor.")
            else:
                print(f"¡Felicidades! Has adivinado el número secreto: {secreto}")
                break
        
        except ValueError:
            print("Error: Ingrese un número válido")

adivina_numero()
```

```python
# Ejercicio 6
def calcular_factorial():
    try:
        num = int(input("Ingrese un número entero para calcular su factorial: "))
        
        if num < 0:
            print("Error: El número debe ser no negativo")
            return
        
        factorial = 1
        for i in range(1, num + 1):
            factorial *= i
        
        print(f"El factorial de {num} es {factorial}")
    
    except ValueError:
        print("Error: Ingrese un número entero válido")

calcular_factorial()
```

```python
# Ejercicio 7
from datetime import datetime

def dias_para_fecha():
    try:
        fecha_str = input("Ingrese una fecha (DD/MM/AAAA): ")
        fecha = datetime.strptime(fecha_str, "%d/%m/%Y").date()
        hoy = datetime.now().date()
        
        if fecha < hoy:
            print("La fecha ingresada ya pasó")
        else:
            dias = (fecha - hoy).days
            print(f"Faltan {dias} días para el {fecha_str}")
    
    except ValueError:
        print("Error: Formato de fecha incorrecto. Use DD/MM/AAAA")

dias_para_fecha()
```

```python
# Ejercicio 8
def registro_estudiantes():
    estudiantes = {'Ana': 90, 'Luis': 78, 'Carlos': 85}
    
    while True:
        print("\nRegistro actual:", estudiantes)
        print("\nMenú:")
        print("1. Agregar estudiante")
        print("2. Actualizar calificación")
        print("3. Eliminar estudiante")
        print("4. Listar estudiantes")
        print("5. Salir")
        
        try:
            opcion = int(input("Opción: "))
            
            if opcion == 1:
                nombre = input("Ingrese el nombre del estudiante: ")
                calificacion = float(input("Ingrese la calificación: "))
                estudiantes[nombre] = calificacion
                print(f"Estudiante {nombre} agregado con calificación {calificacion}")
            
            elif opcion == 2:
                nombre = input("Ingrese el nombre del estudiante: ")
                if nombre in estudiantes:
                    calificacion = float(input("Ingrese la nueva calificación: "))
                    estudiantes[nombre] = calificacion
                    print(f"Estudiante actualizado: {nombre} - {calificacion}")
                else:
                    print("Error: Estudiante no encontrado")
            
            elif opcion == 3:
                nombre = input("Ingrese el nombre del estudiante a eliminar: ")
                if nombre in estudiantes:
                    del estudiantes[nombre]
                    print(f"Estudiante {nombre} eliminado")
                else:
                    print("Error: Estudiante no encontrado")
            
            elif opcion == 4:
                print("\nListado de estudiantes:")
                for nombre, calificacion in estudiantes.items():
                    print(f"{nombre}: {calificacion}")
            
            elif opcion == 5:
                print("Saliendo del sistema...")
                break
            
            else:
                print("Opción no válida")
        
        except ValueError:
            print("Error: Ingrese una opción numérica válida")

registro_estudiantes()
```

```python
# Ejercicio 9
import random
import string

def generar_contrasena():
    try:
        longitud = int(input("Ingrese la longitud deseada para la contraseña: "))
        if longitud <= 0:
            raise ValueError("La longitud debe ser positiva")
        
        caracteres = string.ascii_letters + string.digits + string.punctuation
        contrasena = ''.join(random.choice(caracteres) for _ in range(longitud))
        print(f"Contraseña generada: {contrasena}")
    
    except ValueError as e:
        print(f"Error: {e}")

generar_contrasena()
```

```python
# Ejercicio 10
import os

def buscar_archivos(directorio, subcadena):
    try:
        if not os.path.isdir(directorio):
            raise FileNotFoundError(f"El directorio '{directorio}' no existe.")

        archivos_encontrados = []

        for archivo in os.listdir(directorio):
            ruta_completa = os.path.join(directorio, archivo)

            if os.path.isfile(ruta_completa) and subcadena.lower() in archivo.lower():
                archivos_encontrados.append(archivo)

        if archivos_encontrados:
            print("📂 Archivos encontrados:")
            for nombre in archivos_encontrados:
                print(f"- {nombre}")
        else:
            print("🔍 No se encontraron archivos que contengan esa subcadena.")

    except PermissionError:
        print("Error: Permisos insuficientes para acceder al directorio.")
    except FileNotFoundError as e:
        print(f"Error: {e}")
    except Exception as e:
        print(f"Error inesperado: {e}")

buscar_archivos("/Users/Yennifer Medrano/Downloads/Aprender_python/Clase", "Reporte")
```
