1. Pregunte al usuario cuántos elementos desea ingresar en una lista, luego solicite cada uno de ellos y presente el contenido de la lista y su contenido invertido.

```python
# Pedir al usuario la cantidad de elementos
cantidad_elementos = int(input("¿Cuántos elementos deseas ingresar en la lista? "))

# Inicializar la lista vacía
mi_lista = []

# Pedir al usuario que ingrese cada elemento
for i in range(cantidad_elementos):
    elemento = input(f"Ingresa el elemento {i + 1}: ")
    mi_lista.append(elemento)

# Mostrar la lista original
print("Lista original:", mi_lista)

# Mostrar la lista invertida
lista_invertida = mi_lista[::-1]  # Usamos slicing para invertir la lista
print("Lista invertida:", lista_invertida) 
```

2. Crear una lista de números y solicitar un número. Borrar de la lista ese número en todas las posiciones donde se encuentre. Presentar la lista final

```python
# Crear una lista de números
mi_lista = [1, 2, 3, 4, 2, 5, 2, 6, 7, 2, 8]

# Solicitar un número al usuario
numero_a_borrar = int(input("Ingresa el número que deseas borrar de la lista: "))

# Borrar el número de la lista en todas sus posiciones
mi_lista = [numero for numero in mi_lista if numero != numero_a_borrar]

# Presentar la lista final
print("Lista final:", mi_lista)
```

3. Solicite al usuario dos(2) frases y devuelva una lista con todas las letras que se repiten en la misma posición de ambas listas ejemplo: frase1: "holasena" frase2: "todogana" salida: ["o","n", "a"]

```python
def letras_repetidas_misma_posicion(frase1, frase2):

    letras_repetidas = []
    longitud_minima = min(len(frase1), len(frase2))

    for i in range(longitud_minima):
        if frase1[i] == frase2[i] and frase1[i].isalpha():
            letras_repetidas.append(frase1[i])

    return letras_repetidas

# Pedir al usuario que ingrese las dos frases
frase1 = input("Ingresa la primera frase: ")
frase2 = input("Ingresa la segunda frase: ")

# Encontrar las letras repetidas
resultado = letras_repetidas_misma_posicion(frase1, frase2)

# Mostrar el resultado
print("Letras repetidas en la misma posición:", resultado)
```

1. Procesamiento de Números del 1 al 20 Se requiere desarrollar un programa que trabaje con los números comprendidos entre el 1 y el 20, de forma que se identifiquen y muestren aquellos que sean divisibles por dos, acompañados de su posición en la secuencia. Además, el sistema debe recorrer la colección para determinar cuántos números no cumplen esa propiedad y calcular la suma de los que sí la cumplen. El objetivo es diseñar la lógica necesaria para examinar y procesar los números según las condiciones indicadas.

```python
numeros = list(range(1, 21))  # Crea una lista con los números del 1 al 20
divisibles_por_dos = []
no_divisibles_por_dos = 0
suma_divisibles_por_dos = 0

for indice, numero in enumerate(numeros):
    if numero % 2 == 0:
        divisibles_por_dos.append((numero, indice + 1))  # Añade el número y su posición a la lista
        suma_divisibles_por_dos += numero
    else:
        no_divisibles_por_dos += 1

print("Números divisibles por dos y su posición:")
for numero, posicion in divisibles_por_dos:
    print(f"Número: {numero}, Posición: {posicion}")

print(f"\nCantidad de números no divisibles por dos: {no_divisibles_por_dos}")
print(f"Suma de los números divisibles por dos: {suma_divisibles_por_dos}")
```

2. Administración de Actividades Diarias Imagina una aplicación para gestionar las actividades cotidianas de una persona, donde se disponga de una serie de tareas, algunas de las cuales indican que ya han sido concluidas. El desafío consiste en presentar cada actividad junto a su posición en la colección, filtrar aquellas que contengan una marca de finalización y actualizar el conjunto de actividades. Finalmente, se debe exhibir el estado del conjunto antes y después de aplicar el filtro, permitiendo al usuario observar los cambios realizados.

```python
tareas = [
    "Hacer la compra",
    "Estudiar para el examen",
    "Llamar al médico (Completado)",
    "Pagar las facturas",
    "Hacer ejercicio (Completado)",
    "Leer un libro"
]

print("Lista de tareas inicial:")
for indice, tarea in enumerate(tareas):
    print(f"Posición: {indice + 1}, Tarea: {tarea}")

tareas_completadas = []
tareas_pendientes = []

for tarea in tareas:
    if "(Completado)" in tarea:
        tareas_completadas.append(tarea.replace(" (Completado)", ""))  # Elimina la marca de completado
    else:
        tareas_pendientes.append(tarea)

print("\nLista de tareas completadas:")
for tarea in tareas_completadas:
    print(tarea)

print("\nLista de tareas pendientes:")
for tarea in tareas_pendientes:
    print(tarea)

tareas_actualizadas = tareas_pendientes  # Actualiza la lista original con las tareas pendientes

print("\nLista de tareas actualizada:")
for indice, tarea in enumerate(tareas_actualizadas):
    print(f"Posición: {indice + 1}, Tarea: {tarea}")
```

3. Filtrado y Transformación de Texto Propón un sistema en el que se solicite al usuario una secuencia de caracteres. A partir de la entrada, se debe descomponer el texto en sus componentes individuales y extraer únicamente aquellos que correspondan a letras del alfabeto, omitiendo números y símbolos. Una vez realizado este filtrado, la colección resultante deberá ser convertida en una estructura que no permita modificaciones, y el resultado final se debe mostrar al usuario.

```python
texto_usuario = input("Introduce una secuencia de caracteres: ")

letras = [caracter for caracter in texto_usuario if caracter.isalpha()]

tupla_letras = tuple(letras)

print("Letras filtradas y convertidas a tupla:", tupla_letras)
```

4. Informe Interactivo de Alumnos Considera dos conjuntos de datos paralelos: uno con nombres de estudiantes y otro con sus respectivas calificaciones. El reto consiste en elaborar un informe que relacione cada alumno con su nota, presentando también la posición de cada registro. Posteriormente, el sistema debe permitir al usuario consultar la nota de un alumno específico introduciendo la posición correspondiente, finalizando la consulta si se ingresa un valor fuera del rango válido. La solución deberá vincular ambas colecciones de manera sincronizada y ofrecer una interacción dinámica.

```python
nombres = ["Ana", "Carlos", "Sofía", "Pedro", "Laura"]
calificaciones = [90, 85, 95, 78, 92]

# Informe inicial
print("Informe de alumnos:")
for indice, nombre in enumerate(nombres):
    print(f"Posición: {indice + 1}, Alumno: {nombre}, Calificación: {calificaciones[indice]}")

# Consulta de notas
while True:
    try:
        posicion = int(input("\nIntroduce la posición del alumno (0 para salir): "))
        if posicion == 0:
            break
        if 1 <= posicion <= len(nombres):
            print(f"Alumno: {nombres[posicion - 1]}, Calificación: {calificaciones[posicion - 1]}")
        else:
            print("Posición fuera de rango.")
    except ValueError:
        print("Entrada inválida. Introduce un número entero.")
```

5. Aplicación de Operaciones sobre Valores Numéricos Diseña una aplicación que inicie con una serie de valores numéricos predefinidos. El usuario interactuará mediante un menú que le permita realizar distintas operaciones: calcular la suma total de los valores, reordenarlos según un criterio específico o eliminar uno de ellos según una indicación. La interfaz deberá ofrecer la posibilidad de repetir estas acciones hasta que el usuario decida finalizar. El desafío es organizar el flujo de la interacción y manipular la colección de números de acuerdo a la opción elegida.

```python
valores_iniciales = [10, 5, 20, 15, 30]
valores = valores_iniciales[:]  # Copia la lista original para no modificarla directamente

def mostrar_menu():
    print("\nMenú de operaciones:")
    print("1. Calcular suma total")
    print("2. Reordenar valores")
    print("3. Eliminar valor")
    print("4. Mostrar lista actual")
    print("5. Salir")

def calcular_suma():
    print("Suma total:", sum(valores))

def reordenar_valores():
    criterio = input("Introduce 'ascendente' o 'descendente': ").lower()
    if criterio == "ascendente":
        valores.sort()
    elif criterio == "descendente":
        valores.sort(reverse=True)
    else:
        print("Criterio inválido.")

def eliminar_valor():
    try:
        valor_eliminar = int(input("Introduce el valor a eliminar: "))
        valores.remove(valor_eliminar)
    except ValueError:
        print("Valor no encontrado en la lista.")

def mostrar_lista():
    print("Lista actual:", valores)

while True:
    mostrar_menu()
    opcion = input("Elige una opción: ")

    if opcion == "1":
        calcular_suma()
    elif opcion == "2":
        reordenar_valores()
    elif opcion == "3":
        eliminar_valor()
    elif opcion == "4":
        mostrar_lista()
    elif opcion == "5":
        print("¡Hasta luego!")
        break
    else:
        print("Opción inválida.")
```
