##1
Se dispone de una lista de códigos de productos que puede contener elementos duplicados. El reto consiste en convertir la lista en una colección que elimine los duplicados, agregar un nuevo producto al conjunto y luego descartar un producto específico. Posteriormente, se compararán dos colecciones (por ejemplo, inventarios de dos almacenes) para identificar los productos comunes, aquellos exclusivos de cada inventario y los que no se repiten entre ambos.

```python
inventario_inicial = {101, 102, 103, 104, 105}
print(inventario_inicial)

inventario_inicial.add(110)
print("Inventario con elemento agregado 110:", inventario_inicial)

inventario_inicial.discard(103)
print("Inventario con elemento descartado 103:", inventario_inicial)

inventario_almacen_A = {101, 102, 104, 110}
inventario_almacen_B = {102, 104, 107, 110}

print(f"Diferencia simétrica entre A y B: {inventario_almacen_A.symmetric_difference(inventario_almacen_B)}")
print(f"Diferencia de A - B: {inventario_almacen_A.difference(inventario_almacen_B)}")
print(f"Diferencia de B - A: {inventario_almacen_B.difference(inventario_almacen_A)}")
```

##2
Dos grupos de amigos tienen colecciones con sus géneros musicales favoritos. Se requiere determinar si ambos grupos comparten al menos un género (evaluar si las colecciones son disjuntas o no). Además, se debe presentar la intersección de géneros, la unión de ambos conjuntos y la diferencia simétrica (los géneros que no coinciden entre los grupos).

```python
GrupoA = {'rock', 'pop', 'jazz'}
print(f"Preferencias Grupo A: {(GrupoA)}")
GrupoB = {'pop', 'reggae', 'electrónica'}
print(f"Preferencias Grupo A: {(GrupoB)}")


print(f"GrupoA insertion GrupoB: {GrupoB.intersection(GrupoB)}")

print (f"GrupoA disjoint GrupoB= {GrupoA.isdisjoint(GrupoB)}")
print(f"GrupoA union GrupoB: {GrupoA.union(GrupoB)}")
print(f"GrupoA difference GrupoB:  {GrupoA.difference(GrupoB)}")
print(f"GrupoA difference GrupoB:  {GrupoA.difference(GrupoB)}")
```

##3
Desarrolla un traductor sencillo que utilice un diccionario donde las claves sean palabras en español y los valores sus traducciones en inglés. El usuario ingresará una palabra en español; si la palabra se encuentra en el diccionario, se mostrará su traducción. En caso contrario, se le solicitará al usuario que proporcione la traducción, la cual se agregará al diccionario. El proceso se repetirá hasta que el usuario decida finalizar la consulta.

```python

diccionario = { "hola": "hello", "adiós": "goodbye", "gracias": "thank you",  "por favor": "please"}


while True:
    palabra = input("Ingresa una palabra en español (o escribe 'salir' para terminar): ").lower()

    if palabra == "salir":
        print("¡Hasta luego!")
        break

    if palabra in diccionario:
        print(f"La traducción al inglés de '{palabra}' es: {diccionario[palabra]}")
    else:
        traduccion = input(f"No conozco la traducción de '{palabra}'. ¿Cómo se dice en inglés?: ").lower()
        diccionario[palabra] = traduccion
        print(f"Gracias, he guardado '{palabra}' como '{traduccion}'.")

print("\nDiccionario final:")
for esp, eng in diccionario.items():
    print(f"{esp} -> {eng}")
```

##4
Crea una aplicación que administre el inventario de una tienda usando un diccionario, donde cada clave representa el nombre de un producto y cada valor la cantidad en stock. La aplicación debe permitir consultar la cantidad de un producto, actualizar el stock de uno existente, agregar nuevos productos y eliminar aquellos que ya no estén disponibles. Al final se mostrará el inventario actualizado.

```python
inventario ={"manzanas": 50, "naranjas": 30, "peras": 20}


print(f"Inventario inicial:{inventario}")

while True:
  accion = input("¿Qué quieres hacer? (consultar, actualizar, agregar, eliminar, salir): ")

  if accion == "consultar":
    producto = input("¿Qué producto deseas consultar? ")
    if producto in inventario:
      print(f"Hay {inventario[producto]} {producto} en stock.")


  elif accion == "actualizar":
    producto = input("¿Qué producto deseas actualizar? ")
    if producto in inventario:
      cantidad = int(input(f"Nueva cantidad de {producto}: "))
      inventario[producto] = cantidad
      print(f"Stock de {producto} actualizado a {cantidad}.")

  elif accion == "agregar":
    producto = input("Nombre del nuevo producto: ")
    cantidad = int(input(f"Cantidad de {producto}: "))
    inventario[producto] = cantidad
    print(f"Producto {producto} agregado con {cantidad} unidades.")

  elif accion == "eliminar":
    producto = input("¿Qué producto deseas eliminar? ")
    if producto in inventario:
      del inventario[producto]
      print(f"Producto {producto} eliminado del inventario.")


  elif accion == "salir":
    break

print("\nInventario final:")
print(inventario)
```
