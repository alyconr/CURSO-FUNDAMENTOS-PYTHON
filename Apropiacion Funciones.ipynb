
1. Realice un programa traductor de frutas del español al inglés. Para ello, solicite el nombre de una fruta al usuario y muestre su traducción al inglés. Si la fruta no se encuentra en el diccionario, permítale al usuario agregarla con su respectiva traducción. El programa solo termina cuando el usuario ingresa una fruta que no está en el diccionario y no desea agregarla.

```python
frutas = {'manzana':'apple', 'naranja':'orange', 'platano':'banano', 'pera':'pear', 'coco':'coconut', 'limon':'lemon'}

while True:
    traduccionIngles = input('Ingrese el nombre de la fruta en español: ')
    if traduccionIngles in frutas:
        print(f'La traducción de {traduccionIngles} es: {frutas[traduccionIngles]}')
        break
    elif traduccionIngles not in frutas:
        print(f'La fruta {traduccionIngles} no se encuentra')
        añadir_res = input(f'Desea agregar la traducción de {traduccionIngles}?(sí/no): ')
        if añadir_res == 'sí':
            nueva_fruta = input(f'¿Cuál es la traducción de {traduccionIngles}?: ')
            frutas[traduccionIngles] = nueva_fruta
        elif añadir_res == 'no':
            print('Volviendo al inicio...')
        else:
            print('Escoja una opción válida')
    else:
        print("La fruta no se encuentra en el diccionario.")
        fruta = input("Ingrese el nombre de la fruta en español: ")
        if fruta == "salir":
            break

print("¡Hasta luego!")

```

1. Realiza una función que tome una lista de números enteros y devuelva dos listas ordenadas. La primera con los números pares y la segunda con los números impares

```python
def separar_pares_impares(lista):
  pares = []
  impares=[]
  for numero in lista:
    if numero % 2 == 0:
      pares.append(numero)
    else:
      impares.append(numero)
  pares.sort()
  impares.sort()
  return pares,impares
lista = [2,90,5,7,96,3,45,62,5648,13,156,4,84,56,31,564,86,456,10]
pares, impares = separar_pares_impares(lista)
print("pares",pares)
print("impares",impares)
```

2. Realiza una función que reciba una palabra y devuelva un diccionario donde las claves sean las diferentes letras que se encuentran en la palabra y los valores sean el número de veces que se repite cada letra en la palabra
palabra = "colombia"
{"c":1, "o":2, "l":1, "m":1, "b":1, "i":1, "a":1}

```python
lista = input("ingresa una palabra")
def contar(lista):
  clave = {}
  for letra in lista:
    if letra in clave:
      clave[letra] += 1
    else:
      clave[letra] = 1
  return clave
resultado = contar(lista)
print("palabra",resultado)
```

3. Realiza una función que reciba un número y devuelva una cadena con los nombres de los números recibidos, separando cada número con un guión medio. Por ejemplo, si el número recibido es 134, la función devolverá la cadena "uno - tres - cuatro"

```python
numero = int(input("ingresa un numero"))
claves_numeros = {0:"cero",1:"uno",2:"dos",3:"tres",4:"cuatro",5:"cinco",6:"seis",7:"siete",8:"ocho",9:"nueve"}



def clave_numero(numero):
  resultado = []
  for x in str(numero):
    if int(x) in claves_numeros:
      resultado.append(claves_numeros[int(x)])
    else:
      print("no es un numero")

  return "-_-".join(resultado) #me funciona para colocar a todos de una vez viva =iv
resultado = clave_numero(numero)
if resultado:
  print("Resultado:", resultado)
```
