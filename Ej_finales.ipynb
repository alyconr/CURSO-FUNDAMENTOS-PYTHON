```python
#1
def calculadora():
    try:
        num1 = int(input("Ingrese el primer número: "))
        
        num2 = int(input("Ingrese el segundo número: "))
        
        operador = input("Ingrese el operador (suma(+),resta(-),mulltiplicacion(*),division(/),exponente(**),residuo(//)): ")

        if operador == '+':
            resultado = num1 + num2
        elif operador == '-':
            resultado = num1 - num2
        elif operador == '*':
            resultado = num1 * num2
        elif operador == '/':
            if num2 == 0:
                raise ZeroDivisionError("No se puede dividir entre cero.")
            
            resultado = num1 / num2
        
        
        elif operador == '**':
            resultado = num1 ** num2
        elif operador == '//':
            if num2 == 0:
                raise ZeroDivisionError("No se puede dividir entre cero.")
            resultado = num1 // num2
        else:
            print("Operador no válido.")
            return

        print(f"Resultado: {num1} {operador} {num2} = {resultado}")

    except ValueError:
        print("Error: debe ingresar valores numéricos.")
    except ZeroDivisionError as e:
        print(f"Error: {e}")


calculadora()
```

```python
#2
inventario = {
    'manzanas': 50,
    'naranjas': 30,
    'peras': 20
}

while True:
    print("\n--- Menú de Inventario ---")
    print("1. Mostrar inventario")
    print("2. Agregar producto")
    print("3. Actualizar producto")
    print("4. Eliminar producto")
    print("5. Salir")

    opcion = input("Seleccione una opción (1-5): ")

    if opcion == '1':
        print("\nInventario actual:")
        for producto, cantidad in inventario.items():
            print(f"- {producto}: {cantidad}")

    elif opcion == '2':
        nombre = input("Nombre del producto a agregar: ").lower()
        if nombre in inventario:
            print("Ese producto ya existe.")
        else:
            try:
                cantidad = int(input("Cantidad: "))
                inventario[nombre] = cantidad
                print(f"Producto '{nombre}' agregado con {cantidad} unidades.")
            except ValueError:
                print("Cantidad inválida. Debe ser un número entero.")

    elif opcion == '3':
        nombre = input("Nombre del producto a actualizar: ").lower()
        if nombre in inventario:
            try:
                cantidad = int(input("Nueva cantidad: "))
                inventario[nombre] = cantidad
                print(f"Producto '{nombre}' actualizado a {cantidad} unidades.")
            except ValueError:
                print("Cantidad inválida.")
        else:
            print(f"El producto '{nombre}' no existe.")

    elif opcion == '4':
        nombre = input("Nombre del producto a eliminar: ").lower()
        if nombre in inventario:
            del inventario[nombre]
            print(f"Producto '{nombre}' eliminado del inventario.")
        else:
            print(f"El producto '{nombre}' no existe.")

    elif opcion == '5':
        print("Saliendo del programa. ¡Hasta luego!")
        break

    else:
        print("Opción inválida. Intente de nuevo.")

```

```python
#3
# Solicitar texto al usuario
texto = input("Ingrese un texto: ")

# Verificar si el texto no está vacío
if texto.strip() == "":
    print("Error: no ingresaste ningún texto.")
else:
    # Convertir a minúsculas
    texto = texto.lower()

    # Eliminar signos de puntuación básicos
    for caracter in ",.;:¡!¿?\"'()[]{}":
        texto = texto.replace(caracter, "")

    # Separar palabras
    palabras = texto.split()

    # Crear diccionario de frecuencia
    frecuencia = {}

    for palabra in palabras:
        if palabra in frecuencia:
            frecuencia[palabra] += 1
        else:
            frecuencia[palabra] = 1

    # Mostrar resultados
    print("\nFrecuencia de palabras:")
    for palabra, cantidad in frecuencia.items():
        print(f"{palabra}: {cantidad}")
```

```python
#4
print("Seleccione la conversión:")
print("1. Centímetros a pulgadas")
print("2. Kilómetros a millas")

opcion = input("Opción (1 o 2): ")

if opcion == "1":
    try:
        cm = int(input("Ingrese la cantidad en centímetros: "))
        pulgadas = cm / 2.54
        print(f"Resultado: {cm} cm = {pulgadas:.2f} pulgadas")
    except ValueError:
        print("Error: debe ingresar un número válido.")

elif opcion == "2":
    try:
        km = int(input("Ingrese la cantidad en kilómetros: "))
        millas = km * 0.621371
        print(f"Resultado: {km} km = {millas:.2f} millas")
    except ValueError:
        print("Error: debe ingresar un número válido.")

else:
    print("Opción inválida. Por favor seleccione 1 o 2.")
```

```python
#6
def calcular_factorial():
    try:

        num = int(input("Ingrese un número entero para calcular su factorial: "))


        if num < 0:
            print("Error: El número no puede ser negativo.")
        else:

            factorial = 1
            for i in range(1, num + 1):
                factorial *= i


            print(f"El factorial de {num} es {factorial}")

    except ValueError:
        print("Error: Debes ingresar un número entero válido.")


calcular_factorial()
```

```python
#7
from datetime import datetime

def calcular_dias_faltantes():
    try:

        fecha_str = input("Ingrese una fecha (DD/MM/AAAA): ")


        fecha_objeto = datetime.strptime(fecha_str, "%d/%m/%Y")


        hoy = datetime.today()


        diferencia = fecha_objeto - hoy


        if diferencia.days > 0:
            print(f"Faltan {diferencia.days} días para el {fecha_objeto.strftime('%d/%m/%Y')}.")
        elif diferencia.days == 0:
            print("¡Hoy es el día!")
        else:
            print(f"¡Ya ha pasado la fecha {fecha_objeto.strftime('%d/%m/%Y')}!")

    except ValueError:
        print("Error: El formato de la fecha no es válido. Debe ser DD/MM/AAAA.")


calcular_dias_faltantes()
```

```python
#8
registro_estudiantes = {
    'Ana': 90,
    'Luis': 78,
    'Carlos': 85
}

def agregar_estudiante():
    nombre = input("Ingrese el nombre del estudiante: ")
    try:
        calificacion = int(input(f"Ingrese la calificación de {nombre}: "))
        registro_estudiantes[nombre] = calificacion
        print(f"Estudiante {nombre} agregado con calificación {calificacion}.")
    except ValueError:
        print("Error: La calificación debe ser un número válido.")

def actualizar_calificacion():
    nombre = input("Ingrese el nombre del estudiante: ")
    if nombre in registro_estudiantes:
        try:
            calificacion = int(input(f"Ingrese la nueva calificación de {nombre}: "))
            registro_estudiantes[nombre] = calificacion
            print(f"Estudiante actualizado: {nombre} - {calificacion}")
        except ValueError:
            print("Error: La calificación debe ser un número válido.")
    else:
        print("Error: El estudiante no existe.")

def eliminar_estudiante():
    nombre = input("Ingrese el nombre del estudiante a eliminar: ")
    if nombre in registro_estudiantes:
        del registro_estudiantes[nombre]
        print(f"Estudiante {nombre} eliminado.")
    else:
        print("Error: El estudiante no existe.")

def listar_estudiantes():
    if registro_estudiantes:
        print("\nRegistro de estudiantes:")
        for nombre, calificacion in registro_estudiantes.items():
            print(f"{nombre}: {calificacion}")
    else:
        print("No hay estudiantes registrados.")

def menu():
    while True:
        print("\nMenú:")
        print("1. Agregar estudiante")
        print("2. Actualizar calificación")
        print("3. Eliminar estudiante")
        print("4. Listar estudiantes")
        print("5. Salir")

        opcion = input("Opción: ")

        if opcion == "1":
            agregar_estudiante()
        elif opcion == "2":
            actualizar_calificacion()
        elif opcion == "3":
            eliminar_estudiante()
        elif opcion == "4":
            listar_estudiantes()
        elif opcion == "5":
            print("Saliendo del programa...")
            break
        else:
            print("Error: Opción no válida. Por favor, elija una opción entre 1 y 5.")
menu()
```

```python
#9
import random
import string

def generar_contraseña():
    try:

        longitud = int(input("Ingrese la longitud deseada para la contraseña: "))


        if longitud <= 0:
            print("Error: La longitud debe ser un número entero positivo.")
            return

        # Crear una lista de caracteres posibles (mayúsculas, minúsculas, dígitos y caracteres especiales)
        caracteres_posibles = string.ascii_letters + string.digits + string.punctuation


        contraseña = ''.join(random.choice(caracteres_posibles) for _ in range(longitud))


        print(f"Contraseña generada: {contraseña}")

    except ValueError:
        print("Error: La longitud debe ser un número entero válido.")


generar_contraseña()
```

```python
#10
import os

def buscar_archivos():
    try:
        # Solicitar la ruta del directorio
        directorio = input("Ingrese la ruta del directorio: ")

        # Comprobar si el directorio existe
        if not os.path.isdir(directorio):
            print("Error: El directorio no existe o no es un directorio válido.")
            return

        # Solicitar la subcadena a buscar
        subcadena = input("Ingrese la subcadena a buscar: ")

        # Listar todos los archivos en el directorio
        archivos_encontrados = [archivo for archivo in os.listdir(directorio)
                                if os.path.isfile(os.path.join(directorio, archivo)) and subcadena in archivo]

        # Mostrar los resultados
        if archivos_encontrados:
            print(f"Archivos encontrados que contienen \"{subcadena}\":")
            for archivo in archivos_encontrados:
                print(archivo)
        else:
            print(f"No se encontraron archivos que contengan \"{subcadena}\" en el nombre.")

    except PermissionError:
        print("Error: Permisos insuficientes para acceder a este directorio.")
    except Exception as e:
        print(f"Error: {e}")

# Llamar a la función para buscar los archivos
buscar_archivos()
```
