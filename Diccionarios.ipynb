En una escuela se permite la inscripción de sus alumnos a las siguientes actividades deportivas: futbol (f), baloncesto (b). Realice un programa que permita ingresar el nombre de sus estudiantes e inscribir a cada uno en uno o los dos deportes.
La aplicación debe mostrar los siguientes listados:

Estudiantes inscritos en fulbol     
Estudiantes inscritos en baloncesto           
Estudiantes inscritos en ambos deportes        
Todos los estudiantes inscritos en algún deporte       
Estudiantes inscritos en solo un deporte.      
Genere un menú con las opciones de la aplicación

```python
futbol = []
baloncesto = []

def menu():
    print("\n Menú de Inscripción Deportiva ")
    print("1. Inscribir estudiante")
    print("2. Mostrar estudiantes inscritos en fútbol")
    print("3. Mostrar estudiantes inscritos en baloncesto")
    print("4. Mostrar estudiantes inscritos en ambos deportes")
    print("5. Mostrar todos los estudiantes inscritos en algún deporte")
    print("6. Mostrar estudiantes inscritos en solo un deporte")
    print("7. Salir")

def inscribir_estudiante():
    nombre = input("Ingrese el nombre del estudiante: ").strip()
    print("¿En qué deporte desea inscribir al estudiante?")
    print("1. Fútbol")
    print("2. Baloncesto")
    print("3. Ambos")
    opcion = input("Seleccione una opción (1/2/3): ")
    if opcion == "1":
        futbol.append(nombre)
    elif opcion == "2":
        baloncesto.append(nombre)
    elif opcion == "3":
        futbol.append(nombre)
        baloncesto.append(nombre)
    else:
        print("Opción inválida")

def mostrar_lista(lista, titulo):
    print(f"\n{titulo}")
    if lista:
        for estudiante in lista:
            print(f"- {estudiante}")
    else:
        print("No hay estudiantes en esta lista.")

while True:
    menu()
    opcion = input("Seleccione una opción del menú: ")

    if opcion == "1":
        inscribir_estudiante()
    elif opcion == "2":
        mostrar_lista(futbol, "Estudiantes inscritos en fútbol:")
    elif opcion == "3":
        mostrar_lista(baloncesto, "Estudiantes inscritos en baloncesto:")
    elif opcion == "4":
        ambos = list(set(futbol) & set(baloncesto))
        mostrar_lista(ambos, "Estudiantes inscritos en ambos deportes:")
    elif opcion == "5":
        todos = list(set(futbol) | set(baloncesto))
        mostrar_lista(todos, "Todos los estudiantes inscritos en algún deporte:")
    elif opcion == "6":
        solo_uno = list((set(futbol) ^ set(baloncesto)))
        mostrar_lista(solo_uno, "Estudiantes inscritos en solo un deporte:")
    elif opcion == "7":
        print("Saliendo del programa.")
        break
    else:
        print("Opción inválida. Intente nuevamente.")
```

Realice un programa traductor de frutas del español al inglés. Para ello, solicite el nombre de una fruta al usuario y muestre su traducción al inglés. Si la fruta no se encuentra en el diccionario, permítale al usuario agregarla con su respectiva traducción. El programa solo termina cuando el usuario ingresa una fruta que no está en el diccionario y no desea agregarla.

```python
frutas = {"manzana": "apple", "banana": "banana", "naranja": "orange", "uva": "uva"}

while True:
    fruta_español = input("Ingrese el nombre de una fruta en español: ").lower()
    if fruta_español in frutas:
        print(f"La traducción de {fruta_español} al inglés es: {frutas[fruta_español]}")
        break
    elif fruta_español == "salir":
        print("Saliendo del programa.")
        break

    else:
        traduccion = input(f"La fruta {fruta_español} no se encuentra en el diccionario. Ingrese su traducción al inglés: ").lower()
        frutas[fruta_español] = traduccion
        print(f"La fruta {fruta_español} ha sido agregada al diccionario.")
        print(frutas)
        print(f"La traducción de {fruta_español} al inglés es: {frutas[fruta_español]}")
        break
```

```python
def factorial_recursivo(n):
    if n == 1:
        return 1
    else:
        return n * factorial_recursivo(n-1)

num = int(input("Ingrese un numero: "))
print(f"El factorial de {num} es {factorial_recursivo(num)}")
```
