2. En una escuela se permite la inscripción de sus alumnos a las siguientes actividades deportivas:
futbol (f), baloncesto (b). Realice un programa que permita ingresar el nombre de sus estudiantes e inscribir a cada uno en uno o los dos deportes.
La aplicación debe mostrar los siguientes listados:

Estudiantes inscritos en fulbol
Estudiantes inscritos en baloncesto
Estudiantes inscritos en ambos deportes
Todos los estudiantes inscritos en algún deporte
Estudiantes inscritos en solo un deporte.
Genere un menú con las opciones de la aplicación

```python
futbol = []
baloncesto = []

while True:

    print("1. Inscribir estudiante")
    print("2. Mostrar estudiantes inscritos en fútbol")
    print("3. Mostrar estudiantes inscritos en baloncesto")
    print("4. Mostrar estudiantes inscritos en ambos deportes")
    print("5. Mostrar todos los estudiantes inscritos")
    print("6. Mostrar estudiantes inscritos en solo un deporte")
    print("7. Salir del programa")

    opcion = input("Ingrese una opción: ")

    if opcion == '1':
        nombre = input("Ingrese el nombre del estudiante: ")
        deportes = input("Ingrese los deportes (f para fútbol, b para baloncesto, o ambas): ")
        if 'f' == deportes:
            futbol.append(nombre)
        if 'b' == deportes:
            baloncesto.append(nombre)
    elif opcion == '2':
        print("Estudiantes inscritos en fútbol:", futbol)
    elif opcion == '3':
        print("Estudiantes inscritos en baloncesto:", baloncesto)
    elif opcion == '4':
        print("Estudiantes inscritos en ambos deportes:", futbol.intersection(baloncesto))
    elif opcion == '5':
        print("Todos los estudiantes inscritos:", futbol.union(baloncesto))
    elif opcion == '6':
        solo_futbol = futbol.difference(baloncesto)
        solo_baloncesto = baloncesto.difference(futbol)
        print("Estudiantes inscritos en solo un deporte:", solo_futbol.union(solo_baloncesto))
    elif opcion == '7':
        break
    else:
        print("La opción es inválida.")
```

# Funciones recursivas

Una función recursiva es aquella que en su interior se invoca a si misma

```python
def factorial_recursivo(n):
  if n == 1:
    return 1
  else:
    return n * factorial_recursivo(n-1) # 4 * 4-1= 3 => 4 * 3

num = int(input("Ingrese un número: "))
print(f"El factorial de {num} es {factorial_recursivo(num)}")
```

```python
def atras(num):
  print(num)
  num -= 1
  if num > 0:
     atras(num)

atras(10)
```

1, Realiza una función que tome una lista de números enteros y devuelva dos listas ordenadas. La primera con los números pares y la segunda con los números impares.

```python
def separar_pares_impares(lista):
    pares = sorted([num for num in lista if num % 2 == 0])
    impares = sorted([num for num in lista if num % 2 != 0])
    return pares, impares

numeros = [5, 2, 9, 4, 1, 6, 3, 8]
pares, impares = separar_pares_impares(numeros)
print("Pares:", pares)
print("Impares:", impares)
```

2. Realiza una función que reciba una palabra y devuelva un diccionario donde las claves sean las diferentes letras que se encuentran en la palabra y los valores sean el número de veces que se repite cada letra en la palabra

```python
def contar_letras(palabra):
    contador = {}
    for letra in palabra:
        if letra in contador:
            contador[letra] += 1
        else:
            contador[letra] = 1
    return contador

palabra = "colombia"
resultado = contar_letras(palabra)
print(resultado)
```

3. Realiza una función que reciba un número y devuelva una cadena con los nombres de los números recibidos, separando cada número con un guión medio. Por ejemplo, si el número recibido es 134, la función devolverá la cadena "uno - tres - cuatro"


```python
def numero_a_palabras(numero):
    numeros_en_palabras = { '0': 'cero', '1': 'uno', '2': 'dos', '3': 'tres', '4': 'cuatro', '5': 'cinco', '6': 'seis', '7': 'siete', '8': 'ocho', '9': 'nueve'
    }
    palabras = [numeros_en_palabras[digito] for digito in str(numero)]
    return ' - '.join(palabras)


numero = 2007
resultado = numero_a_palabras(numero)
print(resultado)
```
