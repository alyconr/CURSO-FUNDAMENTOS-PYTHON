```python
def calculadora_simple(num1, num2, operador):

    if operador == '+':
        return num1 + num2
    elif operador == '-':
        return num1 - num2
    elif operador == '*':
        return num1 * num2
    elif operador == '/':
        if num2 == 0:
            return "Error: División por cero"
        return num1 / num2
    elif operador == '**':
        return num1 ** num2
    elif operador == '//':
        if num2 == 0:
            return "Error: División por cero"
        return num1 // num2
    else:
        return "Error: Operador no válido"
try:
    num1 = float(input("Ingrese el primer número: "))
    num2 = float(input("Ingrese el segundo número: "))
    operador = input("Ingrese el operador (+, -, *, /, **, //): ")

    resultado = calculadora_simple(num1, num2, operador)

    print(f"Resultado: {resultado}")

except ValueError:
    print("Error: Por favor, ingrese números válidos.")
```

```python
def mostrar_inventario(inventario):

    if not inventario:
        print("El inventario está vacío.")
    else:
        print("Inventario:")
        for producto, cantidad in inventario.items():
            print(f"- {producto}: {cantidad}")

def agregar_producto(inventario, nombre, cantidad):
    """Agrega un nuevo producto al inventario."""
    if nombre in inventario:
        print(f"Error: El producto '{nombre}' ya existe.")
    else:
        inventario[nombre] = cantidad
        print(f"Producto '{nombre}' agregado.")

def actualizar_stock(inventario, nombre, cantidad):
    """Actualiza la cantidad de un producto existente."""
    if nombre in inventario:
        inventario[nombre] = cantidad
        print(f"Stock de '{nombre}' actualizado.")
    else:
        print(f"Error: El producto '{nombre}' no existe.")

def eliminar_producto(inventario, nombre):
    """Elimina un producto del inventario."""
    if nombre in inventario:
        del inventario[nombre]
        print(f"Producto '{nombre}' eliminado.")
    else:
        print(f"Error: El producto '{nombre}' no existe.")

inventario = {'manzanas': 50, 'naranjas': 30, 'peras': 20}
print("Inventario inicial:")
mostrar_inventario(inventario)

actualizar_stock(inventario, 'peras', 25)
agregar_producto(inventario, 'bananas', 40)
eliminar_producto(inventario, 'naranjas')

print("\nInventario final:")
mostrar_inventario(inventario)
0
```

```python
import re

def analizar_texto(texto):
 try:
        if not texto:
            return "Error: No se ingresó ningún texto."

        texto = texto.lower()
        palabras = re.findall(r'\b\w+\b', texto)

        frecuencia = {}
        for palabra in palabras:
            if palabra in frecuencia:
                frecuencia[palabra] += 1
            else:
                frecuencia[palabra] = 1

        return frecuencia

 except Exception as e:
        return f"Error inesperado: {e}"

texto_ingresado = input("Ingrese un texto: ")
resultado = analizar_texto(texto_ingresado)

if isinstance(resultado, dict):
    print("Frecuencia de palabras:")
    for palabra, frecuencia in resultado.items():
        print(f"{palabra}: {frecuencia}")
else:
    print(resultado)
```

```python
def cm_a_pulgadas(cm):

    return cm * 0.393701

def km_a_millas(km):
    return km * 0.621371

def mostrar_menu():

    print("Seleccione la conversión:")
    print("1. Centímetros a pulgadas")
    print("2. Kilómetros a millas")

def obtener_opcion():

    while True:
        try:
            opcion = int(input("Opción: "))
            if 1 <= opcion <= 2:
                return opcion
            else:
                print("Opción inválida. Por favor, ingrese 1 o 2.")
        except ValueError:
            print("Entrada inválida. Por favor, ingrese un número.")

def obtener_cantidad(unidad):
    while True:
        try:
            cantidad = float(input(f"Ingrese la cantidad en {unidad}: "))
            return cantidad
        except ValueError:
            print("Entrada inválida. Por favor, ingrese un número.")

mostrar_menu()
opcion = obtener_opcion()

if opcion == 1:
    cantidad_cm = obtener_cantidad("centímetros")
    pulgadas = cm_a_pulgadas(cantidad_cm)
    print(f"Resultado: {cantidad_cm} cm = {pulgadas:.2f} pulgadas")
elif opcion == 2:
    cantidad_km = obtener_cantidad("kilómetros")
    millas = km_a_millas(cantidad_km)
    print(f"Resultado: {cantidad_km} km = {millas:.2f} millas")
```

```python
import random

def adivina_el_numero():
    numero_secreto = random.randint(1, 50)
    intento = 1

    print('¡Bienvenido al juego "Adivina el Número"!')
    print('Estoy pensando en un número entre 1 y 50.')

    while True:
        try:
            entrada = input(f"Intento {intento} - Ingresa tu número: ")
            numero_usuario = int(entrada)

            if numero_usuario < 1 or numero_usuario > 50:
                print("Por favor, ingresa un número dentro del rango de 1 a 50.")
                continue

            if numero_usuario < numero_secreto:
                print("El número secreto es mayor.")
            elif numero_usuario > numero_secreto:
                print("El número secreto es menor.")
            else:
                print(f"¡Felicidades! Has adivinado el número secreto: {numero_secreto}")
                break

            intento += 1

        except ValueError:
            print("Entrada inválida. Por favor, ingresa un número entero.")

adivina_el_numero()
```

```python
def calcular_factorial():
    try:
        numero = int(input("Ingrese un número entero para calcular su factorial: "))

        if numero < 0:
            print("El número debe ser un entero no negativo.")
            return

        factorial = 1
        for i in range(1, numero + 1):
            factorial *= i

        print(f"El factorial de {numero} es {factorial}")

    except ValueError:
        print("Entrada inválida. Por favor, ingrese un número entero.")

calcular_factorial()
```

```python
from datetime import datetime

def calcular_dias_para_fecha():
    try:
        entrada = input("Ingrese una fecha (DD/MM/AAAA): ")
        fecha_objetivo = datetime.strptime(entrada, "%d/%m/%Y").date()
        fecha_actual = datetime.today().date()

        diferencia = (fecha_objetivo - fecha_actual).days

        if diferencia > 0:
            print(f"Faltan {diferencia} días para el {fecha_objetivo.strftime('%d/%m/%Y')}.")
        elif diferencia == 0:
            print(f"¡Hoy es el día {fecha_objetivo.strftime('%d/%m/%Y')}!")
        else:
            print(f"La fecha {fecha_objetivo.strftime('%d/%m/%Y')} ya pasó hace {-diferencia} días.")

    except ValueError:
        print("Formato de fecha inválido. Por favor, use el formato DD/MM/AAAA.")

calcular_dias_para_fecha()
```

```python
def mostrar_menu():
    print("\nMenú:")
    print("1. Agregar estudiante")
    print("2. Actualizar calificación")
    print("3. Eliminar estudiante")
    print("4. Listar estudiantes")
    print("5. Salir")

def agregar_estudiante(registro):
    nombre = input("Ingrese el nombre del estudiante: ").strip()
    if nombre in registro:
        print("Este estudiante ya existe.")
    else:
        try:
            calificacion = int(input("Ingrese la calificación: "))
            registro[nombre] = calificacion
            print(f"Estudiante agregado: {nombre} - {calificacion}")
        except ValueError:
            print("Calificación inválida. Debe ser un número entero.")

def actualizar_calificacion(registro):
    nombre = input("Ingrese el nombre del estudiante: ").strip()
    if nombre in registro:
        try:
            calificacion = int(input("Ingrese la nueva calificación: "))
            registro[nombre] = calificacion
            print(f"Estudiante actualizado: {nombre} - {calificacion}")
        except ValueError:
            print("Calificación inválida. Debe ser un número entero.")
    else:
        print("El estudiante no existe.")

def eliminar_estudiante(registro):
    nombre = input("Ingrese el nombre del estudiante a eliminar: ").strip()
    if nombre in registro:
        del registro[nombre]
        print(f"Estudiante eliminado: {nombre}")
    else:
        print("El estudiante no existe.")

def listar_estudiantes(registro):
    if not registro:
        print("No hay estudiantes registrados.")
    else:
        print("Registro de estudiantes:")
        for nombre, calificacion in registro.items():
            print(f"{nombre}: {calificacion}")

def sistema_registro():
    registro = {'Ana': 90, 'Luis': 78, 'Carlos': 85}
    print("Registro de estudiantes inicial:", registro)

    while True:
        mostrar_menu()
        opcion = input("Opción: ").strip()

        if opcion == "1":
            agregar_estudiante(registro)
        elif opcion == "2":
            actualizar_calificacion(registro)
        elif opcion == "3":
            eliminar_estudiante(registro)
        elif opcion == "4":
            listar_estudiantes(registro)
        elif opcion == "5":
            print("Saliendo... Registro final:", registro)
            break
        else:
            print("Opción inválida. Por favor, elige una opción del 1 al 5.")
sistema_registro()
```

```python
import random
import string

def generar_contraseña():
    try:
        longitud = int(input("Ingrese la longitud deseada para la contraseña: "))

        if longitud <= 0:
            print("La longitud debe ser un número entero positivo.")
            return

        caracteres = string.ascii_letters + string.digits + string.punctuation
        contraseña = ''.join(random.choice(caracteres) for _ in range(longitud))

        print("Contraseña generada:", contraseña)

    except ValueError:
        print("Entrada inválida. Por favor, ingrese un número entero.")

generar_contraseña()
```

```python
import os

def buscar_archivos(directorio, subcadena):
    try:
        archivos = os.listdir(directorio)
        encontrados = [archivo for archivo in archivos if subcadena in archivo]
        if encontrados:
            print(f'Archivos encontrados que contienen "{subcadena}":\n')
            for archivo in encontrados:
                print(archivo)
        else:
            print(f'No se encontraron archivos que contengan "{subcadena}".')
    except FileNotFoundError:
        print("El directorio no existe.")
    except PermissionError:
        print("No tiene permisos para acceder a este directorio.")
    except Exception as e:
        print(f"Ocurrió un error: {e}")

ruta = input("Ingrese la ruta del directorio: ")
cadena = input("Ingrese la subcadena a buscar: ")
buscar_archivos(ruta, cadena)
```
